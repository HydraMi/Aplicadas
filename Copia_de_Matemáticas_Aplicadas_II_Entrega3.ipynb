## (a) Ejemplo numérico donde el vecino más cercano en distancia no es el mismo que el vecino más cercano en ángulo

Ejemplo con $( n = 1 )$. Tomamos:

$$
z_1 = 1, \quad z_2 = -10, \quad x = -1.
$$

Calculamos las normas:

$$
\|x - z_1\| = | -1 - 1 | = 2
$$

$$
\|x - z_2\| = | -1 - (-10) | = | -1 + 10 | = 9
$$

Calculamos los ángulos:

Para $( \theta_1 = \angle(x, z_1) )$:

$$
\cos(\theta_1) = \frac{-1}{(1)(1)} = -1.
$$

Para $( \theta_2 = \angle(x, z_2) )$:

$$
\cos(\theta_2) = \frac{10}{(1)(10)} = 1.
$$

Usamos la función $( \arccos )$ para obtener el ángulo en radianes:

$$
\theta_1 = \arccos(-1) = \pi \text{ radianes } = 180^\circ.
$$

$$
\theta_2 = \arccos(1) = 0 \text{ radianes } = 0^\circ.
$$

$$
\angle(x, z_1) = \frac{\pi}{2}, \quad \angle(x, z_2) = 0.
$$

Por lo tanto, el vecino más cercano en distancia es $( z_1 )$, mientras que el vecino más cercano en ángulo es $( z_2 )$. En dimensiones más altas osea que los vectores tienen más de una coordenada.

## (b) Demostración cuando los vectores están normalizados

Si asumimos que los vectores están normalizados, es decir, que

$$
\|z_i\| = 1, \quad \text{para todo } i = 1, \dots, m,
$$

entonces tenemos la siguiente deducción:

$$
\|x - z_j\| \leq \|x - z_i\|
$$

es equivalente a:

$$
\|x - z_j\|^2 \leq \|x - z_i\|^2
$$

Expandemos los cuadrados:

$$
\|x\|^2 - 2x^T z_j + \|z_j\|^2 \leq \|x\|^2 - 2x^T z_i + \|z_i\|^2
$$

Cancelando $( \|x\|^2 )$ y usando que $( \|z_i\| = 1 )$:

$$
-2x^T z_j \leq -2x^T z_i
$$

$$
x^T z_j \geq x^T z_i
$$

Dividiendo por $( \|x\|\|z_j\| )$ y $( \|x\|\|z_i\| )$ (que son iguales porque $( \|z_j\| = \|z_i\| = 1 ))$:

$$
\frac{x^T z_j}{\|x\|\|z_j\|} \geq \frac{x^T z_i}{\|x\|\|z_i\|}
$$

Tomando la función arco coseno en ambos lados (recordando que $( \arccos )$ es decreciente):

$$
\arccos \left( \frac{x^T z_j}{\|x\|\|z_j\|} \right) \leq \arccos \left( \frac{x^T z_i}{\|x\|\|z_i\|} \right)
$$

utilizamos el hecho de que la función arcocoseno es decreciente, es decir, si $( a \geq b )$, entonces:

$$
\arccos(a) \leq \arccos(b).
$$
Finalmente, esto implica que:

$$
\angle(x, z_j) \leq \angle(x, z_i).
$$

##Por lo tanto, cuando los vectores están normalizados (magnitud igual a 1), el vecino más cercano en distancia y el vecino más cercano en ángulo siempre son el mismo vector.
