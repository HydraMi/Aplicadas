El ángulo $θ$ entre dos vectores $x$ y $y$ está dado por la ecuación:

$\cos\theta = \frac{x \cdot y} {\|{x}\| \|{y}\|}$

$x \cdot y = \sum_{i=1}^{n} x_i y_i$

$x \cdot y \geq 0$

 Por lo tanto :

$ \cos\theta = \frac{x \cdot y}{\|{x}\| \|y\|} \geq 0$

Como se dice que el intervalo debe estar entre $[0^\circ, 90^\circ]$

Es decir : $0^\circ \leq \theta \leq 90^\circ$

# Distance versus angle nearest neighbor

## (a) Ejemplo numérico donde el vecino más cercano en distancia no es el mismo que el vecino más cercano en ángulo.

Supongamos que tenemos los siguientes vectores en $\mathbb{R}^2$:

$$
z_1 = (1,0), \quad z_2 = (0,2), \quad x = (1,1)
$$

Calculamos las distancias euclidianas:

$$
\|x - z_1\| = \sqrt{(1-1)^2 + (1-0)^2} = \sqrt{1} = 1
$$

$$
\|x - z_2\| = \sqrt{(1-0)^2 + (1-2)^2} = \sqrt{1+1} = \sqrt{2}
$$

Dado que $1 < \sqrt{2}$, el **vecino más cercano en distancia** es $z_1$.

Ahora calculamos los ángulos entre $x$ y cada $z_i$ usando el coseno:

$$
\cos\theta_1 = \frac{x \cdot z_1}{\|x\| \|z_1\|} = \frac{(1,1) \cdot (1,0)}{\sqrt{2} \cdot 1} = \frac{1}{\sqrt{2}}
$$

$$
\cos\theta_2 = \frac{x \cdot z_2}{\|x\| \|z_2\|} = \frac{(1,1) \cdot (0,2)}{\sqrt{2} \cdot 2} = \frac{2}{2\sqrt{2}} = \frac{1}{\sqrt{2}}
$$

Ambos ángulos son iguales en este caso, pero si tomamos otro ejemplo con más vectores, se pueden encontrar casos donde el vecino por distancia no sea el mismo que el vecino por ángulo.

## (b) Demostración cuando los vectores están normalizados

Si todos los vectores están normalizados, significa que $\|z_i\| = 1$ y $\|x\| = 1$. En este caso, la distancia entre $x$ y $z_i$ es:

$$
\|x - z_i\|^2 = \|x\|^2 + \|z_i\|^2 - 2 x \cdot z_i = 1 + 1 - 2 \cos\theta_i = 2(1 - \cos\theta_i)
$$

Dado que $\cos\theta_i$ es una función decreciente de $\theta_i$, minimizar $\|x - z_i\|$ es lo mismo que maximizar $\cos\theta_i$, lo que implica minimizar $\theta_i$.

Por lo tanto, el vecino más cercano en distancia es el mismo que el vecino más cercano en ángulo cuando los vectores están normalizados.
