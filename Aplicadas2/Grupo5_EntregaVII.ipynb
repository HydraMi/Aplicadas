2. (20 ptos) **Derivadas de orden superior**.
  
  2.1 Sea $\displaystyle f(x,y)=(x^2+y^2)^{3/2}$ calcule $f_{xx},f_{yy}$ y $f_{xy}$.

##Solucion

##2.1

Sea la función

$$
f(x,y) = (x^2 + y^2)^{3/2}
$$

Nos piden calcular las derivadas parciales de segundo orden:

- $f_{xx}$: la derivada parcial de segundo orden con respecto a $x$ dos veces
- $f_{yy}$: la derivada parcial de segundo orden con respecto a $y$ dos veces
- $f_{xy}$: la derivada parcial con respecto a $x$ y luego con respecto a $y$

---

### Derivada parcial $f_x$

Para poder encontrar $f_{xx}$ y $f_{xy}$, primero necesitamos derivar $f(x,y)$ con respecto a $x$.

Recordemos que $f(x,y) = (x^2 + y^2)^{3/2}$ es una función compuesta. Aplicamos la regla de la cadena:

1. Sea $u = x^2 + y^2$, entonces $f(x,y) = u^{3/2}$

2. Derivamos usando la regla de la cadena:

$$
f_x = \frac{d}{dx} \left[ (x^2 + y^2)^{3/2} \right]
= \frac{3}{2}(x^2 + y^2)^{1/2} \cdot \frac{d}{dx}(x^2 + y^2)
$$

3. La derivada de $x^2 + y^2$ con respecto a $x$ es:

$$
\frac{d}{dx}(x^2 + y^2) = 2x
$$

Entonces:

$$
f_x = \frac{3}{2}(x^2 + y^2)^{1/2} \cdot 2x = 3x(x^2 + y^2)^{1/2}
$$

---

### Derivada parcial de segundo orden $f_{xx}$

Ahora derivamos $f_x$ con respecto a $x$ de nuevo:

$$
f_{xx} = \frac{d}{dx} \left[ 3x(x^2 + y^2)^{1/2} \right]
$$

Aquí usamos la regla del producto, porque hay un producto entre $3x$ y $(x^2 + y^2)^{1/2}$:

- Derivada del primer factor: $3$
- Derivada del segundo factor:

$$
\frac{d}{dx}(x^2 + y^2)^{1/2}
= \frac{1}{2}(x^2 + y^2)^{-1/2} \cdot 2x = \frac{x}{(x^2 + y^2)^{1/2}}
$$

Entonces:

$$
f_{xx} = 3(x^2 + y^2)^{1/2} + 3x \cdot \frac{x}{(x^2 + y^2)^{1/2}}
= 3(x^2 + y^2)^{1/2} + \frac{3x^2}{(x^2 + y^2)^{1/2}}
$$

Factorizando el denominador:

$$
f_{xx} = \frac{3(x^2 + y^2) + 3x^2}{(x^2 + y^2)^{1/2}}
= \frac{3(x^2 + y^2 + x^2)}{(x^2 + y^2)^{1/2}}
= \frac{3(2x^2 + y^2)}{(x^2 + y^2)^{1/2}}
$$

---

### Derivada parcial $f_y$

Ahora derivamos $f(x,y)$ con respecto a $y$ para poder calcular $f_{yy}$ y $f_{xy}$.

Al igual que antes, aplicamos la regla de la cadena:

$$
f_y = \frac{3}{2}(x^2 + y^2)^{1/2} \cdot \frac{d}{dy}(x^2 + y^2) = \frac{3}{2}(x^2 + y^2)^{1/2} \cdot 2y = 3y(x^2 + y^2)^{1/2}
$$

---

### Derivada parcial de segundo orden $f_{yy}$

Ahora derivamos $f_y$ con respecto a $y$:

$$
f_{yy} = \frac{d}{dy} \left[ 3y(x^2 + y^2)^{1/2} \right]
$$

Otra vez usamos la regla del producto:

- Derivada del primer factor: $3$
- Derivada del segundo factor:

$$
\frac{d}{dy}(x^2 + y^2)^{1/2} = \frac{1}{2}(x^2 + y^2)^{-1/2} \cdot 2y = \frac{y}{(x^2 + y^2)^{1/2}}
$$

Entonces:

$$
f_{yy} = 3(x^2 + y^2)^{1/2} + 3y \cdot \frac{y}{(x^2 + y^2)^{1/2}} = 3(x^2 + y^2)^{1/2} + \frac{3y^2}{(x^2 + y^2)^{1/2}}
$$

Finalmente:

$$
f_{yy} = \frac{3(x^2 + y^2 + y^2)}{(x^2 + y^2)^{1/2}} = \frac{3(x^2 + 2y^2)}{(x^2 + y^2)^{1/2}}
$$

---

### Derivada mixta $f_{xy}$

Vamos a derivar $f_x = 3x(x^2 + y^2)^{1/2}$ con respecto a $y$:

Usamos de nuevo la regla del producto:

- $3x$ se comporta como una constante (no depende de $y$)
- Derivada de $(x^2 + y^2)^{1/2}$ respecto a $y$:

$$
\frac{d}{dy}(x^2 + y^2)^{1/2} = \frac{y}{(x^2 + y^2)^{1/2}}
$$

Entonces:

$$
f_{xy} = 3x \cdot \frac{y}{(x^2 + y^2)^{1/2}} = \frac{3xy}{(x^2 + y^2)^{1/2}}
$$

Luego de calcular todas las derivadas parciales de segundo orden obtenemos:

- Derivada segunda respecto a $x$:
  $$
  f_{xx} = \frac{3(2x^2 + y^2)}{(x^2 + y^2)^{1/2}}
  $$

- Derivada segunda respecto a $y$:
  $$
  f_{yy} = \frac{3(x^2 + 2y^2)}{(x^2 + y^2)^{1/2}}
  $$

- Derivada mixta $f_{xy}$ (igual a $f_{yx}$):
  $$
  f_{xy} = \frac{3xy}{(x^2 + y^2)^{1/2}}
  $$


  2.2 Sea $\displaystyle f(X)=\cos(A\cdot X)$ donde $X\in\mathbb{R}^n$ y  $A$ es un vector constante en $\mathbb{R}^n$. Esto es

  $$f(x_1,...,x_n)=\cos(a_1x_1+...+a_nx_n).$$

 Muestre que $$f_{x_1x_1}+...+f_{x_nx_n}=-||A||^2f.$$

## Solucion

Tenemos que: $f_{x_1x_1}+...+f_{x_nx_n}=-||A||^2f$ donde:


*   $||A||^2$ es $(a_1^2 + a_2^2 + \cdots + a_n^2)$

*   $f_{x_nx_n}$ significa la segunda derivada de $f$ respecto a $x$

ahora, debemos derivar $f(x_1,...,x_n)=\cos(a_1x_1+...+a_nx_n)$ con respecto a $x$

primera derivada

$f{x_i}= -sin(a_1x_1+...+a_nx_n)*a_i$

segunda derivada

$f{x_ix_i}= -cos(a_1x_1+...+a_nx_n)*a_i^2$

ahora sumanos todas las segundas derivadas con respecto a $x$ de tal forma que quede asi:

$
\sum_{i=1}^{n} \frac{\partial^2 f}{\partial x_i^2} = \sum_{i=1}^{n} \left( -\cos(A \cdot X) \cdot a_i^2 \right)
$

lo que se puede interpretar como:

$f_{x_1 x_1} + \cdots + f_{x_n x_n} = -\|A\|^2 \cos(A \cdot X)$

que es lo mismo que:

$f_{x_1x_1}+...+f_{x_nx_n}=-||A||^2f$

debido a que la $f$ inicial es igual a $\cos(A \cdot X)$
