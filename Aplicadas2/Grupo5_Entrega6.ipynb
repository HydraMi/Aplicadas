1.4 *Jacobiano de una función $F:\mathbb{R}^n\rightarrow\mathbb{R}^n$*. La matriz de derivadas de una función $F:\mathbb{R}^n\rightarrow\mathbb{R}^n$ dada por $F(x)=F(x_1,...,x_n)=(f_1(x),...,f_n(x))$ en un punto $P$ se puede escribir en forma matricial como:

$$DF(P)=\begin{bmatrix}
\nabla f_1(x)^T\\
\vdots\\
\nabla f_n(x)^T
\end{bmatrix}.$$

El determinante de $DF(P)$ se denomina *Jacobiano* de $F$ en $A$ y se denota

$$JF(P)=\det DF(P).$$

Considere la función $F(x,y)=(x+y^2,x^3+xy)$. Encuentre el Jacobiano de $F$ en el punto $P=(1,2)$.

#Solucion

La matriz de derivadas parciales (matriz Jacobiana) de $F$ se construye con los gradientes transpuestos de cada componente:

$$
DF(x,y) = \begin{bmatrix}
\frac{\partial f_1}{\partial x} & \frac{\partial f_1}{\partial y} \\
\frac{\partial f_2}{\partial x} & \frac{\partial f_2}{\partial y}
\end{bmatrix}.
$$

Se calcula las derivadas parciales de cada componente:

\begin{align*}
\frac{\partial f_1}{\partial x} &= \frac{\partial}{\partial x}(x + y^2) = 1, \\
\frac{\partial f_1}{\partial y} &= \frac{\partial}{\partial y}(x + y^2) = 2y, \\
\frac{\partial f_2}{\partial x} &= \frac{\partial}{\partial x}(x^3 + x y) = 3x^2 + y, \\
\frac{\partial f_2}{\partial y} &= \frac{\partial}{\partial y}(x^3 + x y) = x.
\end{align*}

Por lo tanto, la matriz Jacobiana es:

$$
DF(x,y) = \begin{bmatrix}
1 & 2y \\
3x^2 + y & x
\end{bmatrix}.
$$

Evaluamos en el punto $P = (1,2)$:

\begin{align*}
2y &= 4, \\
3x^2 + y &= 3(1)^2 + 2 = 5, \\
x &= 1.
\end{align*}

Entonces:

$$
DF(1,2) = \begin{bmatrix}
1 & 4 \\
5 & 1
\end{bmatrix}.
$$

Calculamos el determinante:

$$
JF(1,2) = \det DF(1,2) = (1)(1) - (4)(5) = 1 - 20 = -19.
$$

Por lo tanto, el Jacobiano de $F$ en el punto $P = (1,2)$ es:

$$
{JF(1,2) = -19}.
$$

 2.1 Encuentre la ecuación del plano tangente a la función $f(x,y)=x+y^2$ en cualquier punto $(a,b)$. Diseñe un applet en GeoGebra que dibuje simultáneamente los planos tangentes a esta función y los vectores normales a dicho plano en el punto de tangencia,  para valores de $-5\leq a,b\leq 5$.

 # Solucion

Sea la función $f(x,y) = x + y^2$. La ecuación del plano tangente a esta función en un punto $(a,b)$ se puede obtener a partir de:

$$
z = f(a,b) + f_x(a,b)(x - a) + f_y(a,b)(y - b),
$$

donde $f_x$ y $f_y$ representan las derivadas parciales de $f$ con respecto a $x$ y $y$, respectivamente.

Calculamos las derivadas parciales de $f$:

\begin{align*}
f_x(x,y) &= \frac{\partial}{\partial x}(x + y^2) = 1, \\
f_y(x,y) &= \frac{\partial}{\partial y}(x + y^2) = 2y.
\end{align*}

evaluando en el punto $(a,b)$:

\begin{align*}
f(a,b) &= a + b^2, \\
f_x(a,b) &= 1, \\
f_y(a,b) &= 2b.
\end{align*}

Sustituyendo en la fórmula del plano tangente:

\begin{align*}
z &= a + b^2 + 1(x - a) + 2b(y - b) \\
  &= x + 2by - b^2.
\end{align*}

Por lo tanto, la ecuación del plano tangente a la función $f(x,y) = x + y^2$ en el punto $(a,b)$ es:

$$
{z = x + 2b y - b^2}.
$$

Adicionalmente, el vector normal al plano tangente en $(a,b)$ está dado por:

$$
\vec{n} = \left\langle f_x(a,b), f_y(a,b), -1 \right\rangle = \left\langle 1, 2b, -1 \right\rangle.
$$

  2.2 Sea $\displaystyle f(x,y)=e^{-(x^2+y^2)}$ y $\displaystyle g(x,y)=\frac{e^{-1}}{x^2+y^2}$. Suponga que  $(a,b)$ es un punto tal que $a^2+b^2=1$.

  a.   Muestre que $f(a,b)=g(a,b)$.

  b.   Encuentre $\nabla f$ y $\nabla g$.

  c. Muestre que la gráfica $f$ es tangente a la gráfica de $g$ en todo punto $(a,b)$. Diseñe un applet en GeoGebra que ilustre esta situación. Sugerencia: Use planos tangentes,  la vista 3D e intersecte secciones transversales con las gráficas de $f$ y $g$.

  ###Respuesta

  a. En primer lugar, se evalua $f(a,b)$, sustituyendo $x=a$ y $y=b$ en $f(x,y)$

  $$f(a,b) = e^{-(a^2 + b^2)}$$
  
  Teniendo en cuenta que $a^2 + b^2 = 1$, se simplifica a

  $$f(a,b) = e^{-1}$$

  Posteriormente, se evalua $g(a,b)$ de la misma manera

  $$g(a,b) = \frac{e^{-1}}{a^2 + b^2}$$

  $$g(a,b) = \frac{e^{-1}}{1}$$

  $$g(a,b) = e^{-1}$$

  Llegandose asi a que ambas funciones, al ser evaluadas en $(a,b)$ son iguales.

  b. En primer lugar, se calcula el vector gradiente de $f$. El vector gradiente de $f$ es el vector de sus derivadas parciales:

  $$\nabla f = (\frac{\partial f}{\partial x} + \frac{\partial f}{\partial y})$$

  Para encontrarla, se encuentra cada derivada parcial haciendo uso de la regla de la cadena.

  $$\frac{\partial f}{\partial x} = e^{-(x^2 + y^2)} * (-2x)$$

  $$\frac{\partial f}{\partial x} = -2xe^{-(x^2 + y^2)}$$

  $$\frac{\partial f}{\partial y} = e^{-(x^2 + y^2)} * (-2y)$$

  $$\frac{\partial f}{\partial y} = -2ye^{-(x^2 + y^2)}$$

  De acuerdo con esto, el vector gradiente de $f$ seria:

  $$\nabla f = (-2xe^{-(x^2 + y^2)} , -2ye^{-(x^2 + y^2)})$$

  Al realizar la evaluacion en $(a,b)$, tomando en cuenta que $a^2 + b^2 = 1$

  $$\nabla f(a,b) = (-2ae^{-(a^2 + b^2)} , -2be^{-(a^2 + b^2)})$$

  $$\nabla f(a,b) = (-2ae^{-1} , -2be^{-1})$$

  Ahora se calcula el vector gradiente de $g$, para lo cual se hace uso de la regla del cociente:

  Para $\frac{\partial g}{\partial x}$:

  $$\nabla g = (\frac{\partial g}{\partial x} + \frac{\partial g}{\partial y})$$

  $$\frac{\partial g}{\partial x} = e^{-1} * \frac{d}{dx}(\frac{1}{x^2 + y^2})$$

  $$\frac{\partial g}{\partial x} = e^{-1} * (\frac{-2x}{(x^2 + y^2)^2})$$


  $$\frac{\partial g}{\partial x} = (\frac{-2xe^{-1}}{(x^2 + y^2)^2})$$

  Para $\frac{\partial g}{\partial y}$:

  $$\frac{\partial g}{\partial y} = e^{-1} * \frac{d}{dy}(\frac{1}{x^2 + y^2})$$

  $$\frac{\partial g}{\partial y} = e^{-1} * (\frac{-2y}{(x^2 + y^2)^2})$$

  $$\frac{\partial g}{\partial y} = (\frac{-2ye^{-1}}{(x^2 + y^2)^2})$$

  Segun lo cual, el vector gradiente seria:

  $$\nabla g = (\frac{-2xe^{-1}}{(x^2 + y^2)^2},\frac{-2ye^{-1}}{(x^2 + y^2)^2}  )$$

  Al evaluar en $(a,b)$, dado que $a^2 + b^2 = 1$, se obtiene que:

  $$\nabla g(a,b) = (\frac{-2ae^{-1}}{(a^2 + b^2)^2},\frac{-2be^{-1}}{(a^2 + b^2)^2}  )$$

  $$\nabla f(a,b) = (-2ae^{-1} , -2be^{-1})$$

Para demostrar que la grafica $f$ es tangente a la grafica $g$ en todo punto $(a,b)$ que cumpla con la condicion dada, se realiza la comparacion de sus gradientes, como en el literal anterior se verifico que estos coinciden en cada punto que satisface la condicion, se concluye que los planos tangentes a las superficies son iguales en esos puntos, por lo tanto, las superficies son tangentes ahi.
