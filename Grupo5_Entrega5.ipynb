1.3 Sea $L$ una función lineal de $\mathbb{R}^n→\mathbb{R}^m$. Muestre que:

  a. Si $L(X) = 0$ y $L(Y) = 0$, entonces $L(X+Y)$ también es $0$.

  b. Si $L(X) = 0$ y $c$ es un número, entonces $L(cX)$ también es $0$.

  # Solucion

#$1.3$

##$Punto$ $A$  

Lo primero que se debe de hacer es demostrar las siguientes propiedades:

### Propiedad 1: Aditividad

Si $L(X) = 0$ y $L(Y) = 0$, entonces $L(X+Y) = 0$.

#### Aplicando la Propiedad de Aditividad

Dado que $L$ es una transformación lineal, cumple la propiedad de aditividad:

$$
L(X + Y) = L(X) + L(Y)
$$

Como se nos ha dado que $L(X) = 0$ y $L(Y) = 0$, sustituimos estos valores en la ecuación:

$$
L(X+Y) = 0 + 0 = 0.
$$

Por lo tanto, se ha demostrado que si $X$ y $Y$ son transformados en el vector cero, su suma tambien lo ser.

### Propiedad 2: Homogeneidad

Si $L(X) = 0$ y $c$ es un numero real, entonces $L(cX) = 0$.

#### Aplicando la Propiedad de Homogeneidad

Dado que $L$ es lineal, tambien cumple la propiedad de homogeneidad:

$$
L(cX) = cL(X)
$$

Sustituyendo el valor dado $L(X) = 0$:

$$
L(cX) = c \cdot 0 = 0.
$$

Por lo tanto, cualquier multiplo escalar de un vector que ya es transformado en cero seguira siendo transformado en cero. Entoces se puede decir que si una transformacion lineal $L$ convierte dos vectores en el vector cero, su suma tambiéen se convertira en el vector cero. Ademas, si un vector ya es transformado en cero, cualquier version escalada de ese vector tambien lo sera. Esto simplemente nos dice que el conjunto de todos los vectores que $L$ convierte en cero tiene una estructura especial (es un subespacio vectorial).

1.4 Muestre que para una función lineal $𝐿$ de  $\mathbb{R}^n→\mathbb{R}^m$, el valor $L(x)$ es una combinación lineal de las columnas $v_j$ de la matriz de $L$:

$$L(x)=x_1v_1+...x_nv_n.$$

#$1.4$

#Solucion

### Definición de la Matriz de Transformación

Sea $A$ la matriz de la transformación $L$, con dimensiones $m \times n$, compuesta por las columnas $v_1, v_2, \dots, v_n$:

$$
A =
\begin{pmatrix}
| & | &  & | \\
 v_1 & v_2 & \dots & v_n \\
| & | &  & |
\end{pmatrix}
$$

Un vector $x$ en $R^n$ se expresa como:

$$x = \begin{pmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{pmatrix}$$

### Multiplicacion de Matriz por Vector

Cuando aplicamos la transformación $L(x)$, realizamos la multiplicacion matricial:

$$ L(x) = A x. $$

Este producto se puede escribir como:

$$
L(x) =
\begin{pmatrix}
| & | &  & | \\
 v_1 & v_2 & \dots & v_n \\
| & | &  & |
\end{pmatrix}
\begin{pmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{pmatrix}
$$

Distribuyendo el producto, obtenemos:

$$ L(x) = x_1 v_1 + x_2 v_2 + \dots + x_n v_n. $$

Esto demuestra que la imagen de $x$ bajo la transformacion $L$ es una combinacion lineal de las columnas de la matriz $A$, con los coeficientes dados por los componentes de $x$.

### Ejemplo Numérico

Como un ejemplo puntual que puedo hacer es que la matriz de la transformación sea:

$$
A = \begin{pmatrix} 2 & -1 \\ 0 & 3 \\ 4 & 5 \end{pmatrix}
$$

Las columnas de $A$ son:

$$ v_1 = \begin{pmatrix} 2 \\ 0 \\ 4 \end{pmatrix}, \quad v_2 = \begin{pmatrix} -1 \\ 3 \\ 5 \end{pmatrix} $$

Tomemos el vector $x$:

$$ x = \begin{pmatrix} 1 \\ 2 \end{pmatrix} $$

Aplicamos $L(x)$:

$$
L(x) = 1 v_1 + 2 v_2 = 1 \begin{pmatrix} 2 \\ 0 \\ 4 \end{pmatrix} + 2 \begin{pmatrix} -1 \\ 3 \\ 5 \end{pmatrix}
$$

Realizando la suma:

$$
L(x) = \begin{pmatrix} 2 + (-2) \\ 0 + 6 \\ 4 + 10 \end{pmatrix} = \begin{pmatrix} 0 \\ 6 \\ 14 \end{pmatrix}
$$

Al aplicar una funcion lineal $L(x)$ es lo mismo que tomar las columnas de su matriz y combinarlas segun los valores del vector $x$. Mejor dicho, cada entrada de $x$ nos dice cuanta cantidad de cada columna se usa en la combinacion.

###Respuesta

El conjunto de nivel de una funcion es el subconjunto de puntos en el espacio en los que la funcion tiene el mismo valor
Se deben hallar las curvas que existen en la funcion $f(x,y)$.

La funcion dada es una funcion indicadora del disco unitario, cuyo proposito es indicar si un punto $(x,y)$ pertenece o no al disco. Los conjuntos de nivel $L_C$ se derivan directamente de la definicion de $f$, la definicion formal de conjunto de nivel es:

$$L_C = [(x,y) | f(x,y) = c]$$

Como $f$ solo toma valores 0 o 1, hay 2 casos posibles:

Cuando $C = 1$, el cual define el interior del disco, incluyendo el borde:

$$f(x,y) = 1 \rightarrow x^2 + y^2 \leq 1$$

Y el caso cuando $C = 0$, el cual define el exterior del disco:

$$f(x,y) = 0  \rightarrow  x^2 + y^2 > 1$$

Para cualquier otro valor de $C$, la funcion nunca toma dichos valores, por lo cual, no harian parte de los conjuntos de nivel.

En el applet se ven representadas diferentes curvas , sin embargo, la unica que realmente es una curva de nivel es $x^2 + y^2 = 1$, ya que la funcion solo toma los valores 1 y 0, las demas circunferencias no son curvas de nivel de $f$. Valores intermedios no son curvas de nivel debido a que se trata de una funcion discontinua.

2.4 Describa geométricamente (por ejemplo, son rectas, parábolas, etc.) los conjuntos de nivel $c$, $f (x, y) = c$ en $\mathbb{R}^2$, para las  funciones que encuentra a continuación. Si no existe la curva de nivel explique por qué. Use un applet o un conjunto de applets para visualizar en 3D todas las gráficas de las funciones y otros para sus respectivos conjuntos de nivel. De ser posible que se vean tanto las superficies como las curvas de nivel simultáneamente.
$$f_1(x,y) = x + 2y, \quad c = -1, 0, 1$$
$$f_2(x,y) = xy, \quad c = -1, 0, 1$$
$$f_3(x,y) = x^2 - y,\quad c = -1, 0, 1$$
$$f_4(x,y) = 1 - x^2 - y^2,\quad c = 0, \frac{1}{2}, 1$$
$$f_5(x,y) = \frac{1}{x^2 + y^2},\quad c = 0, 1, 2.$$



Para $f(x,y)$, sus curvas de nivel estan definidas por la ecuacion $f(x,y) = c$, donde $c$ es una constante. Estas curvas representan los puntos en el plano donde la funcion toma el mismo valor. A continuacion,

1.
   $f_1(x,y) = x + 2y$  
   Esta es una funcion lineal, por lo que sus curvas de nivel son rectas en el plano. Se obtiene:  
   - $c = -1$: $x + 2y = -1$  
   - $c = 0$: $x + 2y = 0$  
   - $c = 1$: $x + 2y = 1$  

   Despejando $y$ en cada caso:  
   $$
   y = -\frac{1}{2}x + \frac{c}{2}
   $$  
   Esto muestra que todas las rectas tienen la misma pendiente $-\frac{1}{2}$, lo que significa que son rectas paralelas entre si.


2.   
   $f_2(x,y) = xy$  
   En este caso es una funcion de producto, el producto de $x$ y $y$ es constante. Se obtiene:  
   - $c = -1$: $xy = -1$, que representa una hiperbola con ramas en los cuadrantes II y IV.  
   - $c = 0$: $xy = 0$, lo que significa que al menos una de las variables debe ser cero. Esto da como resultado las rectas coordenadas $x = 0$ y $y = 0$, es decir, los ejes.  
   - $c = 1$: $xy = 1$, que representa otra hiperbola, pero ahora con ramas en los cuadrantes I y III.  

   En general, las curvas de nivel de esta funcion son hiperbolas que cambian su orientacion segun el signo de $c$.


3.
   $f_3(x,y) = x^2 - y$  
   La ecuacion es una funcion cuadratica y es una familia de parabolas con eje de simetria vertical:  
   - $c = -1$: $x^2 - y = -1 \Rightarrow y = x^2 + 1$ (parabola desplazada una unidad hacia arriba).  
   - $c = 0$: $y = x^2$ (parabola base con vertice en el origen).  
   - $c = 1$: $y = x^2 - 1$ (parabola desplazada una unidad hacia abajo).  

   Estas curvas de nivel son parabolas que se mueven verticalmente segun el valor de $c$.


4.   
   $f_4(x,y) = 1 - x^2 - y^2$  
   Reordenando, $x^2 + y^2 = 1 - c$ y nos percatamos que es una funcion circular, lo que representa circunferencias centradas en el origen :  
   - $c = 0$: $x^2 + y^2 = 1$, que es una circunferencia de radio 1.  
   - $c = \frac{1}{2}$: $x^2 + y^2 = \frac{1}{2}$, una circunferencia mas pequeña con radio $\sqrt{\frac{1}{2}}$.
   - $c = 1$: $x^2 + y^2 = 0$, que representa un punto unico en $(0,0)$.  

   Si $c > 1$, la ecuacion no tiene solucion real, lo que indica que no existen curvas de nivel en ese caso.


5.
   $f_5(x,y) = \frac{1}{x^2 + y^2}$  
  
   Despejando, obtenemos la ecuacion $x^2 + y^2 = \frac{1}{c}$, que representa circunferencias centradas en el origen. Esta es un ejemplo de una funcion racional, ya que esta definida como el cociente de dos expresiones algebraicas.
   - $c = 1$: $x^2 + y^2 = 1$ (circunferencia de radio 1).  
   - $c = 2$: $x^2 + y^2 = \frac{1}{2}$ (circunferencia de radio $\frac{1}{\sqrt{2}}$, mas pequeña).  
   - $c = 0$: No tiene solucion, ya que $x^2 + y^2$ es siempre positivo y nunca puede ser cero para valores reales distintos de $(0,0)$.  

   A medida que $c$ aumenta, el radio de la circunferencia disminuye, y cuando $c \to \infty$, la curva de nivel se acerca al punto $(0,0)$.
