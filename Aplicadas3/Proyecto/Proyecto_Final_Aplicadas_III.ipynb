## **14.1 Clasificación**

---

En el problema de ajuste de datos del capítulo 13, el objetivo era reproducir o predecir el resultado $y$, el cual es un número escalar, a partir de un vector $n$-dimensional $\mathbf{x}$.  
En un problema de **clasificación**, el resultado o variable dependiente $y$ toma solo un número finito de valores, y por esta razón se denomina a veces una **etiqueta** o, en estadística, una **variable categórica**.

En el caso más simple, $y$ tiene solo dos valores posibles, por ejemplo: **verdadero o falso**, o **spam o no spam**.  
Este caso se conoce como el **problema de clasificación binaria** o **clasificación booleana**, ya que el resultado $y$ puede tomar solo dos valores.

---

### **Codificación de la variable de salida**

Se codifica $y$ como un número real, tomando $y = +1$ para representar *verdadero* y $y = -1$ para representar *falso*.  
(También se puede usar $y = +1$ y $y = 0$, o cualquier otro par de valores distintos.)

Al igual que en el ajuste de datos con valores reales, se asume una relación aproximada de la forma:
$$
y \approx f(\mathbf{x})
$$
donde $f : \mathbb{R}^n \rightarrow \{-1, +1\}$.

Nuestro modelo tendrá la forma:
$$
\hat{y} = \hat{f}(\mathbf{x})
$$
donde $\hat{f} : \mathbb{R}^n \rightarrow \{-1, +1\}$.

El modelo $\hat{f}$ se denomina **clasificador**, ya que clasifica los vectores $\mathbf{x}$ en dos grupos:
- Aquellos para los cuales $\hat{f}(\mathbf{x}) = +1$.
- Aquellos para los cuales $\hat{f}(\mathbf{x}) = -1$.

Como en el caso del ajuste de datos, el clasificador $\hat{f}$ se construye a partir de datos observados. Toma estos datos observados y a partir de ellos se genera la función que creemos se comporta como nuestra funcion original,

---

### **Ejemplos de clasificadores booleanos**

- **Detección de spam en correos electrónicos:**  
  El vector $\mathbf{x}$ contiene características de un mensaje de correo, tales como recuento de palabras, número de signos de exclamación o palabras en mayúscula, y datos sobre el origen del mensaje.  
  El resultado es $y = +1$ si el mensaje es spam, y $y = -1$ en caso contrario.  
  Los datos provienen de usuarios que marcan explícitamente ciertos mensajes como “no deseados”.

- **Detección de fraude:**  
  El vector $\mathbf{x}$ incluye características de una transacción con tarjeta de crédito: gasto promedio mensual, precio medio de compras recientes, número de transacciones por categoría, saldo promedio, entre otros.  
  El resultado es $y = +1$ para una transacción fraudulenta y $y = -1$ en caso contrario.  
  Los datos se obtienen de registros históricos que contienen ejemplos verificados de fraude y de transacciones legítimas.

- **Clasificación de documentos:**  
  El vector $\mathbf{x}$ representa un histograma de palabras de un documento.  
  El resultado es $y = +1$ si el documento pertenece a una categoría específica (por ejemplo, política) y $y = -1$ en caso contrario.  
  Los datos para construir el clasificador provienen de un conjunto de documentos previamente etiquetados.

- **Detección de enfermedades:**  
  Cada ejemplo corresponde a un paciente. El resultado $y = +1$ indica que el paciente tiene una enfermedad determinada, mientras que $y = -1$ indica que no la tiene.  
  El vector $\mathbf{x}$ contiene características médicas relevantes: edad, sexo, resultados de pruebas y síntomas específicos.  
  Los datos provienen de registros hospitalarios o estudios clínicos con diagnósticos confirmados.

- **Receptor de comunicaciones digitales:**  
  En un sistema de comunicación electrónica, $y$ representa un bit (0 o 1) que se transmite.  
  El vector $\mathbf{x}$ contiene $n$ mediciones de la señal recibida, y el modelo $\hat{y} = \hat{f}(\mathbf{x})$ se conoce como **bit decodificado**.  
  En este contexto, el clasificador $\hat{f}$ se llama **decodificador** o **detector**.  
  Los datos para construir el decodificador provienen de una secuencia de bits conocida por el receptor, llamada **señal de entrenamiento**.

---

### **Errores de predicción**

Para un punto de datos $(\mathbf{x}, y)$ con resultado predicho $\hat{y} = \hat{f}(\mathbf{x})$, existen cuatro posibilidades:

- **Verdadero positivo (TP):** $y = +1$ y $\hat{y} = +1$  
- **Verdadero negativo (TN):** $y = -1$ y $\hat{y} = -1$  
- **Falso positivo (FP):** $y = -1$ y $\hat{y} = +1$  
- **Falso negativo (FN):** $y = +1$ y $\hat{y} = -1$

En los dos primeros casos la predicción es correcta; en los dos últimos se cometen errores.  
Los **falsos positivos** se denominan *errores de tipo I* y los **falsos negativos** *errores de tipo II*.

Dependiendo de la aplicación, puede importar más minimizar uno de estos tipos de errores que el otro.

---

### **Matriz de confusión**

Para un conjunto de datos
$$
(\mathbf{x}^{(1)}, y^{(1)}), \ldots, (\mathbf{x}^{(N)}, y^{(N)})
$$
y un modelo $\hat{f}$, se pueden contar los casos de cada tipo (TP, TN, FP, FN) y representarlos en una **matriz de confusión**, donde las columnas corresponden al valor predicho $\hat{y}$ y las filas al valor real $y$.

\
\begin{array}{c|cc|c}
\text{Resultado real} & \hat{y} = +1 & \hat{y} = -1 & \text{Total} \\
\hline
y = +1 & N_{tp} & N_{fn} & N_p \\
y = -1 & N_{fp} & N_{tn} & N_n \\
\hline
\text{Total} & N_{tp} + N_{fp} & N_{fn} + N_{tn} & N
\end{array}
\

Los elementos diagonales representan las predicciones correctas, mientras que los elementos fuera de la diagonal representan errores.  
El total de los cuatro valores es $N$, el número total de ejemplos.

---

### **Métricas de desempeño**

A partir de la matriz de confusión se definen varios indicadores:

- **Tasa de error:**  
  
  $$\text{Error} = \frac{N_{fp} + N_{fn}}{N}$$
  
  Es la proporción total de errores cometidos.

- **Tasa de verdaderos positivos (sensibilidad o recall):**  
  
  $$\text{TPR} = \frac{N_{tp}}{N_p}$$
  
  Es la fracción de casos positivos correctamente clasificados.

- **Tasa de falsos positivos (false alarm rate):**  
  
  $$\text{FPR} = \frac{N_{fp}}{N_n}$$
  
  Es la fracción de casos negativos que fueron clasificados incorrectamente como positivos.

- **Especificidad (tasa de verdaderos negativos):**  
  
  $$\text{TNR} = \frac{N_{tn}}{N_n} = 1 - \text{FPR}$$
  
  Es la fracción de casos negativos correctamente identificados.

- **Precisión:**  
  
  $$\text{Precisión} = \frac{N_{tp}}{N_{tp} + N_{fp}}$$
  
  Es la fracción de predicciones positivas que son correctas.

---

### **Ejemplo: Detección de spam**

La siguiente matriz muestra el desempeño de un detector de spam sobre un conjunto de $N = 1266$ correos electrónicos, de los cuales 127 son spam y 1139 no lo son:


\begin{array}{c|cc|c}
\text{Resultado real} & \hat{y} = +1\ (\text{spam}) & \hat{y} = -1\ (\text{no spam}) & \text{Total} \\
\hline
y = +1\ (\text{spam}) & 95 & 32 & 127 \\
y = -1\ (\text{no spam}) & 19 & 1120 & 1139 \\
\hline
\text{Total} & 114 & 1152 & 1266
\end{array}


El modelo obtiene:

$$\text{Error} = \frac{19 + 32}{1266} = 4.03\%$$


$$\text{TPR} = \frac{95}{127} = 74.8\%, \quad \text{FPR} = \frac{19}{1139} = 1.67\%$$


Esto significa que el clasificador detecta aproximadamente el 75\% del spam y clasifica erróneamente solo el 1.7\% de los correos legítimos como spam.

---

### **Validación en problemas de clasificación**

En los problemas de clasificación se evalúan métricas como la tasa de error, la tasa de verdaderos positivos y la tasa de falsos positivos.  
La **validación fuera de muestra** y la **validación cruzada** se realizan utilizando las métricas de desempeño relevantes para el problema.  
En algunas aplicaciones puede ser más importante maximizar la sensibilidad, mientras que en otras se busca minimizar la tasa de falsos positivos, dependiendo del contexto.

## **Sección 14.2.1 — Clasificación de flores *Iris***

Boyd introduce la clasificación de mínimos cuadrados (least squares classification), aplicada al conjunto de datos de las flores Iris

Inicialmente, el objetivo en un problema de clasificación binaria es predecir una etiqueta

$$
y^{(i)} \in \{-1, +1\}
$$

a partir de un vector de características

$$
x^{(i)} \in \mathbb{R}^n
$$

Queremos construir una función de decisión $\hat{f}(x)$ que asigne correctamente la clase a cada muestra

Para ello, Boyd utiliza el modelo lineal:

$$
\tilde{f}(x) = x^\top \beta + v
$$

donde:
- $\beta \in \mathbb{R}^n$ es el vector de coeficientes, que define la orientación del hiperplano
- $v$ es el término de sesgo (bias), que lo desplaza  
- y $\tilde{f}(x)$ produce una salida continua (no discreta)

Para clasificar, se toma el signo de la salida:

$$
\hat{f}(x) = \operatorname{sign}(\tilde{f}(x)) =
\begin{cases}
+1, & \tilde{f}(x) \ge 0\\
-1, & \tilde{f}(x) < 0
\end{cases}
$$



Este modelo se entrena buscando los parámetros $(\beta, v)$ que minimicen el error cuadrático total entre las etiquetas reales y las predichas:

$$
J(\beta, v) = \sum_{i=1}^{N} \left(y^{(i)} - (x^{(i)\top}\beta + v)\right)^2
$$

En forma matricial, se define la matriz de diseño $X \in \mathbb{R}^{N \times (n+1)}$ como:

$$
X =
\begin{bmatrix}
1 & x_1^{(1)} & x_2^{(1)} & \dots & x_n^{(1)}\\
1 & x_1^{(2)} & x_2^{(2)} & \dots & x_n^{(2)}\\
\vdots & \vdots & \vdots & & \vdots \\
1 & x_1^{(N)} & x_2^{(N)} & \dots & x_n^{(N)}
\end{bmatrix}
$$

y el vector de parámetros:

$$
\theta =
\begin{bmatrix}
v \\ \beta
\end{bmatrix}
$$

Entonces la función objetivo se expresa como

$$
J(\theta) = \|X\theta - y\|^2
$$



Minimizando $J(\theta)$ respecto a $\theta$ (proyectar ortogonalmente a
$y$ sobre el espacio columna de $X$), obtenemos las ecuaciones normales:

$$
X^\top X \, \hat{\theta} = X^\top y
$$

cuyo resultado es:

$$
\hat{\theta} = (X^\top X)^{-1} X^\top y
$$

Este vector $\hat{\theta}$ contiene tanto los coeficientes $\beta$ como el término de sesgo $v$ del clasificador lineal óptimo en el sentido de mínimos cuadrados

Ya con esto, podemos ver la aplicación del modelo al conjunto Iris, el cual tiene tiene 150 ejemplos, 3 especies (Setosa, Versicolor y Virginica) y 4 atributos:

1. Largo del sépalo  
2. Ancho del sépalo  
3. Largo del pétalo  
4. Ancho del pétalo  

El objetivo del experimento es clasificar entre:

$$
y = +1 \text{ si la flor es Virginica}
$$
$$
y = -1 \text{ si no lo es}
$$

Usando los 150 ejemplos y aplicando la fórmula de mínimos cuadrados,
Boyd obtiene el modelo:

$$
\tilde{f}(x) = -2.39 - 0.0918x_1 + 0.406x_2 + 0.00798x_3 + 1.10x_4
$$

Para el cual se obtiene que el clasificador final es:

$$
\hat{f}(x) = \operatorname{sign}(\tilde{f}(x))
$$


Mediante los resultados obtenidos se interpreta que los coeficientes indican la importancia relativa de cada variable, de manera que:
- El mayor valor corresponde a $x_4$ (ancho del pétalo), el atributo más relevante para distinguir Virginica
- Los coeficientes de sépalo ($x_1$, $x_2$) son menores, lo que refleja que esas variables tienen menos poder discriminante

Y la matriz de confusión  obtenida fue:

| Verdadero | Pred. +1 | Pred. -1 |
|------------|-----------|----------|
| +1 | 46 | 4 |
| -1 | 7 | 93 |

Con una tasa de error total de:

$$
\text{Error} = \frac{7 + 4}{150} = 7.3\%
$$

Esto nos permite concluir que el método de mínimos cuadrados permite construir clasificadores lineales simples y eficientes, y aunque no está diseñado específicamente para clasificación, funciona bien cuando las clases son aproximadamente separables de forma lineal. En el caso de las flores de Iris, el error del 7.3% demuestra un buen desempeño usando solo una combinación lineal de las cuatro características. El principio utilizado se puede extender a modelos no lineales, agregando términos polinómicos


```python
## Conjunto Iris
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

##Cargar dataset Iris

data = load_iris()
X = data.data
y = np.where(data.target == 2, 1, -1) # Virginica = +1, otras = -1

N, n = X.shape

##Matriz de diseño con columna de unos

X_design = np.column_stack([np.ones(N), X])

##Cálculo de theta por ecuaciones normales

theta = np.linalg.inv(X_design.T @ X_design) @ X_design.T @ y

theta
```

```python
theta_df = pd.DataFrame(theta, index=["v","x1","x2","x3","x4"], columns=["coef"])
theta_df
```

```python
## Predicciones

y_pred = np.sign(X_design @ theta)

## Matriz de confusión manual

confusion = pd.crosstab(pd.Series(y, name="Real"),
pd.Series(y_pred, name="Predicho"))
confusion

## Error
error = np.mean(y_pred != y)
error
```

## **1. Contextualización del Problema**

### **1.1 Introducción a la Clasificación por Mínimos Cuadrados**

En el ámbito del aprendizaje automático, la **clasificación** es una tarea fundamental que consiste en asignar etiquetas categóricas a instancias de datos basándose en sus características.  
El método de **mínimos cuadrados para clasificación**, presentado en el *Capítulo 14* de **Boyd & Vandenberghe**, reformula este problema como un problema de ajuste cuadrático, donde la frontera de decisión se obtiene resolviendo un sistema lineal.

---

### **1.2 Formulación Matemática del Problema**

Dado un conjunto de datos de entrenamiento
${(x^{(i)}, y^{(i)})}_{i=1}^N$, donde:

- $x^{(i)} \in \mathbb{R}^n$ son los vectores de características.  
- $y^{(i)} \in \{-1, +1\}$ son las etiquetas binarias.

Buscamos una función
$f: \mathbb{R}^n \rightarrow \mathbb{R}$  
que minimice el **error cuadrático**:

$$
\min_{\theta} \sum_{i=1}^{N} (y^{(i)} - f(x^{(i)}))^2
$$

donde

$$
f(x) = \theta_1 f_1(x) + \cdots + \theta_p f_p(x)
$$

es una **combinación lineal de funciones base** o características.

---

### **1.3 Clasificador Final**

El clasificador se define mediante la **función signo**:

$$
\hat{f}(x) = \operatorname{sign}(\tilde{f}(x))
$$

donde $\tilde{f}(x)$ representa el ajuste obtenido por mínimos cuadrados.

---

## **2. Marco Teórico y Fundamentos Matemáticos**

### **2.1 Álgebra Lineal en Clasificación**

#### **2.1.1 Formulación Matricial**

El problema de mínimos cuadrados puede expresarse en forma matricial como:

$$
\min_{\theta} \| A\theta - y \|_2^2
$$

donde:

- $A \in \mathbb{R}^{N \times p}$ es la **matriz de diseño** que contiene las características.
- $\theta \in \mathbb{R}^p$ es el **vector de parámetros**.
- $y \in \mathbb{R}^N$ es el **vector de etiquetas**.

---

#### **2.1.2 Solución por Ecuaciones Normales**

La condición de primer orden para el óptimo es:

$$
A^T A \theta = A^T y
$$

Si $A^T A$ es invertible, la solución se obtiene directamente como:

$$
\theta = (A^T A)^{-1} A^T y
$$

Esta expresión representa la **solución analítica** del problema de mínimos cuadrados y define los parámetros del hiperplano separador.

---

### **2.2 Descomposición en Valores Singulares (SVD)**

#### **2.2.1 Teoría de la SVD**

Toda matriz $A \in \mathbb{R}^{m \times n}$ puede descomponerse como:

$$
A = U \Sigma V^T
$$

donde:

- $U \in \mathbb{R}^{m \times m}$ es **ortogonal**, con $U^T U = I_m$  
- $V \in \mathbb{R}^{n \times n}$ es **ortogonal**, con $V^T V = I_n$  
- $\Sigma \in \mathbb{R}^{m \times n}$ es **diagonal**, con valores singulares  
  $\sigma_1 \geq \sigma_2 \geq \cdots \geq \sigma_r > 0$

---

#### **2.2.2 Pseudoinversa y Solución por Mínimos Cuadrados**

La **pseudoinversa de Moore–Penrose** se define como:

$$
A^+ = V \Sigma^+ U^T
$$

donde $\Sigma^+$ es la pseudoinversa de $\Sigma$, con elementos:

$$
\Sigma_{ii}^+ =
\begin{cases}
1 / \sigma_i, & \text{si } \sigma_i > 0 \\
0, & \text{en otro caso}
\end{cases}
$$

La solución más estable (numéricamente robusta) del problema de mínimos cuadrados es:

$$
\theta = A^+ y = V \Sigma^+ U^T y
$$

Esta formulación evita problemas de inversión directa y es la base para métodos computacionales modernos de regresión y clasificación.

---

### **2.3 Interpretación Geométrica**

El clasificador por mínimos cuadrados define un **hiperplano de decisión** en el espacio de características:

$$
\{ x \in \mathbb{R}^n : x^T \beta + v = 0 \}
$$

donde:
- $\beta$ es el vector **normal al hiperplano**,  
- $v$ es el término independiente o **sesgo**.

La regla de clasificación se convierte en:

$$
\hat{y} =
\begin{cases}
+1, & \text{si } \beta^T x + v > 0 \\
-1, & \text{si } \beta^T x + v < 0
\end{cases}
$$

Geométricamente, el modelo busca el hiperplano que **minimiza la suma de los cuadrados de las distancias verticales** entre los puntos de datos y dicho hiperplano.

---

## **3. Implementación del Clasificador Binario**

En el caso de estudio del *Capítulo 14.2.2* (Boyd & Vandenberghe, 2018), se aplica el clasificador de mínimos cuadrados al conjunto **MNIST**, para distinguir el **dígito 0** de los demás (1–9).

El procedimiento general incluye:

1. **Construcción de la matriz de diseño** $A$, utilizando las intensidades de píxeles relevantes como características.  
2. **Cálculo del vector de parámetros** mediante la ecuación normal:  
   $$
   \theta = (A^T A)^{-1} A^T y
   $$
3. **Predicción** de nuevas muestras aplicando:  
   $$
   \hat{y} = \operatorname{sign}(A_{\text{test}} \theta)
   $$
4. **Evaluación** de desempeño usando una matriz de confusión que refleja tasas de error, verdaderos positivos y falsos positivos.

```python

# ## 3. Implementación del Clasificador Binario
# ### 3.1 Configuración Inicial y Carga de Datos

# %%
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

# Configuración para reproducibilidad
np.random.seed(42)

# %%
# Cargar conjunto de datos MNIST
print("=== CARGANDO CONJUNTO DE DATOS MNIST ===")
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X, y = mnist.data, mnist.target.astype(int)

# Normalización de datos
X = X / 255.0

print(f"Dimensión de características: {X.shape}")
print(f"Dimensión de etiquetas: {y.shape}")
print(f"Distribución de clases: {np.bincount(y)}")

# %%
# Preparación para clasificación binaria (dígito 0 vs otros)
y_binary = np.where(y == 0, 1, -1)

# División del conjunto de datos
X_train, X_test, y_train, y_test = train_test_split(
    X, y_binary, test_size=10000, random_state=42, stratify=y_binary
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train, test_size=10000, random_state=42, stratify=y_train
)

print(f"\n=== DIVISIÓN DE DATOS ===")
print(f"Entrenamiento: {X_train.shape[0]} muestras")
print(f"Validación: {X_val.shape[0]} muestras")
print(f"Prueba: {X_test.shape[0]} muestras")
```

3.2 Implementación del Clasificador con SVD

```python

# ### 3.2 Implementación del Clasificador con SVD

class LeastSquaresClassifier:
    """
    Clasificador por Mínimos Cuadrados usando descomposición SVD
    para estabilidad numérica
    """
    def __init__(self):
        self.weights = None
        self.bias = None
        self.singular_values = None

    def fit(self, X, y):
        """Ajusta el modelo usando descomposición SVD"""
        # Añadir término de bias (columna de unos)
        X_aug = np.column_stack([np.ones(X.shape[0]), X])

        # Descomposición SVD para solución estable
        U, s, Vt = np.linalg.svd(X_aug, full_matrices=False)
        self.singular_values = s

        # Calcular pseudoinversa
        s_inv = np.diag(np.where(s > 1e-10, 1.0/s, 0.0))
        X_pinv = Vt.T @ s_inv @ U.T

        # Calcular parámetros óptimos
        params = X_pinv @ y
        self.bias = params[0]
        self.weights = params[1:]

        return self

    def predict_continuous(self, X):
        """Predicción continua (valores reales)"""
        return self.bias + X @ self.weights

    def predict(self, X):
        """Predicción discreta (etiquetas)"""
        continuous_pred = self.predict_continuous(X)
        return np.where(continuous_pred >= 0, 1, -1)

    def get_condition_number(self):
        """Número de condición de la matriz de diseño"""
        if self.singular_values is None:
            return None
        return self.singular_values[0] / self.singular_values[-1]

# %%
# Entrenamiento del clasificador binario
print("=== ENTRENAMIENTO DEL CLASIFICADOR BINARIO ===")
binary_classifier = LeastSquaresClassifier()
binary_classifier.fit(X_train, y_train)

# Número de condición
cond_number = binary_classifier.get_condition_number()
print(f"Número de condición de la matriz: {cond_number:.2f}")
```

3.3 Evaluación del Modelo Binario

```python
# ### 3.3 Evaluación del Modelo Binario

# %%
# Predicciones y evaluación
y_train_pred = binary_classifier.predict(X_train)
y_test_pred = binary_classifier.predict(X_test)

train_accuracy = np.mean(y_train_pred == y_train)
test_accuracy = np.mean(y_test_pred == y_test)

print("=== RESULTADOS CLASIFICACIÓN BINARIA ===")
print(f"Precisión en entrenamiento: {train_accuracy:.4f}")
print(f"Precisión en prueba: {test_accuracy:.4f}")

# %%
# Función para visualizar matriz de confusión
def plot_confusion_matrix(y_true, y_pred, title, ax=None):
    cm = confusion_matrix(y_true, y_pred)

    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 6))

    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Otros', 'Cero'],
                yticklabels=['Otros', 'Cero'], ax=ax)
    ax.set_title(title)
    ax.set_ylabel('Etiqueta Real')
    ax.set_xlabel('Predicción')

    # Métricas detalladas
    tn, fp, fn, tp = cm.ravel()
    tpr = tp / (tp + fn)  # True Positive Rate
    fpr = fp / (fp + tn)  # False Positive Rate
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0

    print(f"Verdaderos Positivos (TP): {tp}")
    print(f"Falsos Positivos (FP): {fp}")
    print(f"Falsos Negativos (FN): {fn}")
    print(f"Verdaderos Negativos (TN): {tn}")
    print(f"Tasa de Verdaderos Positivos: {tpr:.4f}")
    print(f"Tasa de Falsos Positivos: {fpr:.4f}")
    print(f"Precisión: {precision:.4f}")
    print("-" * 50)

    return cm

# %%
# Visualización de matrices de confusión
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

print("=== MATRIZ DE CONFUSIÓN - ENTRENAMIENTO ===")
cm_train = plot_confusion_matrix(y_train, y_train_pred,
                               "Matriz de Confusión - Entrenamiento", ax1)

print("=== MATRIZ DE CONFUSIÓN - PRUEBA ===")
cm_test = plot_confusion_matrix(y_test, y_test_pred,
                              "Matriz de Confusión - Prueba", ax2)

plt.tight_layout()
plt.show()
```

```python
# ## 4. Análisis de Distribuciones y Visualización
# ### 4.1 Distribución de Predicciones Continuas

# %%
def plot_continuous_distributions(classifier, X_train, y_train, X_test, y_test):
    """Visualiza distribuciones de predicciones continuas"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # Distribución en entrenamiento
    train_continuous = classifier.predict_continuous(X_train)
    ax1.hist(train_continuous[y_train == -1], bins=50, alpha=0.7,
             label='Otros dígitos', color='red', density=True)
    ax1.hist(train_continuous[y_train == 1], bins=50, alpha=0.7,
             label='Dígito 0', color='blue', density=True)
    ax1.axvline(x=0, color='black', linestyle='--', label='Umbral')
    ax1.set_xlabel('Valor de $\~{f}(x)$')
    ax1.set_ylabel('Densidad')
    ax1.set_title('Distribución - Entrenamiento')
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    # Distribución en prueba
    test_continuous = classifier.predict_continuous(X_test)
    ax2.hist(test_continuous[y_test == -1], bins=50, alpha=0.7,
             label='Otros dígitos', color='red', density=True)
    ax2.hist(test_continuous[y_test == 1], bins=50, alpha=0.7,
             label='Dígito 0', color='blue', density=True)
    ax2.axvline(x=0, color='black', linestyle='--', label='Umbral')
    ax2.set_xlabel('Valor de $\~{f}(x)$')
    ax2.set_ylabel('Densidad')
    ax2.set_title('Distribución - Prueba')
    ax2.legend()
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

# %%
plot_continuous_distributions(binary_classifier, X_train, y_train, X_test, y_test)
```

5. Implementación del Clasificador Multi-Clase

5.1 Estrategia One-vs-All

```python
# ## 5. Implementación del Clasificador Multi-Clase
# ### 5.1 Estrategia One-vs-All

# %%
class MultiClassLSClassifier:
    """
    Clasificador Multi-Clase usando estrategia One-vs-All
    """
    def __init__(self):
        self.classifiers = []
        self.n_classes = 0

    def fit(self, X, y):
        """Entrena clasificadores one-vs-all para cada clase"""
        self.n_classes = len(np.unique(y))
        self.classifiers = []

        print(f"Entrenando {self.n_classes} clasificadores one-vs-all...")

        for class_label in range(self.n_classes):
            # Crear etiquetas binarias para esta clase
            y_binary = np.where(y == class_label, 1, -1)

            # Entrenar clasificador binario
            classifier = LeastSquaresClassifier()
            classifier.fit(X, y_binary)
            self.classifiers.append(classifier)

            if (class_label + 1) % 2 == 0:
                print(f"Completado: {class_label + 1}/{self.n_classes}")

        return self

    def predict(self, X):
        """Predice usando la regla argmax"""
        n_samples = X.shape[0]
        scores = np.zeros((n_samples, self.n_classes))

        # Obtener scores de cada clasificador
        for i, classifier in enumerate(self.classifiers):
            scores[:, i] = classifier.predict_continuous(X)

        # Regla argmax para decisión final
        return np.argmax(scores, axis=1)

    def predict_proba(self, X):
        """Probabilidades softmax (aproximadas)"""
        n_samples = X.shape[0]
        scores = np.zeros((n_samples, self.n_classes))

        for i, classifier in enumerate(self.classifiers):
            scores[:, i] = classifier.predict_continuous(X)

        # Aplicar softmax para obtener probabilidades
        exp_scores = np.exp(scores - np.max(scores, axis=1, keepdims=True))
        return exp_scores / np.sum(exp_scores, axis=1, keepdims=True)

# %%
# Preparar datos para clasificación multi-clase
y_multi = y

# Dividir datos multi-clase
X_train_multi, X_test_multi, y_train_multi, y_test_multi = train_test_split(
    X, y_multi, test_size=10000, random_state=42, stratify=y_multi
)

print("=== DATOS MULTI-CLASE ===")
print(f"Entrenamiento: {X_train_multi.shape[0]} muestras")
print(f"Prueba: {X_test_multi.shape[0]} muestras")
print(f"Distribución en entrenamiento: {np.bincount(y_train_multi)}")
print(f"Distribución en prueba: {np.bincount(y_test_multi)}")

# %%
# Entrenar clasificador multi-clase
print("\n=== ENTRENAMIENTO MULTI-CLASE ===")
multi_classifier = MultiClassLSClassifier()
multi_classifier.fit(X_train_multi, y_train_multi)
```

5.2 Evaluación Multi-Clase

```python
# ### 5.2 Evaluación Multi-Clase

# %%
# Predicciones multi-clase
y_train_multi_pred = multi_classifier.predict(X_train_multi)
y_test_multi_pred = multi_classifier.predict(X_test_multi)

train_accuracy_multi = np.mean(y_train_multi_pred == y_train_multi)
test_accuracy_multi = np.mean(y_test_multi_pred == y_test_multi)

print("=== RESULTADOS CLASIFICACIÓN MULTI-CLASE ===")
print(f"Precisión en entrenamiento: {train_accuracy_multi:.4f}")
print(f"Precisión en prueba: {test_accuracy_multi:.4f}")

# %%
# Matriz de confusión multi-clase
def plot_multi_confusion_matrix(y_true, y_pred, title):
    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=range(10), yticklabels=range(10))
    plt.title(title)
    plt.ylabel('Etiqueta Real')
    plt.xlabel('Predicción')
    plt.show()

    # Reporte de clasificación
    print(classification_report(y_true, y_pred, digits=4))

# %%
print("=== MATRIZ DE CONFUSIÓN MULTI-CLASE - ENTRENAMIENTO ===")
plot_multi_confusion_matrix(y_train_multi, y_train_multi_pred,
                           "Matriz de Confusión Multi-Clase - Entrenamiento")

print("=== MATRIZ DE CONFUSIÓN MULTI-CLASE - PRUEBA ===")
plot_multi_confusion_matrix(y_test_multi, y_test_multi_pred,
                           "Matriz de Confusión Multi-Clase - Prueba")
```

6.2 Análisis de Ejemplos de Clasificación

```python
# ### 6.2 Análisis de Ejemplos de Clasificación

# %%
def analyze_misclassifications(X, y_true, y_pred, n_examples=10):
    """Analiza ejemplos de clasificaciones incorrectas"""
    misclassified_idx = np.where(y_true != y_pred)[0]

    if len(misclassified_idx) == 0:
        print("¡No hay clasificaciones incorrectas!")
        return

    print(f"Total de clasificaciones incorrectas: {len(misclassified_idx)}")
    print(f"Tasa de error: {len(misclassified_idx)/len(y_true):.4f}")

    # Mostrar ejemplos
    n_display = min(n_examples, len(misclassified_idx))
    fig, axes = plt.subplots(2, 5, figsize=(15, 6))
    axes = axes.ravel()

    for i, idx in enumerate(misclassified_idx[:n_display]):
        axes[i].imshow(X[idx].reshape(28, 28), cmap='gray')
        axes[i].set_title(f'Real: {y_true[idx]}, Pred: {y_pred[idx]}')
        axes[i].axis('off')

    # Ocultar ejes sobrantes
    for i in range(n_display, 10):
        axes[i].axis('off')

    plt.suptitle('Ejemplos de Clasificaciones Incorrectas', fontsize=16)
    plt.tight_layout()
    plt.show()

    # Análisis de patrones de error
    error_matrix = confusion_matrix(y_true, y_pred)
    np.fill_diagonal(error_matrix, 0)  # Eliminar correctos

    plt.figure(figsize=(10, 8))
    sns.heatmap(error_matrix, annot=True, fmt='d', cmap='Reds')
    plt.title('Patrones de Error (Excluyendo Correctos)')
    plt.ylabel('Etiqueta Real')
    plt.xlabel('Predicción')
    plt.show()

# %%
analyze_misclassifications(X_test_multi, y_test_multi, y_test_multi_pred)
```

###Clasificador polinomial con una característica
En este ejercicio se busca construir clasificadores polinomiales de distintos grados para un conjunto de datos generado artificialmente.  
El objetivo es observar cómo el $\textbf{grado del polinomio}$ afecta la capacidad de clasificar correctamente los datos y la $\textbf{tasa de error}$.

**1. Generación del conjunto de datos**

Se generan 200 puntos $x^{(i)}$ uniformemente distribuidos en el intervalo $[-1,1]$, y a cada punto se le asigna una etiqueta $y^{(i)}$ según la regla:

$$
y^{(i)} =
\begin{cases}
+1, & \text{si } -0.5 \le x^{(i)} < 0.1 \text{ o } x^{(i)} \ge 0.5, \\
-1, & \text{en otro caso.}
\end{cases}
$$

Esto genera dos regiones positivas ($y=+1$) separadas por regiones negativas ($y=-1$).  

$\textbf{Código para generar datos y definir función de ajuste:}$


```python
import numpy as np
import matplotlib.pyplot as plt

# 200 puntos uniformes
x = np.linspace(-1, 1, 200)
# Etiquetas según el enunciado
y = np.where(((x >= -0.5) & (x < 0.1)) | (x >= 0.5), 1, -1)

# Visualizamos los datos
plt.figure(figsize=(8,4))
plt.scatter(x, y, color='gray', s=20)
plt.title("Datos generados: +1 y -1")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

- Se generan los 200 puntos y sus etiquetas.  
- La gráfica muestra las regiones positivas ($y=+1$) y negativas ($y=-1$).

**2. Ajuste de polinomios de grado 0 a 8**

Usaremos $\texttt{numpy.polyfit}$ para ajustar polinomios de distintos grados y luego clasificaremos usando el signo del polinomio.

$\textbf{Código:}$

```python
degrees = range(0, 9)
errors = []

plt.figure(figsize=(15,20))
for i, deg in enumerate(degrees):
    # Ajuste polinomial
    coef = np.polyfit(x, y, deg)
    f_tilde = np.polyval(coef, x)
    # Clasificación binaria
    y_hat = np.sign(f_tilde)
    # Error de entrenamiento
    error = np.mean(y_hat != y)
    errors.append(error)

    # Gráfica de polinomio y clasificación
    plt.subplot(5, 2, i+1)
    plt.scatter(x, y, color='gray', s=15, label="Datos reales")
    plt.plot(x, f_tilde, 'b', label=r"$\tilde{f}(x)$ (polinomio)")
    plt.plot(x, y_hat, 'r--', label=r"$\hat{f}(x)$ (signo)")
    plt.title(f"Grado {deg} - Error {error:.3f}")
    plt.ylim(-2,2)
    plt.legend()

plt.tight_layout()
plt.show()
```

$\textbf{Explicación de las gráficas:}$
- Grados 0 y 1: polinomios muy simples → no capturan los cambios → error alto.  
- Grado 3: suficiente para capturar los tres cambios de signo.  
- Grados mayores: pueden sobreajustar, generando oscilaciones innecesarias.

**3. Evolución del error por grado**

$\textbf{Código:}$

```python
plt.figure(figsize=(8,5))
plt.plot(degrees, errors, 'o-', color='purple', linewidth=2)
for i,e in enumerate(errors):
    plt.text(degrees[i], e+0.01, f"{e:.3f}", ha='center')
plt.xlabel("Grado del polinomio")
plt.ylabel("Error de entrenamiento")
plt.title("Evolución del error vs grado")
plt.grid(True)
plt.show()

# Tabla de errores
print("Grado | Error")
for deg, e in zip(degrees, errors):
    print(f"{deg:>5} | {e:.3f}")
```

$\textbf{Explicación:}$
- La gráfica confirma que el error disminuye al aumentar el grado.  
- El grado 3 es el mínimo necesario para clasificar correctamente la mayoría de los datos.

**4. Clasificador cúbico perfecto**

El enunciado propone un cúbico exacto que clasifica perfectamente los datos:

$$
\tilde{f}(x) = c(x + 0.5)(x - 0.1)(x - 0.5), \quad c>0
$$

$\textbf{Código:}$

```python
c = 1.0
f_cubic = c * (x + 0.5)*(x - 0.1)*(x - 0.5)
y_cubic = np.sign(f_cubic)

plt.figure(figsize=(8,4))
plt.scatter(x, y, color='gray', s=15, label="Datos reales")
plt.plot(x, f_cubic, 'b', label=r"Cúbico propuesto $\tilde{f}(x)$")
plt.plot(x, y_cubic, 'r--', label=r"$\hat{f}(x)$ (signo)")
plt.title("Clasificador cúbico perfecto")
plt.ylim(-2,2)
plt.legend()
plt.grid(True)
plt.show()

print("Error cúbico propuesto:", np.mean(y_cubic != y))
```

$\textbf{Explicación:}$
- La curva azul es el polinomio propuesto, la roja el clasificador.  
- Todas las transiciones coinciden con los cambios de clase → error = 0.

**5. Comparación cúbico LS vs cúbico propuesto**

Comparamos el cúbico ajustado por mínimos cuadrados con el cúbico perfecto.

$\textbf{Código:}$

```python
# Ajuste cúbico LS
coef_ls = np.polyfit(x, y, 3)
f_ls = np.polyval(coef_ls, x)
y_ls = np.sign(f_ls)

plt.figure(figsize=(10,5))
plt.scatter(x, y, c=y, cmap='bwr', alpha=0.6, label="Datos reales")
plt.plot(x, f_cubic, 'g-', linewidth=2, label=r"Cúbico propuesto")
plt.plot(x, f_ls, 'k--', linewidth=2, label=r"Cúbico LS")
plt.axhline(0, color='gray', linestyle='--')
plt.title("Comparación cúbico propuesto vs cúbico LS")
plt.xlabel("x")
plt.ylabel(r"$\tilde{f}(x)$")
plt.legend()
plt.grid(True)
plt.show()

error_ls = np.mean(y_ls != y)
print("Error cúbico LS:", error_ls)
```

$\textbf{Interpretación:}$

#Ejercicio 14.7

##Clasificador polinómico con dos características

En este ejercicio se aplica el método de clasificación por mínimos cuadrados
a un conjunto de datos sintético generado en el plano

En primer lugar se generan $200$ puntos aleatorios

$$
x^{(i)} = \begin{bmatrix} x_1^{(i)} \\ x_2^{(i)} \end{bmatrix}
$$

$$
i = 1, \dots, 200
$$

a partir de una distribución normal estándar en dos dimensiones:

$$
x^{(i)} \sim \mathcal{N}(0, I_2)
$$

Las etiquetas se definen como:

$$
y^{(i)} =
\begin{cases}
+1, & \text{si } x_1^{(i)}x_2^{(i)} \ge 0 \\
-1, & \text{si } x_1^{(i)}x_2^{(i)} < 0
\end{cases}
$$

Esto significa:
- Los puntos en el primer y tercer cuadrante tienen $y = +1$
- Los puntos en el segundo y cuarto cuadrante tienen $y = -1$

En este caso el clasificador ideal se puede expresar como:

$$
f(x) = x_1 x_2
$$

ya que su frontera de decisión $f(x)=0$ corresponde exactamente a los ejes coordenados, permitiendo así separar los cuadrantes con error cero

Lo que se busca es ajustar un clasificador de segundo grado mediante mínimos cuadrados de la forma:

$$
\tilde{f}(x) = \theta_1 + \theta_2 x_1 + \theta_3 x_2 + \theta_4 x_1^2 + \theta_5 x_1x_2 + \theta_6 x_2^2
$$

donde $\theta = [\theta_1, \theta_2, \theta_3, \theta_4, \theta_5, \theta_6]^T$
son los parámetros a estimar

Con esto procedemos a construir la matriz de diseño $\Phi \in \mathbb{R}^{N \times 6}$:

$$
\Phi =
\begin{bmatrix}
1 & x_1^{(1)} & x_2^{(1)} & (x_1^{(1)})^2 & x_1^{(1)}x_2^{(1)} & (x_2^{(1)})^2 \\
1 & x_1^{(2)} & x_2^{(2)} & (x_1^{(2)})^2 & x_1^{(2)}x_2^{(2)} & (x_2^{(2)})^2 \\
\vdots & \vdots & \vdots & \vdots & \vdots & \vdots \\
1 & x_1^{(N)} & x_2^{(N)} & (x_1^{(N)})^2 & x_1^{(N)}x_2^{(N)} & (x_2^{(N)})^2
\end{bmatrix}
$$

Con el vector de etiquetas:

$$
y = [y^{(1)}, y^{(2)}, \dots, y^{(N)}]^T
$$

Donde el objetivo de mínimos cuadrados es minimizar:

$$
J(\theta) = \|\Phi \theta - y\|^2
$$


La solución de este clasificador se obtiene mediante las ecuaciones normales:

$$
\Phi^\top \Phi \, \hat{\theta} = \Phi^\top y
$$

de donde:

$$
\hat{\theta} = (\Phi^\top \Phi)^{-1} \Phi^\top y
$$

El vector $\hat{\theta}$ contiene los coeficientes del clasificador polinómico ajustado

Posteriormente, el clasificados predice según el signo de los parámetros obtenidos:

$$
\hat{f}(x) = \operatorname{sign}(\tilde{f}(x)) =
\begin{cases}
+1, & \tilde{f}(x) \ge 0 \\
-1, & \tilde{f}(x) < 0
\end{cases}
$$

Y para obtener el porcentaje de error sobre los datos generados se calcula de la siguiente manera:

$$
\text{Error} =
\frac{1}{N} \sum_{i=1}^{N} \mathbf{1}[\hat{f}(x^{(i)}) \neq y^{(i)}]
$$

donde $\mathbf{1}[\cdot]$ vale 1 si la predicción es incorrecta, y 0 en caso contrario

Ahora bien, para poder visualizar la superficie $\tilde{f}(x)$ de forma geométrica, se definen regiones en el plano de tal forma que:
- $\tilde{f}(x) > 0$ → clase +1  
- $\tilde{f}(x) < 0$ → clase -1  
- $\tilde{f}(x) = 0$ → frontera de decisión  

El gráfico obtenido se interpreta tal que las curvas de nivel donde $\tilde{f}(x)=0$ se aproximan a los ejes $x_1=0$ y $x_2=0$, reproduciendo el patrón cuadrante de las clases


El código Python permite obtener un acercamiento numérico a lo que sería la resolución de un problema de este tipo, en el cual se pueden obtener los coeficientes estimados, la tasa de error y un gráfico por cuadrantes que incluye los puntos generados aleatoriamente y clasificados para cada caso


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Datos
np.random.seed(0)
N = 200
X = np.random.randn(N, 2)
y = np.sign(X[:,0] * X[:,1])
y[y == 0] = 1  # evitar ceros exactos

# 2. Matriz de diseño
Phi = np.column_stack([
    np.ones(N),
    X[:,0], X[:,1],
    X[:,0]**2,
    X[:,0]*X[:,1],
    X[:,1]**2
])

# 3. Ajuste por mínimos cuadrados
theta = np.linalg.inv(Phi.T @ Phi) @ Phi.T @ y

# 4. Predicción y tasa de error
y_pred = np.sign(Phi @ theta)
error = np.mean(y_pred != y)

print("Coeficientes estimados:", theta)
print("Tasa de error:", round(error*100, 2), "%")

# 5. Visualización
x1 = np.linspace(-3,3,200)
x2 = np.linspace(-3,3,200)
X1, X2 = np.meshgrid(x1, x2)

Phi_grid = np.column_stack([
    np.ones(X1.size),
    X1.ravel(), X2.ravel(),
    X1.ravel()**2, X1.ravel()*X2.ravel(), X2.ravel()**2
])

fhat = (Phi_grid @ theta).reshape(X1.shape)

plt.figure(figsize=(6,5))
plt.contour(X1, X2, fhat, levels=[0], colors='k', linewidths=2)
plt.scatter(X[y==1,0], X[y==1,1], c='cornflowerblue', label='y=+1')
plt.scatter(X[y==-1,0], X[y==-1,1], c='salmon', label='y=-1')
plt.legend()
plt.title("Clasificador polinómico (grado 2)")
plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.show()
```
