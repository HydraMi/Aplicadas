### **¿Pertenece $\mathbf b$ al espacio columna $C(A)$?**

El espacio columna $C(A)$ está formado por todas las combinaciones lineales de las columnas de $A$:

$$
C(A)=\{\mathbf{y}\in\mathbb{R}^3:\mathbf{y}=A\mathbf{x}\ \text{para algún }\mathbf{x}\in\mathbb{R}^4\}.

###Comprobacion con codigo Python :

```python
import numpy as np

# Definir matriz A y vector b
A = np.array([[2, -1, 1, 3],
              [-1, 2, 1, -3],
              [0, -1, -1, 1]])

b = np.array([2, 0, 2])

# Rangos
rango_A = np.linalg.matrix_rank(A)
rango_Aumentada = np.linalg.matrix_rank(np.column_stack((A, b)))
num_incognitas = A.shape[1]

print(f"Rango de A: {rango_A}")
print(f"Rango de [A|b]: {rango_Aumentada}")
print(f"Número de incógnitas: {num_incognitas}")

if rango_A != rango_Aumentada:
    print("Sistema INCONSISTENTE - No tiene solución")
else:
    if rango_A == num_incognitas:
        print("Sistema CONSISTENTE - Solución ÚNICA")
    else:
        print("Sistema CONSISTENTE - Infinitas soluciones")
```



### ¿Pertenece $\mathbf b$ al espacio fila $C(A^T)$?

**Definición del espacio fila**

El espacio fila $C(A^T)$ está formado por todas las combinaciones lineales de las filas de la matriz $A$.

Matemáticamente:
$$C(A^T) = \{\mathbf{y} \in \mathbb{R}^n : \mathbf{y} = A^T\mathbf{x} \text{ para algún } \mathbf{x} \in \mathbb{R}^m\}$$

**Análisis dimensional**

- La matriz $A$ tiene dimensiones $3 \times 4$
- Sus filas son vectores en $\mathbb{R}^4$ (porque hay 4 columnas)
- Por lo tanto: $C(A^T) \subset \mathbb{R}^4$
- El vector $\mathbf{b} = \begin{bmatrix} 2\\ 0\\ 2 \end{bmatrix} \in \mathbb{R}^3$

**Conclusión por argumento dimensional**

Un vector de $\mathbb{R}^3$ no puede pertenecer a un subespacio de $\mathbb{R}^4$ porque están en espacios vectoriales diferentes.

$$\mathbf{b} \in \mathbb{R}^3 \quad \text{y} \quad C(A^T) \subset \mathbb{R}^4 \quad \Rightarrow \quad \mathbf{b} \notin C(A^T)$$

### ¿Pertenece $\mathbf b$ al espacio nulo $N(A^T)$?


**Definición del espacio nulo izquierdo**

$$N(A^T) = \{\mathbf{y} \in \mathbb{R}^3 : A^T\mathbf{y} = \mathbf{0}\}$$

**Cálculo de $A^T$**

$$A^T = \begin{bmatrix}
2 & -1 & 0 \\
-1 & 2 & -1 \\
1 & 1 & -1 \\
3 & -3 & 1
\end{bmatrix}$$

**Verificación de la condición $A^T\mathbf{b} = \mathbf{0}$**

Calculamos el producto matriz-vector:

$$A^T\mathbf{b} = \begin{bmatrix}
2 & -1 & 0 \\
-1 & 2 & -1 \\
1 & 1 & -1 \\
3 & -3 & 1
\end{bmatrix}
\begin{bmatrix}
2\\ 0\\ 2
\end{bmatrix}$$

**Cálculo detallado componente por componente**

- **Primera componente:** $2\cdot2 + (-1)\cdot0 + 0\cdot2 = 4$
- **Segunda componente:** $(-1)\cdot2 + 2\cdot0 + (-1)\cdot2 = -2 - 2 = -4$
- **Tercera componente:** $1\cdot2 + 1\cdot0 + (-1)\cdot2 = 2 - 2 = 0$
- **Cuarta componente:** $3\cdot2 + (-3)\cdot0 + 1\cdot2 = 6 + 2 = 8$

**Resultado:**
$$A^T\mathbf{b} = \begin{bmatrix} 4 \\ -4 \\ 0 \\ 8 \end{bmatrix} \neq \begin{bmatrix} 0 \\ 0 \\ 0 \\ 0 \end{bmatrix}$$

**Conclusión**

Para que $\mathbf{b} \in N(A^T)$, debe cumplirse que $A^T\mathbf{b} = \mathbf{0}$. Como esto no se cumple:

$$\mathbf{b} \notin N(A^T)$$

### Espacios representables en $\mathbb{R}^3$

Los dos espacios fundamentales que pueden representarse geométricamente en $\mathbb{R}^3$ son:

1. **$C(A)$ - Espacio columna**: Subespacio de $\mathbb{R}^3$ generado por las columnas de $A$
2. **$N(A^T)$ - Espacio nulo izquierdo**: Subespacio de $\mathbb{R}^3$ formado por vectores ortogonales a todas las filas de $A$

### Base y ecuaciones de cada subespacio

#### Para $C(A)$ (Plano en $\mathbb{R}^3$)

**Vectores generadores:**
$$\mathbf{v}_1 = \begin{bmatrix} 2\\ -1\\ 0 \end{bmatrix}, \quad \mathbf{v}_2 = \begin{bmatrix} -1\\ 2\\ -1 \end{bmatrix}$$

**Ecuación del plano:** Para encontrar la ecuación del plano, calculamos el vector normal mediante producto vectorial:

$$\mathbf{n} = \mathbf{v}_1 \times \mathbf{v}_2 = \begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\
2 & -1 & 0 \\
-1 & 2 & -1
\end{vmatrix} = \begin{bmatrix} 1\\ 2\\ 3 \end{bmatrix}$$

**Ecuación del plano $C(A)$:**
$$x + 2y + 3z = 0$$

#### Para $N(A^T)$ (Recta en $\mathbb{R}^3$)

**Vector generador:**
$$\mathbf{w} = \begin{bmatrix} 1\\ 2\\ 3 \end{bmatrix}$$

**Ecuaciones paramétricas de la recta $N(A^T)$:**
$$\begin{cases}
x = t \\
y = 2t \\
z = 3t
\end{cases}, \quad t \in \mathbb{R}$$

### Relación geométrica entre los espacios

**Teorema Fundamental del Álgebra Lineal:** $C(A) \perp N(A^T)$

**Verificación:**
- $\mathbf{n} \cdot \mathbf{v}_1 = 1\cdot2 + 2\cdot(-1) + 3\cdot0 = 0$
- $\mathbf{n} \cdot \mathbf{v}_2 = 1\cdot(-1) + 2\cdot2 + 3\cdot(-1) = 0$
- $\mathbf{n} = \mathbf{w}$

Esto confirma que $N(A^T)$ es perpendicular a $C(A)$.


### Interpretación geométrica

- **$C(A)$** es un **plano** que pasa por el origen con vector normal $\mathbf{w} = (1,2,3)$
- **$N(A^T)$** es una **recta** que pasa por el origen en la dirección del vector $\mathbf{w} = (1,2,3)$
- La recta $N(A^T)$ es **perpendicular** al plano $C(A)$
- Esta relación ilustra el **Teorema Fundamental del Álgebra Lineal**: el espacio nulo izquierdo es el complemento ortogonal del espacio columna

### Verificación dimensional

$$\dim C(A) + \dim N(A^T) = 2 + 1 = 3 = \dim \mathbb{R}^3$$

Esto muestra que $C(A)$ y $N(A^T)$ son subespacios complementarios ortogonales en $\mathbb{R}^3$.

Al aplicar Gram–Schmidt a las 4 columnas, cuando una columna es combinación de las anteriores va a aparecer el vector cero y este es un vector que no se puede normalizar, por lo que todo este proceso solo produce $r=\operatorname{rank}(A)$ vectores.

### b) Base ortonormal de $\mathbb{R}^3$ que incluya vectores de $\mathcal C(A)$ y de $N(A^T)$

La matriz es
$$
A=\begin{bmatrix}
2&-1&1&3\\
-1&2&1&-3\\
0&-1&-1&1
\end{bmatrix},\qquad \operatorname{rank}(A)=2.
$$

$\quad$

Tenemos los siguientes puntos a destacar:

$\quad$

**Puntos a tener en cuenta:**

- En $\mathbb{R}^3$ todo vector se puede escribir como parte en el espacio columna de $A$: $\mathcal{C}(A)$ más parte en el espacio nulo de $A^T$: $N(A^T)$.

- Dicho de otra forma:
  $
  \mathbb{R}^3 = \mathcal{C}(A) + N(A^T)
  $

- Además, esos dos espacios son perpendiculares entre sí:
  $
  \mathcal{C}(A) \perp N(A^T)
  $

$\quad$

Esto significa que los vectores del espacio columna de \(A\) y los vectores del nulo de \(A^T\) siempre forman ángulo recto.

$\quad$

Además, En $A$, las columnas $c_1=(2,-1,0)^T$ y $c_2=(-1,2,-1)^T$ son independientes, mientras que $c_3=(1,1,-1)^T$ y $c_4=(3,-3,1)^T$ pueden escribirse como combinaciones lineales de $c_1$ y $c_2$. Por lo tanto, $\operatorname{rank}(A)=2$.

$\quad$

**Base ortonormal de $\mathcal C(A)$.**

Tomamos las dos primeras columnas porque on el pivote: $v_1=(2,-1,0)^T$,  $v_2=(-1,2,-1)^T)$ y aplicamos Gram–Schmidt:

$\quad$

$$
q_1=\frac{v_1}{\|v_1\|}=\frac{1}{\sqrt{5}}\,(2,-1,0),
\qquad
w_2=v_2-\frac{\langle v_2,q_1\rangle}{\langle q_1,q_1\rangle}\,q_1=(3,6,-5).
$$

$$
q_2=\frac{w_2}{\|w_2\|}=\frac{1}{\sqrt{70}}\,(3,6,-5).
$$

**Finalmente nos queda:**

$\quad$

es una base ortonormal del espacio vectorial.

```python
import numpy as np
import sympy as sp

def nrm(v):
    return np.linalg.norm(np.array(v, dtype=float))

def fmt(v, prec=6):
    return np.array2string(np.array(v, dtype=float), precision=prec, suppress_small=True)

print("=== (b) Column space C(A) y nullspace N(A^T) ===")
A = np.array([[2,-1, 1, 3],
              [-1,2, 1,-3],
              [0,-1,-1, 1]], dtype=float)

# Rango
rankA = np.linalg.matrix_rank(A)
print("rank(A) =", rankA)

# Tomamos 2 columnas independientes (las dos primeras) y aplicamos Gram–Schmidt
v1 = A[:,0]                  # (2, -1, 0)
v2 = A[:,1]                  # (-1, 2, -1)

q1 = v1 / nrm(v1)
proj_v2_q1 = np.dot(v2, q1) * q1
w2 = v2 - proj_v2_q1
q2 = w2 / nrm(w2)

print("\nBase ortonormal de C(A):")
print("q1 =", fmt(q1))
print("q2 =", fmt(q2))
print("||q1|| =", nrm(q1), " ||q2|| =", nrm(q2))
print("<q1,q2> =", float(np.dot(q1,q2)))

# Nullspace de A^T con Sympy (exacto) y normalización
AT = sp.Matrix(A.T)
N = AT.nullspace()
if not N:
    raise RuntimeError("El nullspace de A^T es vacío (no debería suceder con rank(A)=2).")
n_raw = np.array(N[0], dtype=float).reshape(-1)
n = n_raw / nrm(n_raw)

print("\nVector en N(A^T) (normalizado):")
print("n =", fmt(n))
print("||n|| =", nrm(n))
print("A^T n =", fmt(A.T @ n))
print("<q1,n> =", float(np.dot(q1,n)))
print("<q2,n> =", float(np.dot(q2,n)))

Q = np.column_stack([q1, q2, n])
print("\nQ^T Q (debe ser I3):\n", np.round(Q.T @ Q, 6))

# -------- (c) Espacio vectorial del sistema dado (núcleo) --------
print("\n=== (c) Espacio vectorial (núcleo) del sistema ===")
M = np.array([[ 1, 1, 0, 2],
              [-2,-2, 1,-5],
              [ 1, 1,-1, 3],
              [ 4, 4,-1, 9]], dtype=float)

# Base propuesta del núcleo
b1 = np.array([-1, 1, 0, 0], dtype=float)
b2 = np.array([-2, 0, 1, 1], dtype=float)
print("M @ b1 =", fmt(M @ b1))
print("M @ b2 =", fmt(M @ b2))

# Ortonormalización (Gram–Schmidt)
e1 = b1 / nrm(b1)
b2_perp = b2 - np.dot(b2, e1) * e1
e2 = b2_perp / nrm(b2_perp)

print("\nBase ortonormal del núcleo:")
print("e1 =", fmt(e1), "  ||e1|| =", nrm(e1))
print("e2 =", fmt(e2), "  ||e2|| =", nrm(e2))
print("<e1,e2> =", float(np.dot(e1,e2)))
print("M @ e1 =", fmt(M @ e1))
print("M @ e2 =", fmt(M @ e2))

E = np.column_stack([e1,e2])
print("\nE^T E (debe ser I2):\n", np.round(E.T @ E, 6))

# -------- Valores simbólicos --------
print("\n=== Representación exacta (fracciones) ===")
v1_s = sp.Matrix([2,-1,0])
v2_s = sp.Matrix([-1,2,-1])
q1_s = v1_s / sp.sqrt((v1_s.T*v1_s)[0])
w2_s = v2_s - (v2_s.T*q1_s)[0]*q1_s
q2_s = w2_s / sp.sqrt((w2_s.T*w2_s)[0])
n_s = sp.Matrix([1,2,3]) / sp.sqrt(14)
print("q1 (exacto) =", q1_s)
print("q2 (exacto) =", q2_s)
print("n  (exacto) =", n_s)

e1_s = sp.Matrix([-1,1,0,0]) / sp.sqrt(2)
e2_s = sp.Matrix([-1,-1,1,1]) / 2
print("e1 (exacto) =", e1_s)
print("e2 (exacto) =", e2_s)
```

Suponga que una función $f$ se puede aproximar utilizando un conjunto de funciones $f_j$:

 $$
 f(x) = \sum_{j=0}^m a_jf_j(x),
 $$

 donde las $f_j$ son  funciones predeterminadas de $x$ llamadas  *funciones base*. A la expresión de la derecha se le conoce como **forma lineal**. La ecuación anterior es válida siempre y cuando se tengan suficientes funciones base, algo similar a lo que hemos estudiado en términos de subespacios de $\mathbf{R}^n$. Ahora bien, considere las condiciones de interpolación
$f(t_i) = y_i;i = 1,...,n;$ donde $t_i$ y $y_i$ son un conjunto de datos dados. Formulamos el problema de ajuste de datos :
  $$
 y_i  \approx f(x) = \sum_{j=0}^m c_jf_j(x).
 $$

En este problema lo que buscamos es minimizar
 $$
 \sum_{i=0}^n\left[y_i - \sum_{j=0}^m c_jf_j(x_i)\right]^2,
 $$

 lo cual se traduce en resolver el siguiente sistema de ecuaciones (no necesita probarlo)
 $$
 \frac{\partial S}{\partial c_k} = -2\left[ \sum_{i=0}^n \left[ y_i - \sum_{j=0}^m c_jf_j(x_i) \right] f_k(x_i) \right] = 0, \quad k=0, 1, \ldots m,
 $$

 de donde :
 $$
 \sum_{j=0}^m\left[ \sum_{i=0}^n f_j(x_i)f_k(x_i)\right]c_j = \sum_{i=0}^nf_k(x_i)y_i, \quad k=0, 1, \ldots, m.
 $$
Bajo estas condiciones  se puede formular un sistema de ecuaciones para hallar los coeficientes  $\mathbf{c} = (c_1, c_2, \cdots, c_m)$, como $A\mathbf{c}  = \mathbf{b}$, donde
 $$
 A_{kj} = \sum_{i=0}^n f_j(x_i)f_k(x_i),
 \quad \quad
 b_k = \sum_{i=0}^nf_k(x_i)y_i.
 $$

##a

**Definición (Error en un punto).**
<br>
Dado un conjunto de datos  
$$(x_1,y_1), (x_2,y_2), \ldots, (x_n,y_n),$$  
y un polinomio de ajuste de grado $m$  
$$f(x) = c_0 + c_1x + c_2x^2 + \cdots + c_mx^m,$$  
definimos el **error en el punto $i$** como la diferencia entre el valor real $y_i$ y el valor aproximado $f(x_i)$:  

$$
e_i = y_i - f(x_i).
$$

<br>

**Definición (Función de error total).**
<br>
Dado un conjunto de datos
$$(x_1,y_1), (x_2,y_2), \ldots, (x_n,y_n),$$
y un polinomio de ajuste
$$f(x) = c_0 + c_1x + c_2x^2 + \cdots + c_mx^m,$$
se define la **función de error total** como la suma de los cuadrados de los errores en cada punto:  

$$
S(c_0, \ldots, c_m) = \sum_{i=1}^n \big( y_i - f(x_i) \big)^2.
$$

<br>

**Definición (Forma lineal).**
<br>
Dado un conjunto de funciones predeterminadas de $x$, llamadas **funciones base**,
$$f_0(x), f_1(x), \ldots, f_m(x),$$
una **forma lineal** es una combinación lineal de estas funciones de la forma:

$$
f(x) = c_0 f_0(x) + c_1 f_1(x) + \cdots + c_m f_m(x),
$$

donde los coeficientes $c_0, c_1, \ldots, c_m$ son números reales que se buscan determinar.

<br>

**Definición (Funciones base).**
<br>
Se llama **funciones base** a aquellas funciones predeterminadas $f_j(x)$ que se utilizan
para construir una forma lineal. En el caso del ajuste polinomial, las funciones base son
las potencias de $x$:

$$
f_j(x) = x^j, \quad j = 0,1,2,\ldots,m.
$$


---



Sea la familia de funciones base definida como

$$
f_j(x) = x^j, \quad j = 0,1,2,\ldots,m.
$$

Esta definición de funciones base viene directamente del enunciado, donde se nos indica que
para el caso de ajuste polinomial se utilizan las potencias de $x$.
La razón de esta elección es que un polinomio siempre se construye como
una combinación lineal de potencias de $x$.
Es decir, la forma general de un polinomio de grado $m$ en matemáticas es

$$
f(x) = c_0 + c_1x + c_2x^2 + \cdots + c_mx^m,
$$

lo cual puede escribirse como

$$
f(x) = \sum_{j=0}^m c_j x^j.
$$

De esta manera, las funciones $f_j(x)=x^j$ actúan como los “bloques” que permiten construir
cualquier polinomio de grado $m$, mientras que los coeficientes
$c_0, c_1, \ldots, c_m$ son los valores que necesitamos calcular.

Nuestro objetivo es ajustar este polinomio a un conjunto de datos dados:

$$
(x_1, y_1), (x_2, y_2), \ldots, (x_n, y_n).
$$

En este ajuste, el polinomio rara vez pasa exactamente por todos los puntos.
Por eso, en cada punto aparece una diferencia entre el valor real y el valor aproximado
por el polinomio. A esa diferencia la llamamos "error en el punto $i$":

$$
\text{error en el punto } i = y_i - f(x_i).
$$

Si solo sumamos estos errores, algunos positivos y otros negativos se cancelan.
Por eso los elevamos al cuadrado, así todos cuentan como positivos.

Luego sumamos todos esos errores al cuadrado:

$$
S(c_0,\ldots,c_m) = \sum_{i=1}^n \left( y_i - f(x_i) \right)^2.
$$

La idea es escoger los coeficientes del polinomio de forma que esta suma sea lo más pequeña posible,
es decir, que el polinomio esté lo más cerca posible de los datos.


Al derivar esta expresión con respecto a cada coeficiente $c_k$ y buscar que sea igual a cero, se obtiene el sistema de ecuaciones normales

$$
\sum_{j=0}^m \left( \sum_{i=1}^n f_j(x_i) f_k(x_i) \right)c_j = \sum_{i=1}^n f_k(x_i)y_i,
\quad k=0,1,\ldots,m.
$$

Como en este caso $f_j(x_i) = x_i^j$ y $f_k(x_i) = x_i^k$, el producto que aparece en la suma es

$$
f_j(x_i) \cdot f_k(x_i) = x_i^j \cdot x_i^k = x_i^{j+k}.
$$

De esta manera, los elementos de la matriz $A$ resultan

$$
A_{kj} = \sum_{i=1}^n x_i^{j+k}.
$$

Por otro lado, el lado derecho del sistema se obtiene como

$$
b_k = \sum_{i=1}^n f_k(x_i)y_i = \sum_{i=1}^n x_i^k y_i.
$$

Así, el sistema que debemos resolver se puede escribir como

$$
A \mathbf{c} = \mathbf{b},
$$

donde

$$
\mathbf{c} = \begin{bmatrix} c_0 \\ c_1 \\ \vdots \\ c_m \end{bmatrix},
\quad
\mathbf{b} = \begin{bmatrix} b_0 \\ b_1 \\ \vdots \\ b_m \end{bmatrix}.
$$

La matriz $A$ es de tamaño $(m+1)\times(m+1)$ y sus entradas dependen únicamente de los valores de $x_i$.

Finalmente, se observa que la matriz $A$ es simétrica, ya que

$$
A_{kj} = \sum_{i=1}^n x_i^{j+k} = \sum_{i=1}^n x_i^{k+j} = A_{jk}.
$$

Esto significa que al intercambiar los índices $k$ y $j$ no cambia el valor de $A_{kj}$.

```python
import numpy as np

def gauss_jordan(A, b):
    # Resuelve el sistema Ax=b usando Gauss-Jordan
    A = A.astype(float)
    b = b.astype(float)
    n = len(b)
    # Matriz aumentada
    M = np.hstack([A, b.reshape(-1, 1)])

    for i in range(n):
        # Pivoteo parcial
        max_row = np.argmax(abs(M[i:, i])) + i
        M[[i, max_row]] = M[[max_row, i]]

        # Normalizar fila pivote
        M[i] = M[i] / M[i, i]

        # Eliminar en las demás filas
        for j in range(n):
            if j != i:
                M[j] = M[j] - M[j, i] * M[i]

    return M[:, -1]

def lu_solve(A, b):
    # Resuelve el sistema Ax=b usando descomposición LU
    n = len(b)
    L = np.zeros_like(A, dtype=float)
    U = np.zeros_like(A, dtype=float)

    # Descomposición LU
    for i in range(n):
        for j in range(i, n):
            U[i, j] = A[i, j] - np.dot(L[i, :i], U[:i, j])
        L[i, i] = 1
        for j in range(i + 1, n):
            L[j, i] = (A[j, i] - np.dot(L[j, :i], U[:i, i])) / U[i, i]

    # Sustitución hacia adelante: Ly = b
    y = np.zeros(n)
    for i in range(n):
        y[i] = b[i] - np.dot(L[i, :i], y[:i])

    # Sustitución hacia atrás: Ux = y
    x = np.zeros(n)
    for i in reversed(range(n)):
        x[i] = (y[i] - np.dot(U[i, i+1:], x[i+1:])) / U[i, i]

    return x

def polynomial_fit(x_data, y_data, m):
    m += 1  # grado del polinomio + 1 coeficiente
    x_powers = np.zeros(2 * m)
    b = np.zeros(m)

    # Construcción de sumatorias
    for i in range(2 * m):
        x_powers[i] = np.sum(x_data ** i)
        if i < m:
            b[i] = np.sum((x_data ** i) * y_data)

    # Matriz A
    A = np.zeros((m, m))
    for k in range(m):
        for j in range(m):
            A[k, j] = x_powers[j + k]

    # Resolver con dos algoritmos distintos
    c1 = gauss_jordan(A.copy(), b.copy())  # Solución con Gauss-Jordan
    c2 = lu_solve(A.copy(), b.copy())      # Solución con LU

    return c1, c2, A, b


```

### Justificación:
Utilizamos los métodos de Gauss-Jordan y descomposición LU porque ambos permiten resolver de manera directa el sistema lineal que surge del ajuste polinomial.

En primer lugar, el método de Gauss-Jordan permite entender cómo se transforman las ecuaciones hasta llegar a la solución, lo que lo convierte en una herramienta de aprendizaje y validación inicial.

Al implementarlo manualmente se puede visualizar el efecto de cada operación elemental y comprobar que el sistema efectivamente converge a una solución única.

En el código para la resolución del método Gauss-Jordan se aplican operaciones elementales sobre la matriz aumentada $[A|b]$ hasta transformarla en su forma reducida por filas, con esto se puede obtener directamente la solución del sistema. En la implementación se hace explícito el proceso de eliminación y sustitución, y se muestra paso a paso cómo se llega a los coeficientes del polinomio.


En segundo lugar, la descomposición LU es uno de los métodos más utilizados en la práctica porque reduce la cantidad de operaciones necesarias y resulta más eficiente en problemas grandes. Además su factorización hace posible reutilizar la descomposición cuando se desea resolver el mismo sistema con diferentes vectores de términos independientes $𝑏$

Ambos métodos generan soluciones equivalentes para este problema, porque están resolviendo el mismo sistema de ecuaciones normales.

En el código para la resolución del método LU se factoriza la matriz $A$ como el producto de una matriz triangular inferior $L$ y una matriz triangular superior $U$; con esto el sistema se resuelve en dos etapas mediante sustitución hacia adelante de la forma $(Ly=b)$ y hacia atrás $(Ux=y)$. Con esta implementación es posible obtener una mayor estabilidad y facilidad para reutilizar la factoriación cuando cambian los vectores de términos independientes.

```python
from matplotlib import pyplot as plt

# A continuación encuentra los datos necesarios para hacer la validación de sus resultados
temperaturas = np.array([17.6, 17.2, 17.2, 16.9, 16.5, 16.2, 15.6, 15.0, 14.5, 14.7, 15.1, 15.3, 15.1, 15.3, 15.4, 15.6, 16.0, 15.7, 16.6, 16.8, 16.5, 16.0, 15.6, 14.2, 13.8])
horas = np.arange(25)
plt.plot(horas, temperaturas, 'x')
plt.title("Temperatura")
plt.show()

hours_fine = np.arange(0, 25.1, 0.01)

errores_gj = []
errores_lu = []

# Obtenemos los coeficientes con Gauss-Jordan y LU
for i in range(1, 11):
    plt.title("Ajuste Polinomial de grado {}".format(i))
    c_gj, c_lu, _, _ = polynomial_fit(horas, temperaturas, i)


    # Evaluamos para Gauss-Jordan en los mismos puntos de los datos
    temps_interp_gj = np.zeros(temperaturas.size)
    for j in range(temperaturas.size):
        for k in range(len(c_gj)):
            temps_interp_gj[j] += c_gj[k] * (horas[j] ** k)

    # Evaluamos para Gauss-Jordan en los mismos puntos de los datos
    temps_interp_lu = np.zeros(temperaturas.size)
    for j in range(temperaturas.size):
        for k in range(len(c_lu)):
            temps_interp_lu[j] += c_lu[k] * (horas[j] ** k)


  # Graficamos ambos ajustes contra los datos reales
    plt.plot(horas, temperaturas, 'x', label="Datos reales")
    plt.plot(horas, temps_interp_gj, '-', label="Ajuste GJ")
    plt.plot(horas, temps_interp_lu, '--', label="Ajuste LU")
    plt.legend()
    plt.show()

    # Calculamos el error cuadrático medio (MSE) para cada método
    err_gj = np.mean((temperaturas - temps_interp_gj)**2)
    err_lu = np.mean((temperaturas - temps_interp_lu)**2)

    errores_gj.append(err_gj)
    errores_lu.append(err_lu)

```
