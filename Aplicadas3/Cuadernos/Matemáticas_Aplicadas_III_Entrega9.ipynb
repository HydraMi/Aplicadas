## **Solución Punto 1: Vectores Propios y Matrices Simétricas**

---

### **Parte (a)**

La matriz que se analiza es la siguiente:

$$
A = \begin{bmatrix}
1 & 10^{-15} \\
0 & 1 + 10^{-15}
\end{bmatrix}
$$
$$$$
Para encontrar los valores propios se plantea la ecuación característica $\det(A - \lambda I) = 0$. Al calcular el determinante se obtiene:
$$
\det \begin{bmatrix}
1 - \lambda & 10^{-15} \\
0 & 1 + 10^{-15} - \lambda
\end{bmatrix} = (1 - \lambda)(1 + 10^{-15} - \lambda) = 0
$$
$$$$
De esta ecuación se deduce que los valores propios son:
$$
\lambda_1 = 1, \quad \lambda_2 = 1 + 10^{-15}
$$
$$$$
Para hallar los vectores propios se sustituyen estos valores en la ecuación $(A - \lambda I)\mathbf{v} = 0$.  
En el caso de $\lambda_1 = 1$, al realizar la sustitución se obtiene:
$$
(A - I)\mathbf{v}_1 = 0 \implies \begin{bmatrix}
0 & 10^{-15} \\
0 & 10^{-15}
\end{bmatrix}
\begin{bmatrix}
x \\ y
\end{bmatrix} = \begin{bmatrix}
0 \\ 0
\end{bmatrix}
$$
$$$$
De esta forma se concluye que $y = 0$, y por lo tanto el vector propio asociado es:
$$
\mathbf{v}_1 = \begin{bmatrix} 1 \\ 0 \end{bmatrix}
$$
$$$$
Para el segundo valor propio, $\lambda_2 = 1 + 10^{-15}$, se realiza el mismo procedimiento:
$$
(A - \lambda_2 I)\mathbf{v}_2 = 0 \implies \begin{bmatrix}
-10^{-15} & 10^{-15} \\
0 & 0
\end{bmatrix}
\begin{bmatrix}
x \\ y
\end{bmatrix} = \begin{bmatrix}
0 \\ 0
\end{bmatrix}
$$
$$$$
Al resolver se obtiene la relación $x = y$, y por tanto el vector propio es:
$$
\mathbf{v}_2 = \begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$
$$$$
Con los vectores propios calculados, se determina el ángulo entre ellos.  
El ángulo $\theta$ se obtiene mediante la expresión:
$$
\cos \theta = \frac{\mathbf{v}_1 \cdot \mathbf{v}_2}{\|\mathbf{v}_1\| \|\mathbf{v}_2\|}
$$
$$$$
Al sustituir los valores, se tiene:
$$
\mathbf{v}_1 \cdot \mathbf{v}_2 = 1 \cdot 1 + 0 \cdot 1 = 1
$$
$$$$
Y las normas de los vectores son:
$$
\|\mathbf{v}_1\| = 1, \quad \|\mathbf{v}_2\| = \sqrt{1^2 + 1^2} = \sqrt{2}
$$
$$$$
Por lo tanto:
$$
\cos \theta = \frac{1}{\sqrt{2}} \implies \theta = 45^\circ
$$
$$$$
En consecuencia, los vectores propios de la matriz son $\mathbf{v}_1 = \begin{bmatrix} 1 \\ 0 \end{bmatrix}$ y $\mathbf{v}_2 = \begin{bmatrix} 1 \\ 1 \end{bmatrix}$, con un ángulo entre ellos de $45^\circ$.  
Este resultado muestra que la matriz A es casi idéntica a la identidad, por lo que sus vectores propios también son muy similares a los de la matriz identidad.  
La pequeña perturbación $10^{-15}$ altera ligeramente la estructura de la matriz, provocando que los vectores propios dejen de ser ortogonales.  
El ángulo calculado confirma este comportamiento y se puede comprobar numéricamente usando $cos^{-1}(1/\sqrt{2})$.

#2. (7 ptos) **Máximos y mínimos de una forma cuadrática**

Para la forma cuadrática $\mathbf{x}^TA\mathbf{x}$ vimos un resultado que incluía  las siguientes propiedades:

1. $\lambda_1\leq\mathbf{x}^TA\mathbf{x}\leq\lambda_n$.
2. El máximo de $\mathbf{x}^TA\mathbf{x}$ es $\lambda_n$, y ocurre cuando  $\mathbf{x}$ es un vector propio unitario correspondiente a $\lambda_n$.

Demuestre estas propiedades usando el resultado  $\mathbf{x}^TA\mathbf{x}=\mathbf{y}^T\Lambda\mathbf{y}$ obtenido mediante el cambio de variables $\mathbf{y}=V\mathbf{x}$ donde $V$ es la matriz ortogonal de vectores propios de $A$.

## **Solución Punto 2: Máximos y Mínimos de una Forma Cuadrática**

---

### **Parte (a)**

Se estudia la forma cuadrática $\mathbf{x}^T A \mathbf{x}$ con el objetivo de demostrar que su valor se encuentra acotado entre los valores propios mínimo y máximo de la matriz $A$.  
Dado que $A$ es una matriz simétrica, se puede expresar mediante una diagonalización ortogonal, lo que significa que existen una matriz ortogonal $V$ y una matriz diagonal $\Lambda$ tales que:
$$
A = V \Lambda V^T
$$
$$$$
Al realizar el cambio de variable $\mathbf{y} = V^T \mathbf{x}$, se puede reescribir la forma cuadrática como:
$$
\mathbf{x}^T A \mathbf{x} = \mathbf{y}^T \Lambda \mathbf{y} = \sum_{i=1}^{n} \lambda_i y_i^2
$$
$$$$
Dado que $V$ es ortogonal, se conserva la norma, por lo tanto $\|\mathbf{y}\| = \|\mathbf{x}\| = 1$.  
Cada término de la suma está ponderado por un valor propio $\lambda_i$, los cuales cumplen $\lambda_1 \leq \lambda_i \leq \lambda_n$.  
Al considerar que $\sum y_i^2 = 1$, se obtiene la desigualdad:
$$
\lambda_1 \sum y_i^2 \leq \sum_{i=1}^{n} \lambda_i y_i^2 \leq \lambda_n \sum y_i^2
$$
$$$$
De aquí se deduce que:
$$
\lambda_1 \leq \mathbf{x}^T A \mathbf{x} \leq \lambda_n
$$
$$$$
Esto demuestra que el valor de la forma cuadrática está siempre comprendido entre los valores propios mínimo y máximo de la matriz simétrica $A$.

---

### **Parte (b)**

El valor máximo de $\mathbf{x}^T A \mathbf{x}$ se alcanza cuando el vector $\mathbf{x}$ se alinea con el vector propio asociado al valor propio máximo $\lambda_n$, ya que en ese caso todas las demás componentes de $\mathbf{y}$ son cero y solo $y_n = 1$.  
De manera análoga, el valor mínimo $\lambda_1$ se obtiene cuando $\mathbf{x}$ es el vector propio asociado al valor propio mínimo $\lambda_1$.
$$$$
Por lo tanto, la forma cuadrática alcanza su valor máximo en el vector propio correspondiente a $\lambda_n$ y su valor mínimo en el asociado a $\lambda_1$.  
Estos resultados confirman que los extremos de la forma cuadrática están determinados por los valores propios de la matriz simétrica y que el intervalo de posibles valores está dado por $[\lambda_1, \lambda_n]$.  
Esta propiedad es esencial en el análisis de formas cuadráticas y en la teoría de optimización, pues permite comprender cómo la orientación del vector $\mathbf{x}$ influye en los valores de $\mathbf{x}^T A \mathbf{x}$.

#4. (10 ptos) **Descomposición en valores singulares**

Considere la matriz $A=\begin{bmatrix}
3&0\\
4&5
\end{bmatrix}$.

Paso#1:
Encuentre los valores propios de $A^TA$. Estos valores propios son los denominados *valores singulares* $\sigma_i$ al cuadrado de la matriz $A$  y sus vectores propios unitarios asociados son los respectivos vectores singulares $\mathbf{v}_i$ por izquierda.

Paso#2:
Los vectores singulares por derecha $\mathbf{u}_i$ se calculan mediante $$\mathbf{u}_i=\frac{A\mathbf{v}_i}{\sigma_i}.$$ ¿Que relación existe entre estos vectores $\mathbf{u}_i$  y los vectores propios de $AA^T$?

Paso#2a:
Normalice los vectores  $\mathbf{u}_i$

Paso #3:
Construya las matrices $$U=[\mathbf{u}_1\;\mathbf{u}_2],\quad \Sigma = diag(\sigma_1,\sigma_2)\quad V= [\mathbf{v}_1\; \mathbf{v}_2]$$

La anterior construcción debe permitir comprobar que $$A=U\Sigma V^T$$
y se denomina la *descomposición en valores singulares*, (SVD) por sus siglas en inglés, de $A$. Verifique la igualdad.


* ## Descomposición en valores singulares (SVD) dada la matriz

$$
A = \begin{bmatrix} 3 & 0 \\ 4 & 5 \end{bmatrix},
$$


* ### Paso 1: Cálculo de $A^T A$ y de sus valores propios

$$
A^T = \begin{bmatrix} 3 & 4 \\ 0 & 5 \end{bmatrix}
\quad \Rightarrow \quad
A^T A = \begin{bmatrix} 25 & 20 \\ 20 & 25 \end{bmatrix}
$$

Resolviendo el determinante:

$$
\det(A^T A - \lambda I) = (25 - \lambda)^2 - 400 = 0
\Rightarrow \lambda_1 = 45, \quad \lambda_2 = 5.
$$

Los valores singulares son:

$$
\sigma_1 = \sqrt{45} \approx 6.708, \quad \sigma_2 = \sqrt{5} \approx 2.236.
$$


* ### Paso 2: Cálculo de los vectores singulares por derecha $\mathbf{v}_i$

Para $\lambda_1 = 45$:

$$
(A^T A - 45I)\mathbf{v}_1 = 0 \Rightarrow
\mathbf{v}_1 = \frac{1}{\sqrt{2}}
\begin{bmatrix} 1 \\ -1 \end{bmatrix}
$$

Para $\lambda_2 = 5$:

$$
(A^T A - 5I)\mathbf{v}_2 = 0 \Rightarrow
\mathbf{v}_2 = \frac{1}{\sqrt{2}}
\begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$



* ### Paso 2a: Cálculo y normalización de los vectores singulares por izquierda $\mathbf{u}_i$

Se aplica la relación

$$
\mathbf{u}_i = \frac{A \mathbf{v}_i}{\sigma_i}
$$

Para $\mathbf{v}_1$:

$$
A \mathbf{v}_1 = \frac{1}{\sqrt{2}}
\begin{bmatrix} 3 \\ -1 \end{bmatrix}, \quad
\mathbf{u}_1 = \frac{1}{6.708} \cdot \frac{1}{\sqrt{2}}
\begin{bmatrix} 3 \\ -1 \end{bmatrix}
\approx
\begin{bmatrix} -0.3162 \\ -0.9487 \end{bmatrix}
$$

Para $\mathbf{v}_2$:

$$
A \mathbf{v}_2 = \frac{1}{\sqrt{2}}
\begin{bmatrix} 3 \\ 9 \end{bmatrix}, \quad
\mathbf{u}_2 = \frac{1}{2.236} \cdot \frac{1}{\sqrt{2}}
\begin{bmatrix} 3 \\ 9 \end{bmatrix}
\approx
\begin{bmatrix} -0.9487 \\ 0.3162 \end{bmatrix}
$$

Ambos vectores ya se encuentran normalizados


* ### Paso 3: Construcción de las matrices $U$, $\Sigma$ y $V$

$$
U = \begin{bmatrix}
-0.3162 & -0.9487 \\
-0.9487 & 0.3162
\end{bmatrix}, \quad
\Sigma = \begin{bmatrix}
6.708 & 0 \\
0 & 2.236
\end{bmatrix}, \quad
V = \begin{bmatrix}
-0.7071 & -0.7071 \\
-0.7071 & 0.7071
\end{bmatrix}
$$

Al verificar la igualdad:

$$
A = U \Sigma V^T \approx
\begin{bmatrix} 3 & 0 \\ 4 & 5 \end{bmatrix},
$$

se confirma que la descomposición en valores singulares (SVD) es correcta


* ## Relación entre los vectores $\mathbf{u}_i$ y los vectores propios de $AA^T$

Los vectores $\mathbf{u}_i$ son precisamente los vectores propios de $AA^T$,
y sus valores propios son los cuadrados de los valores singulares de $A$, es decir:

$$
AA^T \mathbf{u}_i = \sigma_i^2 \mathbf{u}_i
$$

De esta forma, los vectores singulares por izquierda representan las direcciones principales de $A$ en el espacio de salida, mientras que los vectores por derecha corresponden a las direcciones principales en el espacio de entrada

- **Paso \#1.** Calcule la factorización SVD
  $$H = U \Sigma V^T,$$
  donde $U\in\mathbb{R}^{n\times n}$ y $V\in\mathbb{R}^{n\times n}$ son ortogonales y $\Sigma$ es diagonal con $ \sigma_1\ge \sigma_2\ge\cdots\ge\sigma_n\ge0$.

- **Paso \#2.** Resuelva $U\mathbf{z}=\mathbf{b}$. Como $U$ es ortogonal,
  $$\mathbf{z}=U^T\mathbf{b}.$$

- **Paso \#3.** Resuelva $\Sigma\mathbf{y}=\mathbf{z}$. Sea $r$ el rango numérico (número de singulares $>\,$tolerancia).
  - Si $z_i\neq 0$ para algún $i>r$ (componentes asociadas a singulares nulas) entonces no existe solución exacta y debe considerarse el problema de mínimos cuadrados (o usar la pseudo-inversa).
  - Si $z_i=0$ para $i>r$, entonces para $i=1,\dots,r$ definimos
    $$y_i = \frac{z_i}{\sigma_i},$$
    y $y_i$ arbitrarios (por ejemplo 0) para $i>r$ si fuera necesario.

- **Paso \#4.** Resuelva $V^T\mathbf{x}=\mathbf{y}$, es decir
  $$\mathbf{x} = V\mathbf{y}.$$

La solución se puede escribir como suma:
$$\mathbf{x}=\sum_{i=1}^r \frac{\mathbf{u}_i^T\mathbf{b}}{\sigma_i}\,\mathbf{v}_i.$$

---
¿Cómo identifico la tolerancia y el rango $r$ en la práctica?

 Una regla práctica para la tolerancia numérica es
$$\text{tol} = \max(m,n)\cdot\varepsilon_{\text{mach}}\cdot\sigma_1,$$
donde $\varepsilon_{\text{mach}}$ es la máquina (por ejemplo `np.finfo(float).eps`). Entonces $r=\#\{i:\sigma_i>\text{tol}\}$.


## Pregunta (ii)

Para $n=5,10,15,20$ resolver con:
- Método A: `np.linalg.solve(H,b)` (solve directo).
- Método B: SVD usando **todas** las singulares (reconstrucción con suma sobre todos los $i$).
- Método C: SVD **truncada** usando solo $r$ singulares (filtrado).

Para cada caso calcule y tabule:
- $$\| \mathbf{b}-H\mathbf{x}\|$$ (residual).
- $$\dfrac{\|\mathbf{x}_{true}-\mathbf{x}\|}{\|\mathbf{x}_{true}\|}$$ (error relativo en $x$).

Adicional: calcule el número de condición aproximado
$$\kappa(H)\approx \frac{\sigma_1}{\sigma_n}\quad\text{o si }\sigma_n\approx0,\ \kappa(H)\approx\infty.$$

#Codigo :


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=6, suppress=True)

def hilbert_matrix(n):
    i = np.arange(n).reshape(-1,1)
    j = np.arange(n).reshape(1,-1)
    return 1.0 / (1 + i + j)

def svd_solve_full(H, b):
    U, s, Vt = np.linalg.svd(H, full_matrices=False)
    x = np.zeros(H.shape[1])
    for i in range(len(s)):
        ui = U[:, i]
        vi = Vt.T[:, i]
        x += (ui.T @ b) / s[i] * vi
    return x, (U, s, Vt)

def svd_solve_truncated(H, b, tol=None):
    U, s, Vt = np.linalg.svd(H, full_matrices=False)
    eps = np.finfo(float).eps
    if tol is None:
        tol = max(H.shape) * eps * s[0]   # regla  para tolerancia
    r = np.sum(s > tol)
    x = np.zeros(H.shape[1])
    for i in range(r):
        ui = U[:, i]
        vi = Vt.T[:, i]
        x += (ui.T @ b) / s[i] * vi
    return x, r, (U, s, Vt, tol)

def run_experiment(ns=[5,10,15,20]):
    rows = []
    tables = {}
    for n in ns:
        H = hilbert_matrix(n)
        x_true = np.ones(n)
        b = H @ x_true

        # Direct solve (Gaussian elimination)
        try:
            x_direct = np.linalg.solve(H, b)
        except np.linalg.LinAlgError:
            x_direct = np.full(n, np.nan)

        # SVD full reconstruction
        x_svd_full, svd_full = svd_solve_full(H, b)

        # SVD truncated (based on tol)
        x_svd_trunc, r, svd_trunc = svd_solve_truncated(H, b)

        # errors
        def rel_err(x):
            return np.linalg.norm(x_true - x) / np.linalg.norm(x_true)
        def res_norm(x):
            return np.linalg.norm(b - H @ x)

        # condition number (using nonzero smallest singular value)
        s = svd_full[1]
        sigma1 = s[0]
        sigman = s[-1] if s[-1] > 0 else np.nan
        cond = sigma1 / sigman if sigman and sigman > 0 else np.inf

        rows.append({'n': n, 'method': 'direct_solve',
                     'residual_norm': res_norm(x_direct),
                     'relative_error': rel_err(x_direct),
                     'cond(H)': cond})
        rows.append({'n': n, 'method': 'svd_full',
                     'residual_norm': res_norm(x_svd_full),
                     'relative_error': rel_err(x_svd_full),
                     'cond(H)': cond})
        rows.append({'n': n, 'method': f'svd_trunc_r={r}',
                     'residual_norm': res_norm(x_svd_trunc),
                     'relative_error': rel_err(x_svd_trunc),
                     'cond(H)': cond})

        tables[n] = {'H': H, 'b': b, 'x_true': x_true,
                     'x_direct': x_direct, 'x_svd_full': x_svd_full,
                     'x_svd_trunc': x_svd_trunc, 'singular_values': s,
                     'r': r, 'tol': svd_trunc[3], 'cond': cond}
    df = pd.DataFrame(rows)
    return df, tables

# Ejecutar experimento
df_results, tables = run_experiment([5,10,15,20])

# Mostrar resultados como DataFrame (Colab lo renderiza bonito)
df_results_display = df_results.copy()
# formatear números para visualización
df_results_display['residual_norm'] = df_results_display['residual_norm'].map(lambda v: f"{v:.3e}")
df_results_display['relative_error'] = df_results_display['relative_error'].map(lambda v: f"{v:.3e}")
df_results_display['cond(H)'] = df_results_display['cond(H)'].map(lambda v: f"{v:.3e}" if np.isfinite(v) else "inf")
display(df_results_display)

# Graficar decay de valores singulares para cada n
for n, info in tables.items():
    s = info['singular_values']
    plt.figure(figsize=(6,3.5))
    plt.semilogy(np.arange(1, len(s)+1), s, marker='o')
    plt.title(f"Singular values decay (n={n}), rank r={info['r']}, tol={info['tol']:.3e}")
    plt.xlabel("Índice i")
    plt.ylabel("σ_i (escala log)")
    plt.grid(True, which='both', ls='--', lw=0.5)
    plt.tight_layout()
    plt.show()

# Imprimir resumen en texto (opcional)
for n, info in tables.items():
    print(f"n = {n}: r = {info['r']}, tol = {info['tol']:.3e}, cond(H) ≈ {info['cond']:.3e}")
    print(" primeras singulares:", np.array2string(info['singular_values'][:min(10,len(info['singular_values']))], precision=6))
    print()
```

En los resultados obtenidos, el método directo $(\texttt{solve}$) y la SVD completa presentan errores relativos y residuales cada vez mayores, mientras que la SVD truncada mantiene una mejor estabilidad numérica.
