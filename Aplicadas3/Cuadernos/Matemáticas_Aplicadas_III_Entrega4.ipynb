### Definición de los vectores:

* Por filas:
$A = (1,2,3), \quad
B = (3,1,4), \quad
C = (-1,1,0)$

* Por columnas:
$D = (1,3,-1), \quad
E = (2,1,1), \quad
F = (3,4,0)$

Los vectores $A,B,C$ corresponden a las filas y los vectores $D,E,F$ a las columnas de la matriz $A$.

De aquí podemos eliminar algunos vectores que no aportan información al espacio, de la siguiente manera


### Para espacio fila:

Queremos verificar si el vector $C$ puede expresarse como combinación lineal de $A$ y $B$:

$$ a(1,2,3) + b(3,1,4) = (-1,1,0) $$

Esto nos da el sistema de ecuaciones:

$$
\begin{cases}
a+3b=-1 \\
2a+b=1 \\
3a+4b=0
\end{cases}
$$

De la segunda ecuación despejamos para $a$:

$$ a = \frac{1-b}{2} $$

Sustituyendo en la primera:

$$\frac{1-b}{2} + 3b = -1$$

$$ \frac{1+5b}{2} = -1 $$
$$ 5b=-3 $$
$$ b=-\frac{3}{5} $$

Ahora sustituimos $b$ en $a = \frac{1-b}{2}$:

$$ a = \frac{1-(-\tfrac{3}{5})}{2} = \frac{4}{5}$$

Con esto concluimos que:

$$ C = \frac{4}{5}A - \frac{3}{5}B$$

es decir, $C$ es combinación lineal de $A$ y $B$ que no es linealmente independiente de estos vectores, de lo cual se puede decir que una base del espacio fila es:

$$ \{A,B\} = \{(1,2,3), (3,1,4)\}$$

Algebraicamente:

$$ v = a(1,2,3) + b(3,1,4), \quad a,b \in \mathbb{R} $$

Lo cual desarrollado del sistema de ecuaciones planteado inicialmente queda como:

$$ v = (a+3b, 2a+b, 3a+4b) $$

### Para espacio columna

Queremos verificar si el vector $F$ puede expresarse como combinación lineal de $D$ y $E$:

$$ x(1,3,-1) + y(2,1,1) = (3,4,0) $$

Esto nos da el sistema de ecuaciones:

$$
\begin{cases}
x+2y=3 \\
3x+y=4 \\
-x+y=0
\end{cases}
$$

De la tercera ecuación despejamos para $y$:

$$ y=x $$

Sustituyendo el despeje anterior en la primera ecuación:

$$ x+2x=3 $$
$$ 3x=3 $$
$$ x=1 $$

Entonces:

$$ y=1 $$

Esto nos permiete concluir que:

$$ F = D + E $$

por lo tanto, $F$ es combinación lineal de $D$ y $E$, no es linealmente independiente de los vectores $D$ y $E$, y una base del espacio columna es:

$$ \{D,E\} = \{(1,3,-1), (2,1,1)\} $$

Algebraicamente:

$$ v = x(1,3,-1) + y(2,1,1), \quad x,y \in \mathbb{R} $$

Desarrollando del sistema de ecuaciones planteado:

$$ v = (x+2y, 3x+y, -x+y) $$

De las verificaciones realizadas, se puede concluir que en ambos casos uno de los tres vectores pudo expresarse como combinación lineal de los otros dos, lo cual significa que los espacios fila y columna no son linealmente independientes en su totalidad. Además, cada espacio está generado únicamente por dos vectores linealmente independientes, es decir:

$$ \dim(\operatorname{Row}(A)) = 2 $$
$$ \dim(\operatorname{Col}(A)) = 2 $$  

Esto confirma que tanto el espacio fila como el espacio columna forman un plano en $\mathbb{R}^3$


```python
import sympy as sp

# Definimos los vectores fila
A = sp.Matrix([1, 2, 3])
B = sp.Matrix([3, 1, 4])
C = sp.Matrix([-1, 1, 0])

# Definimos los vectores columna
D = sp.Matrix([1, 3, -1])
E = sp.Matrix([2, 1, 1])
F = sp.Matrix([3, 4, 0])

# Espacio fila
a, b = sp.symbols('a b')
eqs_fila = sp.Eq(a*A + b*B, C)
sol_fila = sp.solve(eqs_fila, (a, b))

print("Espacio Fila:")
print("Solución:", sol_fila)
print()

# Espacio columna
x, y = sp.symbols('x y')
eqs_col = sp.Eq(x*D + y*E, F)
sol_col = sp.solve(eqs_col, (x, y))

print("Espacio Columna:")
print("Solución:", sol_col)
```

##a.

Para comprobar si los vectores son linealmente independientes, lo que hacemos es plantear la combinación lineal igualada al vector cero. Esto significa que tomamos los vectores que queremos analizar y los multiplicamos por números (escalares) para ver si podemos sumarlos y obtener el vector nulo (el que tiene todas sus componentes en cero).
<br>
Esta ecuación se llama ecuación vectorial y su propósito es verificar si existe alguna manera de combinar los vectores para que su suma dé cero.
<br>
Es decir, buscamos si existe una combinación de la forma:
$$c_{1}\cdot a_{1}+c_{2}\cdot a_{2}+...+c_{n}\cdot a_{n}=0$$
donde $c_{1},c_{2},c_{n}$ son nuestros escalares.
<br>
Ahora, si la única solución es que todos los escalares sean cero
$$c_{1}=0,c_{2}=0,c_{n}=0$$
entonces esta es la solución trivial y los vectores son linealmente independientes.
<br>
Si encontramos al menos un escalar distinto de cero que cumpla la ecuación, eso se llama combinación no trivial y significa que los vectores son linealmente dependientes, es decir, uno de ellos puede escribirse como suma o múltiplo de los otros.
<br>
***
A continuación aplicamos esto al caso concreto partimos de la matriz dada por el enunciado:
$$A=\begin{bmatrix}
1&2&3\\
3&1&4\\
-1&1&0
\end{bmatrix}$$
de la cual tomamos como $a_{1}$ y $a_{2}$ las dos primeras columnas, es decir
$$a_{1}=(1,3,-1),a_{2}=(2,1,1)$$
Para $a_{3}$, tomaremos por ejemplo
$$a_{3}=(4,2,5)$$
Cada número en las posiciones de los vectores se usa para formar la ecuación. Al igualar la combinación lineal con el vector cero, tomamos cada posición y escribimos una ecuación para cada una de ellas.
<br>
Planteamos entonces la ecuación vectorial:
$$\alpha_{1}a_{1}+\alpha_{2}a_{2}+\alpha_{3}a_{3}=0$$
donde $\alpha_{1},\alpha_{2},\alpha_{3}$ son los escalares desconocidos. Al sustituir las componentes de $a_{1},a_{2},a_{3}$ que son los coeficientes que mencionamos anteriormente, esta ecuación se transforma en un sistema de tres ecuaciones:
$$\begin{cases}
\alpha_{1} + 2\alpha_{2} + 4\alpha_{3} = 0 \\
3\alpha_{1} + \alpha_{2} + 2\alpha_{3} = 0 \\
-\alpha_{1} + \alpha_{2} + 5\alpha_{3} = 0
\end{cases}
$$
Aquí cada número viene directamente de los valores de las columnas elegidas.
<br>
Ahora representamos ese sistema con la matriz aumentada.
<br>
Partimos de la combinación lineal:
$$
\alpha_1 \begin{pmatrix} 1 \\ 3 \\ -1 \end{pmatrix}
+ \alpha_2 \begin{pmatrix} 2 \\ 1 \\ 1 \end{pmatrix}
+ \alpha_3 \begin{pmatrix} 4 \\ 2 \\ 5 \end{pmatrix}
= \begin{pmatrix} 0 \\ 0 \\ 0 \end{pmatrix}
$$
La escribimos como matriz aumentada:
$$
\left[\begin{array}{ccc|c}
1 & 2 & 4 & 0\\
3 & 1 & 2 & 0\\
-1 & 1 & 5 & 0
\end{array}\right]
$$
Eliminamos el $3$ en la fila 2 usando $F_2 \leftarrow F_2 - 3F_1$.  
Primero multiplicamos la fila 1 por $3$:  

$$
3 \cdot [1 \quad 2 \quad 4] = [3 \quad 6 \quad 12]
$$

Luego restamos componente por componente a la fila 2:  

$$
[3 \quad 1 \quad 2] - [3 \quad 6 \quad 12] = [0 \quad -5 \quad -10]
$$

Por lo tanto, la nueva fila 2 es:  

$$
F_2 = [0 \quad -5 \quad -10]
$$


$$
\begin{bmatrix}
1&2&4&|&0\\
0&-5&-10&|&0\\
-1&1&5&|&0
\end{bmatrix}
$$

Eliminamos el $-1$ en la fila 3 usando $F_3 \leftarrow F_3 + F_1$.  
Primero dejamos la fila 1 como está:  

$$
F_1 = [1 \quad 2 \quad 4]
$$

Luego sumamos componente por componente a la fila 3:  

$$
[-1 \quad 1 \quad 5] + [1 \quad 2 \quad 4] = [0 \quad 3 \quad 9]
$$

Por lo tanto, la nueva fila 3 es:  

$$
F_3 = [0 \quad 3 \quad 9]
$$


$$
\begin{bmatrix}
1&2&4&|&0\\
0&-5&-10&|&0\\
0&3&9&|&0
\end{bmatrix}
$$

Normalizamos la fila 2 dividiéndola entre $-5$.  
Tomamos la fila 2 actual:  

$$
F_2 = [0 \quad -5 \quad -10]
$$

Dividimos cada componente entre $-5$:  

$$
0 \div (-5)=0, \quad -5 \div (-5)=1, \quad -10 \div (-5)=2
$$

Por lo tanto, la nueva fila 2 es:  

$$
F_2 = [0 \quad 1 \quad 2]
$$


$$
\begin{bmatrix}
1&2&4&|&0\\
0&1&2&|&0\\
0&3&9&|&0
\end{bmatrix}
$$

Eliminamos el $3$ en la fila 3 usando $F_3 \leftarrow F_3 - 3F_2$.  
Tomamos la fila 3 actual:  

$$
F_3 = [0 \quad 3 \quad 9]
$$

Restamos $3$ veces la fila 2:  

$$
3 - 3(1) = 0, \quad 9 - 3(2) = 3
$$

Por lo tanto, la nueva fila 3 es:  

$$
F_3 = [0 \quad 0 \quad 3]
$$


$$
\begin{bmatrix}
1&2&4&|&0\\
0&1&2&|&0\\
0&0&3&|&0
\end{bmatrix}
$$

Dividimos cada número de la fila 3 entre $3$ para que el pivote sea $1$:  

$$
0/3=0, \quad 0/3=0, \quad 3/3=1
$$

Así, la nueva fila 3 queda:  

$$
F_3 = [0 \quad 0 \quad 1]
$$


$$
\begin{bmatrix}
1&2&4&|&0\\
0&1&2&|&0\\
0&0&1&|&0
\end{bmatrix}
$$

Interpretamos el sistema resultante:  
De la última fila: $\alpha_3=0$  
De la segunda fila: $\alpha_2+2\alpha_3=0 \Rightarrow \alpha_2=0$  
De la primera fila: $\alpha_1+2\alpha_2+4\alpha_3=0 \Rightarrow \alpha_1=0$

$$
\boxed{\alpha_1=0,\quad \alpha_2=0,\quad \alpha_3=0}
$$

Podemos entonces concluir que, como en la solución obtenida todos los escalares $c_1, c_2, c_3$ son iguales a $0$, no existe ninguna combinación no trivial de los vectores que nos dé el vector nulo. Esto significa que ninguno de los tres vectores puede escribirse como combinación lineal de los otros, confirmando su independencia lineal de acuerdo con la definición.

Por lo tanto, los vectores $a_1, a_2, a_3$ son **linealmente independientes**.

$$
\begin{cases}
\alpha_1 + 2\alpha_2 + 4\alpha_3 = 1 \\
3\alpha_1 + \alpha_2 + 2\alpha_3 = 1 \\

De la primera fila:

$$
\boxed{\alpha_1=\tfrac{1}{5}, \quad \alpha_2=-\tfrac{2}{15}, \quad \alpha_3=\tfrac{4}{15}}
$$

Podemos concluir que el vector $b$ sí es combinación lineal de los vectores de $S$, porque encontramos valores de $\alpha_1, \alpha_2, \alpha_3$ que cumplen la ecuación. Además, como el sistema tiene solución única (no aparecen variables libres), esta es la única forma posible de escribir $b$ como combinación lineal de los elementos de $S$.


```python
import numpy as np

# Define los tres vectores del conjunto S.
a1 = np.array([1, 3, -1])
a2 = np.array([2, 1, 1])
a3 = np.array([4, 2, 5])  # <- ya corregido según tu indicación

# Define el vector objetivo b.
b = np.array([1, 1, 1])

# 1. CONSTRUIR EL SISTEMA
# Se crea la matriz A donde las columnas son los vectores a1, a2, a3.
A = np.column_stack((a1, a2, a3))

print("--- RESOLUCIÓN DEL SISTEMA Ax = b ---")
print(f"La matriz A (vectores de S) es:\n{A}")
print(f"\nEl vector b es: {b}")

# 2. RESOLVER EL SISTEMA
# Se utiliza la función np.linalg.solve() para encontrar el vector de escalares 'x'
# tal que A*x = b.
try:
    escalares = np.linalg.solve(A, b)

    print("\n¡El sistema tiene una solución única!")
    print("Esto confirma numéricamente que b SÍ es una combinación lineal de los vectores en S.")

    print("\nLa solución (los escalares) es:")
    print(f"  alpha_1 = {escalares[0]:.4f}  (Equivalente a 1/5)")
    print(f"  alpha_2 = {escalares[1]:.4f}  (Equivalente a -2/15)")
    print(f"  alpha_3 = {escalares[2]:.4f}  (Equivalente a 4/15)")

except np.linalg.LinAlgError:
    print("\nEl sistema no se puede resolver. El vector b no es una combinación lineal (o no lo es de forma única).")
```

##c.

Para decidir si el conjunto $S=\{a_1,a_2,a_3\}$ genera todo $\mathbb{R}^3$ basta comprobar si las tres columnas son linealmente independientes. Si lo son (determinante de la matriz formada por las columnas distinto de cero), entonces generan $\mathbb{R}^3$ y cualquier vector de $\mathbb{R}^3$ puede escribirse como combinación lineal de los elementos de $S$.

Partimos de:
$$
a_{1}=\begin{pmatrix}1\\3\\-1\end{pmatrix},\qquad
a_{2}=\begin{pmatrix}2\\1\\1\end{pmatrix},\qquad
a_{3}=\begin{pmatrix}4\\2\\5\end{pmatrix}
$$

Formamos la matriz con estas columnas:
$$
A=\begin{bmatrix}
1 & 2 & 4\\[4pt]
3 & 1 & 2\\[4pt]
-1 & 1 & 5
\end{bmatrix}
$$

Calculamos el determinante de $A$. Con la fórmula se obtiene:
$$
\det(A)=1(1\cdot 5 - 2\cdot 1)-2(3\cdot 5 - 2\cdot(-1))+4(3\cdot 1 - 1\cdot(-1)) = 3 - 34 + 16 = -15
$$

Como $\det(A)=-15\neq 0$, las columnas son linealmente independientes y, en $\mathbb{R}^3$, tres vectores independientes generan todo $\mathbb{R}^3$.


```python
import numpy as np

# --- CONFIGURACIÓN ---
# Define los tres vectores del conjunto S.
a1 = np.array([1, 3, -1])
a2 = np.array([2, 1, 1])
a3 = np.array([4, 2, 5])  # <- Cambiado según tu conjunto S

# Define un vector de ejemplo para probar la combinación lineal.
vector_ejemplo = np.array([4, 7, 1])

# --- ANÁLISIS NUMÉRICO ---

# 1. PRUEBA DE GENERACIÓN (¿FORMAN UNA BASE?)
# Se crea una matriz 'A' donde las columnas son los vectores a1, a2, a3.
A = np.column_stack((a1, a2, a3))

# Se calcula el determinante de la matriz A.
# Si el determinante es diferente de cero, los vectores son linealmente
# independientes. En un espacio 3D, 3 vectores linealmente independientes
# siempre generan todo el espacio.
determinante = np.linalg.det(A)

print("--- 1. ANÁLISIS DE LA BASE ---")
print(f"La matriz A formada por los vectores de S es:\n{A}")
print(f"\nEl determinante de A es: {determinante:.2f}")

if determinante != 0:
    print("RESULTADO: Como el determinante no es cero, los vectores son linealmente independientes y generan todo R^3.")
else:
    print("RESULTADO: El determinante es cero, por lo que los vectores NO generan todo R^3.")

# Separador para mayor claridad
print("\n" + "="*50 + "\n")

# 2. DEMOSTRACIÓN PRÁCTICA (ENCONTRAR ESCALARES)
# Se resuelve el sistema de ecuaciones A*x = vector_ejemplo para encontrar
# los escalares 'x' (que son alpha_1, alpha_2, alpha_3).
print("--- 2. DEMOSTRACIÓN CON UN VECTOR DE EJEMPLO ---")
print(f"Buscando los escalares para el vector v = {vector_ejemplo}...")

try:
    # Intenta resolver el sistema lineal.
    escalares = np.linalg.solve(A, vector_ejemplo)

    print("\nSolución encontrada")
    print(f"  alpha_1 = {escalares[0]:.4f}")
    print(f"  alpha_2 = {escalares[1]:.4f}")
    print(f"  alpha_3 = {escalares[2]:.4f}")
    print("\nRESULTADO: Esto demuestra numéricamente que el vector SÍ es una combinación lineal de los elementos de S.")

except np.linalg.LinAlgError:
    # Esto solo ocurriría si el determinante fuera cero.
    print(f"El sistema no tiene una solución única para el vector v = {vector_ejemplo}.")

```

i. Independencia lineal de vectores apilados. Considere los vectores apilados

$$\mathbf{c}_1=\begin{bmatrix}a_1\\b_1\end{bmatrix},\quad\cdots \mathbf{c}_k=\begin{bmatrix}a_k\\b_k\end{bmatrix},$$

donde $a_1,.., a_k$ son $n-$vectores y $b_1,...,b_k$ son $m-$vectores.

(a) Suponga que $a_1 ,..., a_k$ son linealmente independientes. (No hacemos suposiciones sobre los vectores $b_1,...,b_k.$) ¿Podemos concluir que los vectores apilados $c_1,...,c_k$ son linealmente independientes?

(b) Suponga ahora que $a_1,..., a_k$ son linealmente dependientes. (De nuevo, sin suposiciones sobre $b_1,...,b_k.$) ¿Podemos concluir que los vectores apilados $c_1,...,c_k$ son linealmente dependientes?

---
##**$Solucion$ $i$**

Sean
$$
\mathbf{c}_i=\begin{bmatrix}a_i\\[4pt]b_i\end{bmatrix},\qquad i=1,\dots,k,
$$
con $a_i\in\mathbb{R}^n$ y $b_i\in\mathbb{R}^m$.


## $a).$

Si $a_1,\dots,a_k$ son linealmente independientes → ¿$c_1,\dots,c_k$ independientes?

Consideremos la combinación
$$
\alpha_1\mathbf{c}_1+\cdots+\alpha_k\mathbf{c}_k=\mathbf{0}.
$$
Por bloques esto equivale a las dos ecuaciones
$$
\alpha_1 a_1+\cdots+\alpha_k a_k = 0 \quad\text{(en }\mathbb{R}^n\text{)},
$$
$$
\alpha_1 b_1+\cdots+\alpha_k b_k = 0 \quad\text{(en }\mathbb{R}^m\text{)}.
$$
Como $a_1,\dots,a_k$ son linealmente independientes, la primera ecuación implica
$$
\alpha_1=\cdots=\alpha_k=0.
$$
Por tanto la única combinación que anula a los $\mathbf{c}_i$ es la trivial, y $c_1,\dots,c_k$ son **linealmente independientes**.

Por lo anteriror se puede concluir que si son linealmente independientes, dado que la independencia de los $a_i$ asegura la independencia de los $\mathbf{c}_i$, sin importar los $b_i$.

## $b).$

 Si $a_1,\dots,a_k$ son linealmente dependientes → ¿podemos concluir que $c_1,\dots,c_k$ son dependientes?

Si los $a_i$ son dependientes, existen escalares $\beta_1,\dots,\beta_k$, no todos cero, tales que
$$
\beta_1 a_1 + \cdots + \beta_k a_k = 0.
$$

Para que los $\mathbf{c}_i$ sean dependientes, se requiere además que la **misma combinación** anule las segundas componentes:
$$
\beta_1 b_1 + \cdots + \beta_k b_k = 0.
$$

Sin embargo, esta condición **no está garantizada**: depende de los $b_i$.  
Es posible que los $a_i$ sean dependientes y, aun así, los $\mathbf{c}_i$ resulten independientes.

**Ejemplo:**  
Sea $k=2$, con
$$
a_1=a_2=1, \qquad b_1=0,\quad b_2=1.
$$

Entonces
$$
\mathbf{c}_1=\begin{bmatrix}1\\0\end{bmatrix}, \qquad
\mathbf{c}_2=\begin{bmatrix}1\\1\end{bmatrix}.
$$

Aunque $a_1$ y $a_2$ son dependientes (pues son iguales), los vectores apilados **no** lo son:  
no existe un escalar $\lambda$ tal que $\mathbf{c}_2 = \lambda \mathbf{c}_1$.  

Geométricamente, en el plano $\mathbb{R}^2$, $\mathbf{c}_1$ es horizontal y $\mathbf{c}_2$ es diagonal, por lo que no están alineados.

Entonces la respuesta es, no siempre. La dependencia de los $a_i$ no implica necesariamente la dependencia de los $\mathbf{c}_i$; se necesita que los $b_i$ satisfagan la misma relación de dependencia.
