### ii)

En la integral original, para el límite externo se tiene que $y$ va de $[0,1]$ y para el orden interno se tiene que $x$ va desde $y$ hasta $1$ dado un $y$; inicialmente el dominio de integración es ${(x,y):0\leq y \leq 1, y \leq x \leq 1}$, por lo que para el área se tiene  una recta $y=0$ que es el límite inferior, una recta $y=1$ que es el límite superior, una curva $x=y$ que es la diagonal y una recta $x=1$ que es la vertical, lo cual formaria la figura del applet contenida entre los puntos $A, B, C$. Al cambiar el orden se busca pasar a $dydx$, es por ello que se busca reescribir la misma región pero en términos de $x$, por lo que el $x$ mínimo en la región es $0$ y el máximo es $1$ $(0\leq x \leq1)$, de igual manera para un $x$ fijo se tiene que el valor en $y$ varía desde el límite inferior $y=0$ hasta la curva diagonal $y=x$ $(0\leq y \leq x)$

La reescritura del orden de integración es de la forma
$$\int_{0}^1 \int_{0}^{x} \sin (x^2)\, dy\, dx.$$

Es más fácil reesribir el orden de la integral para $dy dx$ porque no es conveniente integrar $sin(x^2)$ respecto a $x$ debido a que esta no tiene una antiderivada, por lo que al invertir el orden, la integral pasa a ser respecto a $y$ donde $sin(x^2)$ es una constante




### iii)
Resolver para el orden $dy dx$:
$$\int_{0}^1 \int_{0}^{x} \sin (x^2)\, dydx$$
Primero integramos para $dy$
$$\int_{0}^1 sin(x^2) y\Big|_{0}^x dx$$
Evaluamos los límites para $y$

$$\int_{0}^1 sin(x^2)(x)-sin(x^2)(0) dx$$
$$\int_{0}^1 (x)sin(x^2)dx$$

Ahora se puede aplicar sustitución $u$ $du$, en donde se facilita la antiderivada de forma que $u=x^2$ y $du =2xdx → xdx=\frac{1}{2}du$
Ya teniendo esta sustitución, podemos aplicarla en la integral definida

$$\int_{0}^1 (\frac{1}{2})sin(u)du$$ \
$$\frac{1}{2}\int_{0}^1 sin(u)du$$

Integramos para $u$

$$\frac{1}{2}(-cos(u))\Big|_{0}^1$$

Reescribimos en términos de $x$

$$\frac{1}{2}(-cos(x^2))\Big|_{0}^1$$

Evaluamos los límites para $x$
$$\frac{1}{2}(-cos(1^2))-\frac{1}{2}(-cos(0^2))$$ \
$$\frac{1}{2}(-cos(1))-\frac{1}{2}(-1)$$ \
$$\frac{1}{2}(-cos(1))+\frac{1}{2}(1)$$ \
$$\frac{1}{2}(1-cos(1))$$

```python
# iii)
import sympy as sp

# Definir variables
x, y = sp.symbols('x y')

# Integrando
f = sp.sin(x**2)

# --- Orden original: dx luego dy ---
I1 = sp.integrate(sp.integrate(f, (x, y, 1)), (y, 0, 1))

# --- Orden cambiado: dy luego dx ---
I2 = sp.integrate(sp.integrate(f, (y, 0, x)), (x, 0, 1))

# Resultados
print("Orden original (dx luego dy):", sp.simplify(I1))
print("Orden cambiado (dy luego dx):", sp.simplify(I2))
print("Valor numérico:", float(I1))
```

### iii)
Resolver para el orden $dydx$
 $$\int_{0}^1 \int_{-x}^{x} (x+y)^2\, dydx$$

 Primero calculamos la integral interior $y$
  $$\int_{0}^1 \int_{-x}^{x} (x^2+2xy+y^2)\, dydx$$

 Resolvemos término a término
 * Para $x^2$
 $$\int_{-x}^{x} x^2 dy$$ \
 $$x^2y\Big|_{-x}^{x}$$ \
 $$[x^2(x)]-[x^2(-x)]$$ \
 $$x^2(2x)$$ \
 $$2x^3$$

 * Para $2xy$
 $$\int_{-x}^{x} 2xy dy$$ \
 $$2x\int_{-x}^{x} y dy$$ \
 $$2x(\frac{y^2}{2})\Big|_{-x}^{x}$$ \
 $$[2x(\frac{x^2}{2})]-[2x(\frac{(-x)^2}{2})]$$ \
 $$2x(\frac{x^2}{2}-\frac{x^2}{2})$$ \
 $$2x\cdot 0$$ \
 $$0$$

 * Para $y^2$
 $$\int_{-x}^{x} y^2 dy$$ \
 $$\frac{y^3}{3}\Big|_{-x}^{x}$$ \
 $$\frac{x^3}{3}-(-\frac{x^3}{3})$$ \
 $$\frac{2}{3}x^3$$


 Ahora se suman los términos obtenidos
 $$2x^3+0+\frac{2}{3}x^3$$ \
 $$\frac{8}{3}x^3$$

 Se integra para $x$
 $$\int_{0}^1 \frac{8}{3}x^3 dx$$ \
 $$\frac{8}{3} \int_{0}^1 x^3 dx$$ \
 $$\frac{8}{3} (\frac{x^4}{4})\Big|_{0}^{1}$$ \
 $$\frac{8}{3}(\frac{1}{4})$$ \
 $$\frac{8}{12}$$ \
 $$\frac{2}{3}$$

```python
# iii)
import sympy as sp

# Definimos variables
x, y = sp.symbols('x y', real=True)

# Función integrando
f = (x + y)**2

# -----------------------------
# 1) Orden original: dx luego dy
# Región: -1 <= y <= 1, |y| <= x <= 1
I1 = ( sp.integrate(sp.integrate(f, (x, y, 1)), (y, 0, 1))  # y >= 0
     + sp.integrate(sp.integrate(f, (x, -y, 1)), (y, -1, 0)) )  # y <= 0

# -----------------------------
# 2) Orden cambiado: dy luego dx
# Región: 0 <= x <= 1, -x <= y <= x
I2 = sp.integrate(sp.integrate(f, (y, -x, x)), (x, 0, 1))

# -----------------------------
print("Integral en orden original =", I1.simplify())
print("Integral en orden cambiado =", I2.simplify())
```

Por lo que al cambiar el orden para describir la misma región pero invirtiendo $y$ y $x$, teniendo en cuenta que $y$ puede tomar $1$ como valor mínimo y $e^x=e^1$ como valor máximo, entonces el límite en $y$ está dado por $1 \leq y \leq e$, y por cada $y$ fijo dentro del rango se cumple la condición de que $y \leq e^x$, esto para $x$ se re interpreta como $x \geq ln\, y$ que a su vez está limitado como $x \leq 1$, de manera que el límite para $x$ está dado por $ln\, y \leq x \leq 1$

La reescritura del orden de integración es la siguiente
$$\int_{1}^e \int_{ln\, y}^{1} 1 \, dx\, dy.$$

Para este caso, el cambio del orden puede no ser tan necesario como en los casos observados anteriormente ya que en el orden original la integral interior tiene antiderivada directa, y a su vez  en el orden cambiado $ln\, y$ resulta igualmente manejable. ### iii)
Resolver para el orden $dxdy$
$$\int_{1}^{e}\int_{ln\, y}^{1} 1 dxdy$$

Primero integramos para $x$
$$\int_{1}^{e} x\Big|_{ln\, y}^{1} dy$$
Evaluamos los límites de $x$
$$\int_{1}^{e} 1-ln\, y \, dy$$

Ahora separamos por términos para integrar $ln\, y$ por partes y evaluamos para el límite $y$
* Para $1$
$$y \Big|_{1}^{e}$$ \
$$e-1$$
* Para $ln\, y$
Integrando por partes tenemos entonces $u=ln\,y$, $dv=dy$, $du=\frac{1}{y}dy$, $v=y$

Aplicamos la forma
$$uv-\int vdu$$ \
$$ln\,y\cdot y-\int y\cdot \frac{1}{y}dy$$ \
$$ln\,y\cdot y-\int 1dy$$ \
$$ln\,y \cdot y-y \Big|_{1}^{e}$$ \
$$(e\cdot ln\, e -e)-(1\cdot ln\, 1 -1)$$ \
$$(e\cdot 1-e)-(1\cdot 0-1)$$ \
$$0-1$$ \
$$-1$$

Ahora agrupamos los resultados de evaluar el límite por cada término y llegamos al resultado exacto de la integral
$$(e-1)-1$$ \
$$e-2$$


```python
# iii)
import sympy as sp

# Definimos las variables
x, y = sp.symbols('x y')

# --- Integral en el orden original: dy dx ---
integral_original = sp.integrate(sp.integrate(1, (y, 1, sp.exp(x))), (x, 0, 1))

# --- Integral con orden cambiado: dx dy ---
integral_cambiada = sp.integrate(sp.integrate(1, (x, sp.log(y), 1)), (y, 1, sp.E))

# Mostramos los resultados
print("Integral con orden original (dy dx):", integral_original)
print("Integral con orden cambiado (dx dy):", integral_cambiada)
```

##ii.
En los límites de integración dados podemos observar que la región $D$:

La variable $y$ varía en el intervalo $0 \leq y \leq 1$.

Para un valor fijo de $y$, la variable $x$ varía en el intervalo $y \leq x \leq \sqrt{2 - y^{2}}$.

La desigualdad $x \leq \sqrt{2 - y^{2}}$ puede reescribirse elevando al cuadrado ambos lados como $x^{2} \leq 2 - y^{2}$. Reordenando se obtiene $x^{2} + y^{2} \leq 2$, lo cual representa el interior de la circunferencia de radio $\sqrt{2}$ centrada en el origen. Además, los límites dados son $0 \leq y \leq 1$, por lo que la región se encuentra en el primer cuadrante.

Por otra parte, debido a la condición de la desigualdad $y \leq x$, la región está a la derecha de la recta $x = y$.

En consecuencia, la región $D$ está delimitada por la recta $x = y$, el eje $y = 0$ y el arco de la circunferencia definida por $x^{2} + y^{2} = 2$.

Ahora, para cambiar el orden de integración, debemos proyectar la región sobre el eje $x$ y así poder reescribir la integral en el orden $dy , dx$.

Para un valor fijo de $x$, la variable $y$ comienza en la base de la región, es decir, en $y = 0$. El límite superior de $y$ depende de la posición de $x$:

Para valores de $x$ cercanos al origen, la frontera superior está dada por la recta $y = x$.

Para valores de $x$ mayores, la frontera superior está dada por la semicircunferencia $x^{2} + y^{2} = 2$.

Ambas fronteras se cruzan en el punto que satisface el sistema:
	​
$$\begin{cases}
x = y, \\[4pt]
x^2 + y^2 = 2
\end{cases}$$

Sustituyendo $y = x$ en la ecuación de la circunferencia:

$$x^2 + x^2 = 2 \quad \Rightarrow \quad 2x^2 = 2$$

Dividimos entre $2$:

$$x^2 = 1 \quad \Rightarrow \quad x = \pm 1$$

Dado que en la región se cumple $y \geq 0$ y $x \geq 0$, tomamos $x = 1$ y $y = 1$.

Esto implica que:

- Para $0 \leq x \leq 1$, el límite superior de $y$ es $y = x$.

- Para $1 \leq x \leq \sqrt{2}$, el límite superior de $y$ es $y = \sqrt{2 - x^{2}}$.

Por lo tanto, la integral doble original:


$$\int_0^1\int_y^{\sqrt{2−y^2}} f (x, y)\, dx dy,$$

puede escribirse como:  

$$\int_0^1 \int_0^x f(x,y) \, dy \, dx \;+\; \int_1^{\sqrt{2}} \int_0^{\sqrt{2 - x^2}} f(x,y) \, dy \, dx$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de la recta y = x
def recta(x):
    return x

# Definición de la semicircunferencia superior: x^2 + y^2 = 2
def semicirc(x):
    return np.sqrt(2 - x**2)

# Valores de x para graficar
x_recta = np.linspace(0, 1, 200)  # Para la recta (hasta la intersección)
x_semicirc = np.linspace(1, np.sqrt(2), 200)  # Para la semicircunferencia

# Crear figura
fig, ax = plt.subplots(figsize=(6, 6))

# Graficar recta y semicircunferencia
ax.plot(x_recta, recta(x_recta), 'r', label='$y = x$')
ax.plot(x_semicirc, semicirc(x_semicirc), 'b', label='$x^2 + y^2 = 2$')

# Rellenar región
# Parte inferior: y = x, parte superior: semicircunferencia
x_fill = np.linspace(0, np.sqrt(2), 400)
y_lower = np.where(x_fill <= 1, recta(x_fill), 0)  # recta solo hasta x=1
y_upper = semicirc(x_fill)
mask = x_fill <= np.sqrt(2)

# Rellenar para x en [0, 1] entre recta y semicircunferencia
x_fill1 = np.linspace(0, 1, 300)
plt.fill_between(x_fill1, recta(x_fill1), semicirc(x_fill1), color='lightblue', alpha=0.5)

ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_aspect('equal', adjustable='box')
ax.legend()
ax.set_title('Región de integración')

plt.grid(True)
plt.show()
```
