#1. (8 ptos) *Integrales iteradas.*

  Halle el valor de las siguientes integral iteradas: $$i)\,\int\int_R \cos x\sin y\, dx dy, \quad ii)\,\int\int_R \,\frac{y}{1+x^2}\, dx dy$$ donde $R=[-1,0]\times[1,2]$.

# Argumento sobre el dominio
En primer lugar, elegimos integrar en el oden $dxdy$ porque el integrando se puede separar como $f(x)g(y)$ y por lo tanto resulta más conveniente integrar primero las partes naturales en $x$, específicamente en la segunda integral iterada para la cual al separar $y$ como constante resulta más sencillo integrar $\frac{1}{1+x^2}$ como $arctan(x)$ de forma inmediata, es decir, gracias a esto se puede obtener la antiderivada de forma más directa. <br>
Ahora bien, el dominio $R=[-1,0]\times[1,2]$ se eligió de la forma ${}(x,y):-1\leq x\leq 0, 1\leq y\leq 2$ porque el enunciado define $R$ como un rectángulo de base $x$ entre $-1$ y $0$ y altura en $y$ entre $1$ y $2$, por lo que al escribir las integrales como iteradas con orden $dxdy$, el intervalo interno corresponde a $x$, manteniendo a $y$ fijoo y por ser rectángulo sus límites son constantes.

### Solución 1.$i$)
$$ \int_1^2\int_{-1}^0 cosx seny dxdy $$

Siguiendo el orden dado $dxdy$, vamos a integrar primero respecto $x$:
$$ \int_1^2 senx seny dy \Big|_{-1}^0 $$

Se evalúan los límites para $[-1, 0]$:
$$ \int_1^2 (sen(0) seny - sen(-1)seny) $$
$$ \int_1^2 -sen(-1)seny $$
$$ \int_1^2 sen(1)seny $$
Después de evaluar los límites para $x$, se integra respecto a $y$:
$$ sen(1) \cdot (- cosy \Big|_{1}^2)$$

Se evalúan los límites para $[1, 2]$:
$$ (sen(1) \cdot (-cos(2))) - (sen(1) \cdot (-cos(1))) $$

$$ = 0.8048 $$


```python
#Solución numérica en código
import sympy as sp

# Definir variables
x, y = sp.symbols('x y')

# Definir la función
f = sp.cos(x) * sp.sin(y)

# Límites de integración
lim_x = (x, -1, 0)
lim_y = (y, 1, 2)

# Preguntar orden al usuario
orden = input("¿Integrar primero en dx o dy? (dx/dy): ").strip().lower()

if orden == "dx":
    # Primero en x, luego en y
    resultado = sp.integrate(f, lim_x, lim_y)
elif orden == "dy":
    # Primero en y, luego en x
    resultado = sp.integrate(f, lim_y, lim_x)
else:
    print("Opción no válida.")
    resultado = None

if resultado is not None:
    print("Resultado:", resultado)
```

##Solución

El Principio de Cavalieri establece lo siguiente:
Si dos sólidos tienen la misma altura y sus secciones transversales a cualquier nivel (paralelas a la base) tienen igual área, entonces ambos sólidos tienen el mismo volumen.
<br>
Ahora, aplicamos este principio a la figura dada:
<br>
Observamos que la figura tiene una altura de $7$ unidades, y tanto la base inferior como la superior son rectángulos de aparentemente la misma forma y área. Es decir, podemos calcular el área de la base como:
$$Base = ancho\cdot largo$$
<br>
Donde:
<br>
* Altura = $7$ unidades
* Base inferior = $3\cdot 5$
* Base superior = $3\cdot 5$
<br>

Al cortar el sólido con planos horizontales a distintas alturas, obtenemos secciones transversales que siempre son rectángulos con el mismo valor de área. Esto significa que, aunque los lados pudieran deformarse en la forma, la superficie de cada sección a lo largo de la altura no cambia.
<br>
Por lo tanto, como en todos los niveles las secciones tienen el mismo área, se cumple el Principio de Cavalieri. Esto significa que el sólido tiene el mismo volumen que un prisma recto con base de $3\cdot 5$ y altura $7$.
<br>
**Aplicación**
<br>
El sólido tiene una altura de $h=7$ unidades. Aunque la forma de la base cambia gradualmente a lo largo de la altura, el área de cada sección transversal se mantiene constante en todos los niveles. Sabemos que, mientras el área transversal sea la misma en cada altura, el volumen será igual al de un prisma recto con esa base y altura. Podemos entonces calcular el volumen usando la siguiente fórmula:

$$V=A_{base}\cdot h$$

Dado que tanto la base inferior como la base superior tienen el mismo área, podemos tomar la base inferior como referencia. Así:

$$A_{base}=3\cdot 5=15$$

Ahora multiplicamos por la altura:

$$V=15\cdot 7=105$$

De acuerdo con el desarrollo anterior, podemos concluir que el volumen del sólido es de $105$ unidades cúbicas, asumiendo que la transformación de la base mantiene la misma área en todas las alturas.
Esto es correcto gracias al Principio de Cavalieri, el cual nos permite calcular el volumen sin necesidad de conocer una función explícita que describa la variación de la base.

##a)  
El Principio de Cavalieri establece que si dos sólidos tienen la misma altura y las áreas de sus secciones transversales tomadas a la misma altura son iguales, entonces los dos sólidos tienen el mismo volumen.
Para calcular el volumen del sólido de la Figura $(a)$, consideramos que éste se genera al rotar la curva $y = f(x)$ alrededor del eje $x$ en el intervalo $[a,b]$.

##Secciones transversales:

Al tomar un corte perpendicular al eje $x$ en un punto fijo $x$, la sección transversal del sólido es un disco con radio igual a $f(x)$.  
El área de este disco es  
$$
A(x) = \pi [f(x)]^2.
$$

Esto es dado que seguimos la siguiente formula para hallar el area de un circulo, la cual es:
$$A=\pi \cdot r^{2}$$
En este caso, reemplazamos $r$ con la función $f(x)$ ya que la función es la que delimita la figura. Al evaluar el punto correspondiente, que es el centro de la circunferencia, la distancia entre este punto central y el límite del círculo está dada por la función misma. Por eso, el radio del área en este caso es la propia función. Aplicamos el Principio de Cavalieri porque, aunque el sólido tenga variaciones en su forma a diferentes alturas (como un cilindro deformado), las secciones transversales siguen siendo discos y podemos integrar sus áreas a lo largo del eje $x$ para hallar el volumen.
<br>

##Integración para el volumen:

Sumamos el volumen de todos estos discos a lo largo del intervalo $[a,b]$, lo que equivale a integrar las áreas transversales:  
$$
V = \int_a^b A(x) \, dx = \pi \int_a^b [f(x)]^2 \, dx.
$$

Esta integral representa el volumen del sólido generado por la rotación de la curva $y=f(x)$ alrededor del eje $x$.

Por lo tanto, utilizando el método de discos y el principio de Cavalieri, concluimos que el volumen del sólido es:  
$$
V = \pi \int_a^b [f(x)]^2 \, dx.
$$

## **Cálculo del volumen del sólido con las condiciones dadas**

El sólido está delimitado por:

- Los planos $(xz)$ y $(yz)$ (lo que implica $( x \geq 0 )$ y $( y \geq 0 )$).
- El plano $( xy )$ (es decir, $( z = 0 )$).
- Las restricciones $( x \leq 1 )$ y $( y \leq 1 )$.
- La superficie superior $( z = x^2 + y^4 )$.

El volumen se calcula integrando $( z = x^2 + y^4 )$ sobre la región $( 0 \leq x \leq 1,\ 0 \leq y \leq 1 )$:

$$
V = \int_0^1 \int_0^1 \big( x^2 + y^4 \big) \, dy \, dx
$$


**1. Integramos respecto a \(y\):**  

$$
\int_0^1 \big( x^2 + y^4 \big) \, dy
= \int_0^1 x^2 \, dy + \int_0^1 y^4 \, dy
$$

$$
= x^2 y \Big|_0^1 + \frac{y^5}{5} \Big|_0^1
$$

$$
= x^2(1) + \frac{1}{5} = x^2 + \frac{1}{5}
$$


**2. Integramos respecto a \(x\):**  

$$
\int_0^1 \left( x^2 + \frac{1}{5} \right) dx
= \int_0^1 x^2 \, dx + \frac{1}{5} \int_0^1 dx
$$

$$
= \frac{x^3}{3} \Big|_0^1 + \frac{1}{5} x \Big|_0^1
$$

$$
= \frac{1}{3} + \frac{1}{5}
= \frac{5}{15} + \frac{3}{15}
= \frac{8}{15}
$$


**Conclusión:**  
El volumen del sólido es:  

$$
V = \frac{8}{15}

#5. (12 puntos)
i. Sea $f$ continua en $[a, b]$ y $g$ continua en $[c, d]$. Explique o muestre por qué $$\int\int_R f(x)g(y)\, dx dy=\int_a^bf(x)\,dx\int_c^d g(y)\, dy,$$ donde $R=[a,b]\times[c,d]$.

ii. Aunque el teorema de Fubini se cumple para la mayoría de las funciones
en la práctica, este ejemplo da una función en la que aparentemente falla puesto que:

$$\int\int_R\, \frac{x^2-y^2}{(x^2+y^2)^2} \,dx dy=-\frac{\pi}{4}$$ y

$$\int\int_R\, \frac{x^2-y^2}{(x^2+y^2)^2} \,dydx=\frac{\pi}{4},$$

donde $R=[0,1]\times[0,1]$. Explique por qué este ejemplo no contradice el teorema de Fubini.

iii. Sea $f$ continua, $f ≥ 0$, sobre el rectángulo $R$. Si $\int\int_R\, f\, dA = 0$, explique por qué se debe tener que $f = 0$ en $R$.


## Solución i
Se quiere demostrar que si $f$ es continua en $[a, b]$ y $g$ continua en $[c, d]$, se cumple que $$\int\int_R f(x)g(y)\, dx dy=\int_a^bf(x)\,dx\int_c^d g(y)\, dy,$$ donde $R=[a,b]\times[c,d]$.

Dado que $f(x)$ depende unicamente de la variable $x$ y $g(y)$ solo depende de la variable $y$. Esto significa que, al realizar la integración respecto a $x, g(y)$ puede considerarse como una constante y extraerse fuera de esa integral:

$$\int_{c}^{d} \left( \int_{a}^{b} f(x)g(y)\,dx \right) dy = \int_{c}^{d} g(y) \left( \int_{a}^{b} f(x)\,dx \right) dy$$

Ahora, como la integral $\int_{a}^{b}f(x)dx$ es un valor que no depende de $y$, por lo que puede extraerse fuera de la integral en $y$. Así obtenemos:

$$\left( \int_{a}^{b} f(x)\,dx \right) \left( \int_{c}^{d} g(y)\,dy \right)$$

De este modo, queda demostrado que la integral doble sobre la región $R$ puede descomponerse como el producto de dos integrales simples, una que involucra únicamente a $x$ y otra que involucra únicamente a $y$.




## Solución ii

El Teorema de Fubini establece que el orden de integración en una integral doble puede intercambiarse si la función es integrable en el dominio $R$. El Teorema de Fubini nos dice que si $f(x,y)$ es una función absolutamente integrable sobre un dominio rectangular $R=[a,b]\times [c,d]$ es decir:

$$\int\int_{R} |f(x,y)|\,dx\,dy < \infty$$

entonces se cumple que:
$$\
\iint_{R} f(x,y) \, dx \, dy
= \int_{a}^{b} \left[ \int_{c}^{d} f(x,y) \, dy \right] dx
= \int_{c}^{d} \left[ \int_{a}^{b} f(x,y) \, dx \right] dy
\$$
En este problema, la función dada es:

$$f(x,y) = \frac{x^2 - y^2}{(x^2 + y^2)^2}$$

Sin embargo, esta función presenta un problema en el punto $(0,0)$ dado que el denominador $(x^2 + y^2)^2$ tiende a cero cuando $x,y \to 0$ lo que provoca que la función crezca sin límite (se vuelva no acotada, es decir, se vuelve convergente, justamente la condición que exige el Teorema de Fubini para poder intercambiar el orden de integración sin alterar el resultado).
<br>
Esto significa que la integral debe interpretarse como una integral impropia, ya que en el origen existe una singularidad (punto en el dominio donde la función no está definida o tiende a infinito). En integrales impropias, el resultado puede depender del orden en el que se tome el límite, lo que explica que:

$$\int\int_R\, \frac{x^2-y^2}{(x^2+y^2)^2} \,dx dy=-\frac{\pi}{4}$$ y $$\int\int_R\, \frac{x^2-y^2}{(x^2+y^2)^2} \,dydx=\frac{\pi}{4},$$  

den resultados distintos.

El motivo es entonces que la función no es absolutamente integrable, es decir:

$$\int\int_{R} |f(x,y)|\,dx\,dy = \infty,$$

Por lo tanto, el Teorema de Fubini no se puede aplicar directamente a este caso. Esto explica por qué los valores de las integrales difieren, sin que esto contradiga el teorema.



## Solución iii
Sabiendo que $f$ es continua y $f\geq 0$ en el rectángulo $R$, y suponiendo que $\int \int_R f dA = 0$ <br>
Por teorema del valor medio para integrales dobles, existe un punto $(u_0, v_0) \in R$ tal que
$$ \int \int_R f dA = f(u_0, v_0) \cdot Area(R)$$
Es decir, la integral doble mide el valor promedio $f(u_0, v_0) \times $el área del rectángulo <br>
Como $Area>0$ se puede deducir que $f(u_0, v_0)=0$
Ahora bien, por contradicción suponemos que $f$ no es idénticamente nula, entonces existe un punto $(x_0, y_0) \in R$ con $f(u_0, v_0) > 0$. Por lo que, por continuidad existe un $m>0$ y un rectángulo $Q \subset R$ alrededor de $x_0, y_0$, para el cual
$$ f(x,y) \geq m $$ para todo $(x,y) \in Q$, es decir que la función sigue siendo positiva y al menos igual a un número fijo $m>0$<br>
Si en $Q$ la función es al menos $m$, entonces
$$ \int \int_Q f dA \geq \int \int_Q mdA $$
Pero $\int \int_Q mdA=m \cdot Area(Q)$, porque intergrar una constante en $Q$ es igual a multiplicar la constante por el área, de manear que
$$ \int \int_Q fdA \geq m \cdot Area(Q) $$
Y como tenemos que $Q \subset R$, la integral sobre todo $R$ es mayor o igual que la integral solo sobre $Q$
$$ \int \int_R fdA \geq \int \int_Q fdA $$
Ahora, el promedio de $f$ sobre $R$ es
$$ \frac{1}{Area(R)}\int \int_R fdA $$
Si dividimos la desigualdad anterior entre $Area(R)$, se obtiene
$$\frac{1}{Area(R)}\int \int_R fdA \geq m \cdot \frac{Area(Q)}{Area(R)}$$
Y como $m > 0$ y $Area(Q) > 0$, el número es positivo <br>
Entonces con lo anterior, el promedio de $f$ en $R$ satisface que
$$ \frac{1}{Area(R)} \int \int_R fdA \geq \frac{1}{Area(R)} \int \int_Q fdA \geq m \cdot \frac{Area(Q)}{Area(R)} > 0$$

Pero por teorema del valor medio ese promedio sería exactamente $f(u_0, v_0)$, es decir, ese promedio es positivo porque en una parte de $R$ la función es mayor que cero, así que obtendríamos que $f(u_0, v_0) > 0$ contradiciendo que $f(u_0, v_0) = 0$ que se habia deducido anteriormente a partir de que la integral es cero. Por lo que la contradicción muestra que no puede existir un $(x_0, y_0)$ con $f(x_0, y_0) > 0$, ya que en el mismo punto no se pueden tener dos valores distintos, haciendo que la suposición inicial de que existe un punto donde $f$ es positiva, debe de ser falsa y la única posibilidad es que $f$ sea cero en todo el rectángulo $R$ teniendo en cuenta la continuidad
