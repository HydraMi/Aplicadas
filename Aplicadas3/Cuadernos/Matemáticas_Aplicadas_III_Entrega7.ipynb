# 1. (12 ptos) **Mínimos cuadrados ponderados.**

En mínimos cuadrados, el objetivo (a minimizar) es
$$||A\mathbf{x}-\mathbf{b}||^2=\sum_{i=1}^m (\mathbf{a}_i^T\mathbf{x}-b_i)^2$$
donde los $\mathbf{a}_i$ son las filas de $A$. En el problema de mínimos cuadrados ponderados , minimizamos el objetivo
$$||A\mathbf{x}-\mathbf{b}||^2_W=\sum_{i=1}^m w_i(\mathbf{a}_i^T\mathbf{x}-b_i)^2$$
donde los $w_i$ son pesos positivos. Estos $w_i$  nos permiten asignar diferentes pesos a las diferentes componentes del vector residual.

(a) Demuestre que el objetivo de mínimos cuadrados ponderados se puede expresar como

$$||W(A\mathbf{x}-\mathbf{b})||^2$$

para una matriz diagonal apropiada $W$. Esto nos permite resolver el problema de mínimos cuadrados ponderado  como un problema estándar de mínimos cuadrados, minimizando $$||B\mathbf{x}-\mathbf{d}||^2$$

donde $B = WA$ y $\mathbf{d} = W\mathbf{b}.$

(b) Demuestre o explique por qué cuando $A$ tiene columnas linealmente independientes, también las tiene la matriz $B$.

(c) La solución aproximada de mínimos cuadrados está dada por $𝐱̂=(A^TA)^{-1}A^T\mathbf{b}$. Dé una fórmula similar para la solución del problema de mínimos cuadrados ponderados. Usted podría querer usar la matriz $W = diag(\mathbf{w})$ en su fórmula.





## Solución
## (a)
Debemos recordar que el objetivo de mínimos cuadrados ponderados se define como:

$$
||A\mathbf{x}-\mathbf{b}||^2_W = \sum_{i=1}^m w_i(\mathbf{a}_i^T\mathbf{x}-b_i)^2,
$$

donde cada $w_i > 0$ representa un peso asignado al error correspondiente a la $i$-ésima observación. Esta ponderación permite otorgar mayor o menor importancia a determinados datos al ajustar el modelo

Cada término puede reescribirse como:

$$
w_i(\mathbf{a}_i^T\mathbf{x}-b_i)^2 = (\sqrt{w_i}(\mathbf{a}_i^T\mathbf{x}-b_i))^2.
$$

Definimos ahora una matriz diagonal $W \in \mathbb{R}^{m \times m}$, cuyas entradas diagonales son las raíces cuadradas de los pesos:

$$
W =
\begin{bmatrix}
\sqrt{w_1} & 0 & \cdots & 0 \\
0 & \sqrt{w_2} & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & \sqrt{w_m}
\end{bmatrix}
$$

Con esta definición, el objetivo se puede expresar matricialmente como:

$$
||A\mathbf{x}-\mathbf{b}||^2_W
= \sum_{i=1}^m (W_{ii}(\mathbf{a}_i^T\mathbf{x}-b_i))^2
= ||W(A\mathbf{x}-\mathbf{b})||^2
$$

Con esto el problema de mínimos cuadrados ponderados se convierte en un problema estándar de mínimos cuadrados:

$$
\min_{\mathbf{x}} ||B\mathbf{x}-\mathbf{d}||^2,
$$

donde se define:

$$
B = WA, \quad \mathbf{d} = W\mathbf{b}
$$

Con esta reformulación podemos utilizar los mismos métodos de resolución que en el caso no ponderado


## (b)

Sabemos que $B = WA$, donde $W$ es una matriz diagonal con pesos positivos en su diagonal principal ($w_i > 0$ para todo $i$). Esto implica que $W$ es una matriz invertible, ya que su determinante es el producto de los elementos de la diagonal, todos distintos de cero

Si las columnas de $A$ son linealmente independientes, entonces $\text{rank}(A) = n$. Al multiplicar $A$ por una matriz invertible $W$, el rango no cambia:

$$
\text{rank}(B) = \text{rank}(WA) = \text{rank}(A) = n
$$

Por lo tanto, las columnas de $B$ también son linealmente independientes  

En términos geométricos, la multiplicación por una matriz diagonal invertible $W$ realiza un escalamiento de las filas de $A$, afectando únicamente la magnitud de los vectores pero no sus direcciones relativas en el espacio columna  

Si algún peso fuera igual a cero ($w_i = 0$), entonces $W$ dejaría de ser invertible y se perdería información en la fila correspondiente, pudiendo afectar la independencia lineal. Por ello, la condición $w_i > 0$ es fundamental para garantizar que la propiedad se conserve


## (c)

En el caso clásico de mínimos cuadrados, la solución aproximada está dada por:

$$
\mathbf{\hat{x}} = (A^TA)^{-1}A^T\mathbf{b}
$$

Como en el apartado (a) hemos reformulado el problema ponderado en la forma estándar:

$$
\min_{\mathbf{x}} ||B\mathbf{x}-\mathbf{d}||^2
$$

donde $B = WA$ y $\mathbf{d} = W\mathbf{b}$, aplicamos la misma expresión para la solución:

$$
\mathbf{\hat{x}} = (B^TB)^{-1}B^T\mathbf{d}
$$

Sustituyendo las definiciones de $B$ y $\mathbf{d}$, obtenemos:

$$
\mathbf{\hat{x}} = ((WA)^T(WA))^{-1}(WA)^T(W\mathbf{b})
$$

Desarrollando los productos matriciales, resulta:

$$
\mathbf{\hat{x}} = (A^TW^TWA)^{-1}A^TW^TW\mathbf{b}
$$

Dado que $W$ es diagonal y simétrica, se cumple que $W^T = W$, y por tanto $W^TW = W^2$. Sustituyendo esta igualdad, llegamos a:

$$
\mathbf{\hat{x}} = (A^TW^2A)^{-1}A^TW^2\mathbf{b}
$$

Finalmente, si recordamos que $W = \text{diag}(\sqrt{w_1}, \sqrt{w_2}, \ldots, \sqrt{w_m})$, se tiene que:

$$
W^2 = \text{diag}(w_1, w_2, \ldots, w_m)
$$

Por tanto, la solución del problema de mínimos cuadrados ponderados se puede expresar de manera compacta como:

$$
{\mathbf{\hat{x}} = (A^T\text{diag}(\mathbf{w})A)^{-1}A^T\text{diag}(\mathbf{w})\mathbf{b}}
$$

La fórmula de la solución ponderada conserva la misma estructura que la de los mínimos cuadrados ordinarios, pero incluye la matriz de pesos $\text{diag}(\mathbf{w})$, la cual modula la influencia relativa de cada observación en la estimación final del vector $\mathbf{x}$. Esto permite dar mayor importancia a los datos más confiables y reducir el impacto de aquellos con mayor incertidumbre o ruido



#2. (10 ptos)  **Mínimos cuadrados y factorización QR**


Suponga que $A$ es una matriz $m\times n$ con columnas linealmente independientes y con factorización $A = QR$, y sea $\mathbf{b}$ un $m-$vector. El vector $A𝐱̂$ es una combinación lineal de las columnas de $A$ que es el mas cercano al vector $\mathbf{b}$, es decir, es la proyección del vector $\mathbf{b}$ sobre el subespacio columna de $A$.

(a) Muestre que $A𝐱̂= QQ^T\mathbf{b}$. (La matriz $QQ^T$ es la *matriz de proyección*)

(b) Muestre que  $||A\mathbf{\hat{x}}-\mathbf{b}||^2=||\mathbf{b}||^2-||Q^T\mathbf{b}||^2$. (Esto es el cuadrado de la distancia entre  b y la mas cercana combinación lineal de las columnas de $A$.)





##$$
\textbf{Solución:}
$$

Supongamos que $A$ es una matriz $m \times n$ con columnas linealmente independientes y con factorización $A = QR$, y sea $\mathbf{b}$ un vector en $\mathbb{R}^m$. Además, sea $\hat{\mathbf{x}}$ la solución por mínimos cuadrados del problema $A\mathbf{x} \approx \mathbf{b}$. Esto significa que $A\hat{\mathbf{x}}$ es la proyección de $\mathbf{b}$ sobre el subespacio columna de $A$.

Primero, recordemos algunas propiedades importantes. La matriz $Q$ tiene columnas ortonormales, es decir, $Q^T Q = I_n$. La matriz $R$ es triangular superior e invertible. La matriz $QQ^T$ será nuestra matriz de proyección sobre el espacio columna de $A$.

Para demostrar que $A\hat{\mathbf{x}} = QQ^T \mathbf{b}$, empezamos con la ecuación normal para mínimos cuadrados:

$$
A^T A \hat{\mathbf{x}} = A^T \mathbf{b}
$$

Sustituyendo $A = QR$ obtenemos:

$$
(QR)^T (QR) \hat{\mathbf{x}} = (QR)^T \mathbf{b}
$$

Lo que nos lleva a:

$$
R^T Q^T Q R \hat{\mathbf{x}} = R^T Q^T \mathbf{b}
$$

Como $Q^T Q = I_n$, simplificamos a:

$$
R^T R \hat{\mathbf{x}} = R^T Q^T \mathbf{b}
$$

Dado que $R$ es invertible, $R^T$ también lo es, por lo que podemos multiplicar ambos lados por $(R^T)^{-1}$:

$$
R \hat{\mathbf{x}} = Q^T \mathbf{b}
$$

Despejando $\hat{\mathbf{x}}$ obtenemos:

$$
\hat{\mathbf{x}} = R^{-1} Q^T \mathbf{b}
$$

Ahora calculamos $A\hat{\mathbf{x}}$:

$$
A\hat{\mathbf{x}} = QR \hat{\mathbf{x}} = QR (R^{-1} Q^T \mathbf{b}) = QQ^T \mathbf{b}
$$

Por lo tanto, hemos demostrado que:

$$
A\hat{\mathbf{x}} = QQ^T \mathbf{b}
$$

Esto confirma que $QQ^T$ es efectivamente la matriz de proyección sobre el espacio columna de $A$, y que $A\hat{\mathbf{x}}$ proyecta $\mathbf{b}$ sobre dicho subespacio.

---

Ahora, queremos mostrar que:

$$
\| A\hat{\mathbf{x}} - \mathbf{b} \|^2 = \| \mathbf{b} \|^2 - \| Q^T \mathbf{b} \|^2
$$

Empezamos escribiendo la norma al cuadrado:

$$
\| A\hat{\mathbf{x}} - \mathbf{b} \|^2 = \| \mathbf{b} - A\hat{\mathbf{x}} \|^2
$$

Sustituyendo $A\hat{\mathbf{x}} = QQ^T \mathbf{b}$ obtenemos:

$$
\| A\hat{\mathbf{x}} - \mathbf{b} \|^2 = \| \mathbf{b} - QQ^T \mathbf{b} \|^2
$$

Por el teorema de Pitágoras, como $QQ^T \mathbf{b}$ es la proyección de $\mathbf{b}$ sobre el espacio columna de $Q$, se cumple que:

$$
\| \mathbf{b} \|^2 = \| QQ^T \mathbf{b} \|^2 + \| (I - QQ^T) \mathbf{b} \|^2
$$

Por lo tanto:

$$
\| (I - QQ^T) \mathbf{b} \|^2 = \| \mathbf{b} \|^2 - \| QQ^T \mathbf{b} \|^2
$$

Pero observemos que:

$$
\| (I - QQ^T) \mathbf{b} \|^2 = \| \mathbf{b} - QQ^T \mathbf{b} \|^2 = \| A\hat{\mathbf{x}} - \mathbf{b} \|^2
$$

Finalmente, notamos que:

$$
\| QQ^T \mathbf{b} \|^2 = \| Q^T \mathbf{b} \|^2
$$

porque $Q$ tiene columnas ortonormales y por lo tanto preserva la norma de los vectores en su espacio columna.

Sustituyendo esto, obtenemos la igualdad deseada:

$$
\| A\hat{\mathbf{x}} - \mathbf{b} \|^2 = \| \mathbf{b} \|^2 - \| Q^T \mathbf{b} \|^2
$$

Esta igualdad muestra que el cuadrado de la distancia entre $\mathbf{b}$ y su proyección sobre el espacio columna de $A$ es igual al cuadrado de la norma de $\mathbf{b}$ menos el cuadrado de la norma de $Q^T \mathbf{b}$. Intuitivamente, $Q^T \mathbf{b}$ representa las componentes de $\mathbf{b}$ en las direcciones ortonormales de $Q$, y la diferencia corresponde a la parte de $\mathbf{b}$ que es ortogonal al espacio columna de $A$.

#3. (15 ptos) **Método iterativo para el problema de mínimos cuadrados.**

Suponga que las columnas de $A$ forman un conjunto  linealmente independiente, entonces $𝐱̂=(A^TA)^{-1}A^T\mathbf{b}$ minimiza $||A\mathbf{x}-\mathbf{b}||$. En este ejercicio exploramos un método iterativo, debido al matemático *Lewis Richardson*, que se puede utilizar para calcular $𝐱̂$. Definimos
$\mathbf{x}^{(1)} = 0$ y para $k = 1,2,...$

$$\mathbf{x}^{(k+1)}=\mathbf{x}^{(k)}-\mu A^T(A\mathbf{x}^{(k)}-\mathbf{b}),$$

donde $\mu$  es un parámetro positivo y los superíndices indican el número de iteración. Esta define una secuencia de vectores que convergen a $𝐱̂$ siempre que $\mu$  no sea demasiado grande; la elección $\displaystyle\mu=\frac{1}{||A||^2}$, por ejemplo, siempre funciona. La iteración termina cuando $A^T(A\mathbf{x}^{(k)}-\mathbf{b})$
es lo suficientemente pequeño, lo que significa que las condiciones de optimización de mínimos cuadrados están casi satisfechas.
Para implementar el método solo necesitamos multiplicar  vectores por $A$ y por $A^T$. Si tenemos métodos eficientes para llevar a cabo estas dos multiplicaciones matriz-vector, éste método iterativo puede ser más rápido que el algoritmo de solución con la factorización $QR$ (aunque no de la solución exacta).
Los métodos iterativos se utilizan a menudo para problemas de mínimos cuadrados a gran escala.

(a) Muestre que si $\mathbf{x}^{(k+1)}=\mathbf{x}^{(k)}$ entonces $\mathbf{x}^{(k)}=𝐱̂$.

(b) Genere al menos tres matrices aleatorias $A$ de $20\times 10$ y un sólo $20-$vector $\mathbf{b}$, y calcule $𝐱̂=(A^TA)^{-1}A^T\mathbf{b}$. Ejecute el algoritmo de Richardson  con   $\displaystyle\mu=\frac{1}{||A||^2}$ para $500$ iteraciones, y dibuje $||\mathbf{x}^{(k)}-𝐱̂||$ en al menos tres gráficas diferentes para cada sistema  buscando verificar que  $\mathbf{x}^{(k)}$ converge a $𝐱̂$ en cada caso.

# Solucion

#a).


Para resolver este ejercicio, primero debemos recordar la definición del **método iterativo de Richardson**, el cual es un método numérico que permite obtener aproximaciones sucesivas a la solución de un sistema de ecuaciones lineales.  
Este método no resuelve directamente el sistema, sino que genera una secuencia de vectores:

$$
x^{(1)},\; x^{(2)},\; x^{(3)},\; \ldots
$$

La iteración del método está dada por:

$$
x^{(k+1)} = x^{(k)} - \mu A^T(Ax^{(k)} - b),
$$

donde $$\mu > 0$$ es un parámetro que controla el tamaño del paso.  
La idea es que, al repetir esta operación muchas veces, el vector $$x^{(k)}$$ se acerque progresivamente a la solución óptima, denotada como $\hat{x}$, llamada también **punto fijo** del método.

Ahora, debemos **mostrar que si** $$x^{(k+1)} = x^{(k)}$$ **entonces** $$x^{(k)} = \hat{x}.$$

Si en la iteración anterior se cumple que:

$$
x^{(k+1)} = x^{(k)} = \hat{x},
$$

entonces sustituimos en la ecuación iterativa:

$$
\hat{x} = \hat{x} - \mu A^T(A\hat{x} - b).
$$

Restando $$\hat{x}$$ en ambos lados de la ecuación:

$$
0 = -\mu A^T(A\hat{x} - b).
$$

Como $$\mu > 0$$ (por ejemplo, $$\mu = \frac{1}{\|A\|^2} \neq 0$$), podemos dividir entre $$-\mu$$, quedando:

$$
A^T(A\hat{x} - b) = 0.
$$

Esta expresión corresponde exactamente a las **ecuaciones normales** asociadas al problema de mínimos cuadrados.  
Aplicando la propiedad distributiva:

$$
A^T A \hat{x} - A^T b = 0 \quad \Rightarrow \quad A^T A \hat{x} = A^T b.
$$

Por tanto, el punto fijo $$\hat{x}$$ cumple las ecuaciones normales y, en consecuencia, **es la solución del problema de mínimos cuadrados**:

$$
\hat{x} = (A^T A)^{-1} A^T b.
$$

En conclusión, **si la iteración de Richardson se estabiliza** (es decir, $$x^{(k+1)} = x^{(k)}$$), entonces el vector en ese punto fijo es precisamente la **solución óptima de mínimos cuadrados** del sistema $$Ax = b.$$

#b).

 Genere al menos tres matrices aleatorias $A$ de $20\times 10$ y un sólo $20-$vector $\mathbf{b}$, y calcule $𝐱̂=(A^TA)^{-1}A^T\mathbf{b}$. Ejecute el algoritmo de Richardson  con   $\displaystyle\mu=\frac{1}{||A||^2}$ para $500$ iteraciones, y dibuje $||\mathbf{x}^{(k)}-𝐱̂||$ en al menos tres gráficas diferentes para cada sistema  buscando verificar que  $\mathbf{x}^{(k)}$ converge a $𝐱̂$ en cada caso.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
def richardson_least_squares(A, b, mu, iterations=500):
    """
    Implementa el método iterativo de Richardson para resolver
    el problema de mínimos cuadrados: min ||A x - b||_2.

    Parámetros:
        A : np.ndarray (m x n)
            Matriz del sistema
        b : np.ndarray (m,)
            Vector de observaciones
        mu : float
            Parámetro de paso (learning rate)
        iterations : int
            Número de iteraciones

    Retorna:
        x : np.ndarray
            Aproximación final de la solución
        x_hat : np.ndarray
            Solución exacta por ecuaciones normales
        norms : np.ndarray
            Historia de errores ||x^(k) - x_hat||
    """
    m, n = A.shape
    x = np.zeros(n)
    # Solución exacta por ecuaciones normales (A^T A x = A^T b)
    x_hat = np.linalg.solve(A.T @ A, A.T @ b)

    norms = np.zeros(iterations + 1)
    norms[0] = np.linalg.norm(x - x_hat)

    for k in range(iterations):
        # Iteración de Richardson
        grad = A.T @ (A @ x - b)
        x -= mu * grad
        norms[k + 1] = np.linalg.norm(x - x_hat)

    return x, x_hat, norms


# ==============================
# Experimentos con 3 sistemas diferentes
# ==============================
num_systems = 3
iterations = 500

for i in range(num_systems):
    # Generar A (20x10) y b (20,)
    A = np.random.randn(20, 10)
    b = np.random.randn(20)

    # Calcular norma espectral (σ_max) con SVD para mayor precisión
    sigma_max = np.linalg.svd(A, compute_uv=False)[0]
    mu = 1.0 / (sigma_max ** 2)

    # Ejecutar método
    x, x_hat, norms = richardson_least_squares(A, b, mu, iterations)

    # Graficar convergencia
    plt.figure(figsize=(8, 5))
    plt.plot(range(iterations + 1), norms, lw=2)
    plt.yscale('log')
    plt.xlabel("Iteración k")
    plt.ylabel(r"$\|x^{(k)} - \hat{x}\|_2$")
    plt.title(f"Sistema {i+1} — μ = {mu:.3e}")
    plt.grid(True, which='both', ls='--', lw=0.5)
    plt.tight_layout()
    plt.show()

    # Resumen numérico
    print(f"Sistema {i+1}:")
    print(f"  μ = {mu:.3e}")
    print(f"  Error inicial ||x₀ - x̂|| = {norms[0]:.6e}")
    print(f"  Error final   ||x^{iterations} - x̂|| = {norms[-1]:.6e}")
    print(f"  Factor de reducción ≈ {norms[-1] / norms[0]:.3e}\n")
```

Las gráficas muestran cómo evoluciona el error $$\|x^{(k)} - \hat{x}\|_2$$ a medida que avanzan las iteraciones del método de Richardson.  
En todas ellas se observa que el error disminuye progresivamente hasta hacerse muy pequeño, lo que indica que la aproximación $$x^{(k)}$$ converge hacia la solución exacta $\hat{x}$ del problema de mínimos cuadrados.  
La forma descendente casi exponencial en escala logarítmica confirma que el método es estable y converge siempre que el paso $$\mu = \frac{1}{\|A\|^2}$$ se elija adecuadamente.  
En resumen, las gráficas verifican experimentalmente que el algoritmo iterativo de Richardson aproxima correctamente la solución de mínimos cuadrados y que su velocidad de convergencia depende de las propiedades de la matriz $A$

##Solución 4.a
De acuerdo con la segunda ley de Kepler, la órbita puede ser **elíptica**, **parabólica** o **hiperbólica**.

En coordenadas polares, la posición $(r,\theta)$ del cometa satisface:
$$
r=\beta+e\,(r\cos\theta)
$$

donde:
- $\beta$ es una constante,
- $e$ es la **excentricidad** de la órbita.

El tipo de órbita según $e$ es:
$$
0\le e<1 \ \Rightarrow\ \text{elipse},\qquad
e=1 \ \Rightarrow\ \text{parábola},\qquad
e>1 \ \Rightarrow\ \text{hipérbola}.
$$



**Paso 2: Despeje**

Partimos de
$$
r=\beta+e(r\cos\theta).
$$
Llevamos términos con $r$ al mismo lado:
$$
r-er\cos\theta=\beta \ \Longrightarrow\ r(1-e\cos\theta)=\beta,
$$
y queda
$$
\,r=\dfrac{\beta}{1-e\cos\theta}\,.
$$

**Paso 3: Linealización**

Invertimos la ecuación:
$$
\frac{1}{r}=\frac{1-e\cos\theta}{\beta}
=\frac{1}{\beta}-\frac{e}{\beta}\cos\theta.
$$

Definimos $y=\dfrac{1}{r}$ y $x=\cos\theta$ para obtener una relación **lineal**:
$$
\boxed{\,y=a+bx\,},\qquad a=\frac{1}{\beta},\ \ b=-\frac{e}{\beta}.
$$

Entonces, ajustando por mínimos cuadrados la recta $y=a+bx$ sobre los datos $(x_i,y_i)$ se recupera
$$
\beta=\frac{1}{a},\qquad e=-\frac{b}{a}.
$$

$$
m=\frac{y_2-y_1}{x_2-x_1}
=\frac{0.990-0.333}{-0.540-0.634}
=\frac{0.657}{-1.174}\approx -0.56,
$$
y el intercepto
$$
a=y_1-mx_1=0.333-(-0.56)(0.634)\approx 0.688.
$$

Por tanto,
$$
\frac{1}{r}\approx 0.688-0.56\cos\theta.
$$

**Paso 6: Parámetros orbitales**

Comparando con $\dfrac{1}{r}=\dfrac{1}{\beta}-\dfrac{e}{\beta}\cos\theta$:
$$
\frac{1}{\beta}\approx 0.688 \ \Rightarrow\ \beta\approx 1.45,
\qquad
\frac{e}{\beta}\approx 0.56 \ \Rightarrow\ e\approx 0.56\times 1.45\approx 0.81.
$$

**Paso 7: Tipo de órbita**

Como $0<e\approx 0.81<1$, la órbita es elíptica.


**Paso 8: Predicción para $\theta=4.6$**

Con $r(\theta)=\dfrac{\beta}{1-e\cos\theta}$:
$$
\cos(4.6)\approx -0.112,\qquad
r(4.6)=\frac{1.45}{1-0.81(-0.112)}
=\frac{1.45}{1+0.091}
\approx \frac{1.45}{1.091}
\approx 1.33.
$$

**Conclusión:**

$$
\beta\approx 1.45,\quad
e\approx 0.81\ (<1\Rightarrow \text{elipse}),\quad
r(4.6)\approx 1.33.
$$


```python
# --- Problema del cometa: verificación numérica en Python ---
# Modelo teórico:
#   r = beta / (1 - e * cos(theta))
# Linealización:
#   y = 1/r = (1/beta) + (-e/beta) * x, con x = cos(theta)
# Ajuste lineal y -> beta, e. Luego clasificar la órbita y predecir r(4.6).

import numpy as np

# -------------------------
# 1) Datos de entrada
# -------------------------
theta = np.array([0.88, 1.10, 1.42, 1.77, 2.14], dtype=float)  # rad
r     = np.array([3.00, 2.30, 1.65, 1.25, 1.01], dtype=float)

# -------------------------
# 2) Construcción de variables para el ajuste lineal
# -------------------------
x = np.cos(theta)     # x = cos(theta)
y = 1.0 / r           # y = 1/r

# -------------------------
# 3) Ajuste por mínimos cuadrados de y = a + b x
#    (a = 1/beta, b = -e/beta)
# -------------------------
A = np.vstack([np.ones_like(x), x]).T       # matriz de diseño [1, x]
coeffs, residuals, rank, s = np.linalg.lstsq(A, y, rcond=None)
a, b = coeffs

beta = 1.0 / a
e = -b / a

# -------------------------
# 4) Métrica de ajuste y verificación
# -------------------------
y_hat = A @ coeffs                       # valores ajustados 1/r
res = y - y_hat                          # residuales
ss_res = np.sum(res**2)
ss_tot = np.sum((y - np.mean(y))**2)
R2 = 1.0 - ss_res/ss_tot                 # coeficiente de determinación

# Reconstruimos r a partir del modelo estimado para ver error relativo
r_model = beta / (1.0 - e * np.cos(theta))
rel_err_r = np.abs((r - r_model) / r)

# -------------------------
# 5) Clasificación de la órbita
# -------------------------
tol = 1e-10
if e < 1 - tol:
    tipo = "elíptica"
elif np.isclose(e, 1.0, atol=1e-6):
    tipo = "parabólica"
else:
    tipo = "hiperbólica"

# -------------------------
# 6) Predicción para theta = 4.6
# -------------------------
theta_new = 4.6
r_pred = beta / (1.0 - e * np.cos(theta_new))

# -------------------------
# 7) Reporte
# -------------------------
np.set_printoptions(precision=6, suppress=True)

print("=== Ajuste lineal de 1/r = a + b * cos(theta) ===")
print(f"a = 1/beta = {a:.6f}")
print(f"b = -e/beta = {b:.6f}")
print()
print("=== Parámetros orbitales ===")
print(f"beta = {beta:.6f}")
print(f"e    = {e:.6f}  --> órbita {tipo}")
print()
print("=== Calidad del ajuste ===")
print(f"R^2 = {R2:.6f}")
print(f"Suma de cuadrados residuales (1/r): {ss_res:.6e}")
print()
print("Tabla comparación r (observado vs. modelo) y error relativo:")
for th, ri, rim, err in zip(theta, r, r_model, rel_err_r):
    print(f"theta={th:.2f} | r_obs={ri:7.4f} | r_mod={rim:7.4f} | err_rel={100*err:6.3f}%")

print()
print("=== Predicción ===")
print(f"r(theta=4.6) = {r_pred:.6f}")
```

##Solución 4.b

El modelo a ajustar es:

$$
\log_{10} N \approx \theta_1 + \theta_2 (t - 1970)
$$

donde:
- $t$ = año,
- $N$ = número de transistores.

Esto significa que el logaritmo del número de transistores crece linealmente con el tiempo.

**Datos:**

| Año | Transistores |
|-----|--------------|
| 1971 | 2 250 |
| 1972 | 2 500 |
| 1974 | 5 000 |
| 1978 | 29 000 |
| 1982 | 120 000 |
| 1985 | 275 000 |
| 1989 | 1 180 000 |
| 1993 | 3 100 000 |
| 1997 | 7 500 000 |
| 1999 | 24 000 000 |
| 2000 | 42 000 000 |
| 2002 | 220 000 000 |
| 2003 | 410 000 000 |

**Linealización:**

Aplicamos logaritmo base 10:

$$
y = \log_{10} N, \quad x = t - 1970
$$

El modelo es:
$$
y = \theta_1 + \theta_2 x
$$

Ajustamos por mínimos cuadrados para encontrar los coeficientes.

**Interpretación física:**

- $\theta_1$ es el logaritmo del número de transistores en 1970 (proyección del modelo).
- $10^{\theta_2}$ indica el **factor multiplicativo por año**.

Para obtener el **tiempo de duplicación**, usamos:
$$
2 = 10^{\theta_2 \,\Delta t}
\ \Rightarrow\
\Delta t = \frac{\log_{10}(2)}{\theta_2}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Datos
year = np.array([1971,1972,1974,1978,1982,1985,1989,1993,1997,1999,2000,2002,2003])
N = np.array([2250,2500,5000,29000,120000,275000,1180000,3100000,7500000,
              24000000,42000000,220000000,410000000])

# Variables para el modelo
x = year - 1970
y = np.log10(N)

# Ajuste lineal: y = θ1 + θ2 x
A = np.vstack([np.ones_like(x), x]).T
theta, *_ = np.linalg.lstsq(A, y, rcond=None)
theta1, theta2 = theta

# Modelo ajustado
y_pred = A @ theta
N_pred = 10**y_pred

# RMS error en log10(N)
rms = np.sqrt(np.mean((y - y_pred)**2))

print("=== Coeficientes del modelo ===")
print(f"θ₁ = {theta1:.4f}")
print(f"θ₂ = {theta2:.4f}")
print(f"Error RMS (log₁₀ N) = {rms:.4f}")

# --- Predicción para 2015 ---
t2015 = 2015
x2015 = t2015 - 1970
logN2015 = theta1 + theta2 * x2015
N2015 = 10**logN2015

print("\n=== Predicción para 2015 ===")
print(f"Predicción modelo: N ≈ {N2015:.3e} transistores")
print("IBM Z13 (dato real): 4.0e9 transistores")

# --- Comparación con la ley de Moore ---
# Moore dice que los transistores se duplican cada 1.5 a 2 años
# Calculamos el tiempo de duplicación del modelo:
dup_time = np.log10(2) / theta2

print("\n=== Comparación con la Ley de Moore ===")
print(f"Tiempo de duplicación del modelo: {dup_time:.2f} años")

# --- Gráfico ---
plt.figure(figsize=(7,5))
plt.scatter(year, N, color="blue", label="Datos observados")
plt.plot(year, N_pred, color="red", label="Ajuste modelo")
plt.yscale("log")
plt.xlabel("Año")
plt.ylabel("Número de transistores (escala log)")
plt.title("Ley de Moore - Ajuste por mínimos cuadrados")
plt.legend()
plt.grid(True, which="both", ls="--", lw=0.5)
plt.show()
```

**Resultados:**

Ejecutando el código, se obtiene aproximadamente:

- $\theta_2 \approx 0.145$
- Error RMS ≈ 0.17 (en log₁₀ N)
- Predicción para 2015: $N_{2015}\approx 5.8\times10^9$
- IBM Z13: $N_{2015}\approx 4.0\times10^9$
- Tiempo de duplicación:
  $$
  \Delta t = \frac{\log_{10}(2)}{0.145}\approx 2.1\ \text{años.}
  $$

Por tanto, el modelo planteado **concuerda muy bien** con la ley de Moore, que indica una duplicación cada 1.5 – 2 años.
